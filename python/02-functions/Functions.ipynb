# **Functions** (Függvények)

- A függvények olyan programrészletek amelyek valamilyen céltudatos feladatot látnak el. 
- A függvény **csak akkor fut le, ha meghívjuk** azt. 
- Általában olyankor **használjuk, ha az adott programrészt többször is fel szeretnénk használni a
kódban**, ezért elegendő egyszer megírni a függvényt, **szükség esetén pedig csak meghívjuk** a kívánt helyen. 

- Egy függvényt a `def` kulcsszóval definiálunk. 
- A függvény **paramétere** egy vagy több változó amely(ek) **a függvény definiálásakor, a zárójeleiben** van(nak) felsorolva.
- A függvénynek lehetnek **argumentumai** amelyeket **a függvény meghívásakor, a zárójeleiben** adhatunk át neki. Így lehetővé teszik hogy bemeneti adatokat fogadjanak és ezekkel az értékekkel különböző műveleteket hajthassnak végre a fejlesztők. *Technikailag a kettő ugyanaz.*

- A `return`a függvény egy eredményét (visszatérési értékét) adja vissza. Ha nem használunk return utasítást, akkor alapértelmezés szerint **None** értéket ad vissza.
Akkor van szükségünk rá, ha ki kell számolni egy értéket és visszaadni egy változónak *(így pl. egy sima print() esetén nem szükséges)*

### Függvények elnevezése:

- Az elnevezés legyen kifejező és érthető, **ige vagy igei kifejezés** *(Clean Code)*
- **snake_case** stílus ajánlott
- A legtöbb elvárás amely a változókra igaz, a függvényekre is érvényes (ne legyen se foglalt kulcsszó se speciális karakter, számmal nem kezdődhet, case-sensitive stb.)
- Privát jellegű függvények _ *(aláhúzással)* kezdődjenek

Paraméter nélküli függvény:

```python
def to_welcome():
    print("Hello!")
    
to_welcome()
```

Paraméteres függvény:

```python
def to_welcome(name):
    print("Hello",name,end="!")
    
to_welcome("User")
```

Ugyanez a példa, return használatával:

```python
def to_welcome(name):
    return "Hello " + name + "!"

message = to_welcome("User")
print(message)
```

**1. Feladat:**

Írj egy függvényt amely paraméterként megkap egy számot, és eldönti erről a számról, hogy
tökéletes szám-e. 

Tökéletes számnak tekintjük azt a számot, amely az önmagánál kisebb 
osztóinak az összege. Pl. 6 egy tökéletes szám, mert 6 = 3 + 2 + 1

```python
enteredNumber = int(input("Enter a number: "))

def is_perfect(number):
  sum = 0
  for i in range(1, number):
    if number % i == 0:
      sum += i
  if sum == number:
    return True
  else:
    return False

if is_perfect(enteredNumber):
  print(f"{enteredNumber} is a perfect number!")
else:
  print(f"{enteredNumber} is not a perfect number!")
```
