# **OOP (Object Oriented Programming)**

Az objektumorientált programozás a kód szervezésének egy olyan módja, amely **objektumokat (object)** és **osztályokat (class)** használ a valós világbeli entitások és azok viselkedésének ábrázolására. Segíti, hogy moduláris, karbantartható és skálázható alkalmazásokat készíthessünk. 

- Az OOP-ban az **objektumok attribútumokkal** rendelkeznek – olyan dolgok, amelyek meghatározott adatokkal rendelkeznek, és metódusok segítségével bizonyos műveleteket hajthatnak végre.

    - A kódot osztályokba és objektumokba rendezi.
    - Támogatja az enkapszulációt az adatok és metódusok csoportosításához.
    - Lehetővé teszi az **öröklődést** az **újrafelhasználhatóság** és a hierarchia érdekében.
    - Lehetővé teszi a **polimorfizmust** a rugalmas módszermegvalósítás érdekében.


## Class - Osztály

Objektumok gyűjteménye. Összefogja az adatokat és a függvényeket, így könnyebben kezelhetők és használhatók. Amikor új osztályt hozunk létre, egy új típusú objektumot definiálunk. Ezután több példányt is létrehozhatunk ebből az objektumtípusból.

- `class` kulcsszó segítségével hozhatjuk lére
- Attribútumok olyan változók, amelyek az osztályhoz tartoznak.  `. operátorral` érhető el. pl. `car.brand`

## Object - Objektum

- Egy osztály egy **példánya**. Az osztály egy adott implementációját reprezentálja, és saját adatokat tárol. 

    -  **Pl. Tesla egy Car típúsú objektum**

- Egy objektum a következőkből áll:
    - **Állapot (state)** : az objektum attribútumai által képviselt állapot, amely egy objektum tulajdonságait tükrözi. `Pl. autó márkája, modelle, színe, sebessége, üzemanyagszintje`

    - **Viselkedés (behavior)**: egy objektum metódusai reprezentálják, és egy objektum más objektumokra adott válaszát tükrözi. `Pl. autó fékezés, gyorsítás, kanyarodás, beindítás / leállítás, tankolás`
    
    - **Azonosság (identity):** egyedi nevet ad egy objektumnak, és lehetővé teszi, hogy egy objektum más objektumokkal interakcióba lépjen. `Pl. autó alvázszáma, rendszáma`

Egy objektum létrehozása magában foglalja egy osztály példányosítását, létrehozzuk az adott osztály egy új példányát. Ezt a folyamatot **objektum példányosításnak** is nevezik.


```python
class Car:
    wheels = 4

    def __init__(self, brand, model):
        self.brand = brand  
        self.model = model  

    def __str__(self):
        return f"{self.brand} is a(n) {self.model} model."


car1 = Car("Audi", "A8")
car2 = Car("Tesla", "Model 3")
car3 = Car("Ford", "Mustang")

print(car1)
print(car2)
print(car3)

print(f"The {car2.brand} has {car2.wheels} wheels.")
```

Magyarázat:

- **Car osztály**, autó objektumok számára. 
- A kerék (wheels) egy **osztályattribútum**, ami azt jelenti, hogy az osztály minden példányára igaz.
- Az` __init__()` egy **konstruktor metódus**, amely automatikusan lefut egy új objektum létrehozásakor. Az objektumadatok **inicializálására** szolgál.

- `__str__() metódus` lehetővé teszi egy objektum egyéni karakterlánc definiálását


- A `self` az aktuális objektumra utal, lehetővé téve minden objektum számára a saját adatainak tárolását és elérését.
- A `self.brand` és a `self.model` példányattribútumok, amelyek az osztályból létrehozott minden Car objektumra egyediek. (Autónként változó, egyedi értékek)
