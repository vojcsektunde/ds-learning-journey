# **OOP (Object Oriented Programming)**

Az objektumorientált programozás a kód szervezésének egy olyan módja, amely **objektumokat (object)** és **osztályokat (class)** használ a valós világbeli entitások és azok viselkedésének ábrázolására. Segíti, hogy moduláris, karbantartható és skálázható alkalmazásokat készíthessünk. 

- Az OOP-ban az **objektumok attribútumokkal** rendelkeznek – olyan dolgok, amelyek meghatározott adatokkal rendelkeznek, és metódusok segítségével bizonyos műveleteket hajthatnak végre.

    - Lehetővé teszi a **polimorfizmust** a rugalmas módszermegvalósítás érdekében.


## Class - Osztály

Objektumok gyűjteménye. Összefogja az adatokat és a függvényeket, így könnyebben kezelhetők és használhatók. Amikor új osztályt hozunk létre, egy új típusú objektumot definiálunk. Ezután több példányt is létrehozhatunk ebből az objektumtípusból.

- `class` kulcsszó segítségével hozhatjuk lére
- Attribútumok olyan változók, amelyek az osztályhoz tartoznak.  `. operátorral` érhető el. pl. `car.brand`

## Object - Objektum

- Egy osztály egy **példánya**. Az osztály egy adott implementációját reprezentálja, és saját adatokat tárol. 

    -  **Pl. Tesla egy Car típúsú objektum**

- Egy objektum a következőkből áll:
    - **Állapot (state)** : az objektum attribútumai által képviselt állapot, amely egy objektum tulajdonságait tükrözi. `Pl. autó márkája, modelle, színe, sebessége, üzemanyagszintje`

    - **Viselkedés (behavior)**: egy objektum metódusai reprezentálják, és egy objektum más objektumokra adott válaszát tükrözi. `Pl. autó fékezés, gyorsítás, kanyarodás, beindítás / leállítás, tankolás`
    
    - **Azonosság (identity):** egyedi nevet ad egy objektumnak, és lehetővé teszi, hogy egy objektum más objektumokkal interakcióba lépjen. `Pl. autó alvázszáma, rendszáma`

Egy objektum létrehozása magában foglalja egy osztály példányosítását, létrehozzuk az adott osztály egy új példányát. Ezt a folyamatot **objektum példányosításnak** is nevezik.


```python
class Car:
    wheels = 4

    def __init__(self, brand, model):
        self.brand = brand  
        self.model = model  

    def __str__(self):
        return f"{self.brand} is a(n) {self.model} model."


car1 = Car("Audi", "A8")
car2 = Car("Tesla", "Model 3")
car3 = Car("Ford", "Mustang")

print(car1)
print(car2)
print(car3)

print(f"The {car2.brand} has {car2.wheels} wheels.")
```

Magyarázat:

- **Car osztály**, autó objektumok számára. 
- A kerék (wheels) egy **osztályattribútum**, ami azt jelenti, hogy az osztály minden példányára igaz.
- Az` __init__()` egy **konstruktor metódus**, amely automatikusan lefut egy új objektum létrehozásakor. Az objektumadatok **inicializálására** szolgál.

- `__str__() metódus` lehetővé teszi egy objektum egyéni karakterlánc definiálását


- A `self` az aktuális objektumra utal, lehetővé téve minden objektum számára a saját adatainak tárolását és elérését.
- A `self.brand` és a `self.model` példányattribútumok, amelyek az osztályból létrehozott minden Car objektumra egyediek. (Autónként változó, egyedi értékek)



`__new__() metódus`
Ez a metódus felelős egy osztály új példányának létrehozásáért. Memóriát foglal le és visszaadja az új objektumot. A metódus meghívása az __init__ metódus előtt történik.

- Megváltoztathatatlan (Immutable) beépített típusok származtatásakor
- (Singleton) tervezési minta megvalósításakor

### **1. Öröklés**

Az öröklődés lehetővé teszi egy osztály *(child, gyermekosztály)* számára, hogy egy másik osztály *(parent, szülőosztály)* tulajdonságait és metódusait átvegye. Támogatja a **hierarchikus osztályozást, azok modellezését** és elősegíti a kód újrafelhasználását.

```python
class Vehicle:
    def __init__(self, color):
        self.color = color

class Car(Vehicle):
    def __init__(self, color, doors):
        super().__init__(color)
        self.doors = doors

class Motorcycle(Vehicle):
    def __init__(self, color, helmet):
        super().__init__(color)
        self.helmet = helmet

my_car = Car("red", 4)
print("Car color:", my_car.color, "- Doors:", my_car.doors)

my_bike = Motorcycle("black", True)
print("Motorcycle color:", my_bike.color, "- Helmet:", my_bike.helmet)
```

Magyarázat:

- `A super() függvény` a Python metódusfeloldási sorrendjét (MRO) követő szuperosztály metódusainak meghívására szolgál. Különösen gyakran használják a gyermekosztály `__init__()` metódusában az örökölt attribútumok inicializálására. Így a **gyermekosztály kihasználhatja a szülőosztály funkcionalitását.**

### **2. Polimorfizmus**
A polimorfizmus jelentése **„ugyanaz a művelet, eltérő viselkedés”**. Lehetővé teszi, hogy az azonos nevű függvények vagy metódusok **eltérően működjenek** attól függően, hogy milyen típusú objektummal vagy adattal dolgozik.

*Pl. minden hangszer kiad valamilyen hangot, de más módszerrel: pengetéssel, ütéssel, billentyűleütéssel, fújással stb.* A hangkiadás mindegyikre igaz, csak a módszere tér el.


#### Polimorfizmus típusai:

1. **Fordítási idejű polimorfizmus**
    - A Pythonban nincs hagyományos, fordítási idejű (statikus) polimorfizmus (például metódustúlterhelés – method overloading – mint a C++-ban) de hasonló viselkedést ér el alapértelmezett argumentumokkal, `*args` és `**kwargs` paraméterekkel, mivel a metódusok feloldása futásidőben történik.

    - Lehetővé teszi több, azonos nevű, de eltérő paraméterekkel rendelkező metódus használatát, a megfelelő metódus végrehajtás előtti kiválasztásával. 




2. **Futásidejű polimorfizmus** (felülírás)
    - Egy metódus viselkedését a program futása közben döntik el, a hívó objektum alapján. Ez a metódusok felülbírálásával történik. Egy gyermekosztály metódusának felülbírálása a szülőosztályban már definiált metódus saját verzióját biztosítja.

```python
class Employee:
    def work(self):
        return "Doing general office work"

class Manager(Employee):
        def work(self):
            return "Managing the team and planning meetings"

class Developer(Employee):
    def work(self):
        return "Writing and debugging code"

employees = [Manager(), Developer(), Employee()]
for emp in employees:
    print(emp.work())
```

> `A work()` metódus minden alosztályban *(Manager, Developer)* más-más üzenetet ad vissza. 

>  Python futás közben **(runtime)** dönti el, hogy az adott objektumhoz tartozó melyik work() metódust kell meghívni, attól függően, hogy az employees lista aktuális eleme pontosan melyik osztályból jött létre.

#### **Beépített függvények**
A Python beépített függvényei, mint például a `len()` és a `max()` polimorfikusak, mivel különböző adattípusokkal dolgoznak, és az átadott objektum típusa alapján adnak vissza eredményeket. 

Ez jól mutatja dinamikus természetét, ahol ugyanaz a függvénynév a bemenettől függően módosítja a viselkedését.

#### **Polimorfizmus függvényekben**
A polimorfizmus lehetővé teszi a függvények számára, hogy különböző objektumtípusokkal működjenek, amennyiben azok támogatják a kívánt viselkedést. 

A kacsa típuskezelés **(duck tipization)** használata arra összpontosít, hogy egy objektum rendelkezik-e a szükséges metódusokkal, nem pedig a típusára, így **rugalmas és újrafelhasználható kódot tesz lehetővé**.



```python
class PDFExporter:
    def save(self, data):
        return f"Saving data in PDF format: '{data}'"

class CSVExporter:
    def save(self, data):
        return f"Saving data in CSV format: '{data}'"

def export_data(exporter, data):
    print(exporter.save(data))

export_data(PDFExporter(), "Report") 
export_data(CSVExporter(), "Report") 
```

> - 1. **Nincs közös ős:** A `PDFExporter` és a `CSVExporter` két teljesen **független osztály.** Nem örökölnek semmilyen közös Exporter nevű osztályból. 
> - 2. **A közös interfész (a duck tipization):** Mindkét osztálynak van egy pontosan ugyanúgy kinéző metódusa: save(self, data). Ez a közös viselkedésük.
> - 3. **A export_data függvény:** Amikor meghívod az `export_data(exporter, data)` függvényt, **nem ellenőrzi**, hogy az átadott exporter objektum egy PDFExporter-e vagy egy CSVExporter-e.  A függvénynek **csak egyetlen elvárása van:** a kapott objektumnak legyen egy `.save()` nevű metódusa, amit meg lehet hívni egy paraméterrel.

#### Miért van szükségünk Enkapszulációra?

#### **Hozzáférési specifikátorok**
A **hozzáférési specifikátorok** meghatározzák, hogy az osztálytagok (változók és metódusok) hogyan érhetők el az osztályon kívülről. Segítenek a beágyazás megvalósításában az adatok láthatóságának szabályozásával. 

##### Háromféle hozzáférési specifikátor létezik:
- **Nyilvános (Public)** - osztályon belülről, kívülről vagy más modulokból bárhonnan elérhetők. **Aláhúzás nélküli**

- **Védett (Protected)** - csak az osztályon és alosztályain belül lehet hozzáférni. **Egyetlen aláhúzásjel előtaggal** definiáljuk

- **Magán (Private)**  - nem lehet közvetlenül hozzáférni az osztályon kívülről. Ezeket a hozzáférés korlátozására és a belső adatok védelmére használjuk. **Dupla aláhúzásjel előtaggal** definiáljuk

```python
class CreditCard:
    def __init__(self, card_holder, card_number, pin_code):
        self.card_holder = card_holder      # (Public): Bárhonnan elérhető
        self._card_type = "Visa"            # Védtt (Protected): Belső használatra / alosztályoknak
        self.__card_number = card_number    # (Private): Rejtett adat 
        self.__pin_code = pin_code          # (Private): Rejtett adat
        self.__balance = 0.0                # (Private): Csak ellenőrzötten módosítható

    #  Getter: Biztonságos hozzáférést biztosít az egyenleghez
    def get_balance(self, entered_pin):
        if entered_pin == self.__pin_code:
            return self.__balance
        else:
            print("ERROR: Invalid PIN code! Access denied.")
            return None

    # Setter: Validációval ellátott befizetés
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Success: Deposited +{amount}")
            self.__log_transaction(f"Deposited {amount}") # Belső privát metódus hívása
        else:
            print("ERROR: Deposit amount must be positive!")

    # Magán metódus: Csak az osztályon belülről hívható meg
    def __log_transaction(self, message):
        print(f"[LOG]: {message}")


my_card = CreditCard("Alice Anderson", "1234-5678-9876-5432", "4321")

print(f"Card Holder: {my_card.card_holder}") # Nyilvános, működik
print(f"Card Type: {my_card._card_type}")     # Védett, működik, de jelezzük vele a belső státuszt

# print(my_card.__balance) # <- AttributeError-t eredményezne, mert el van rejtve!

print("\n--- Transactions ---")
#  Pénz befizetése a nyilvános metóduson keresztül (automatikusan futtatja a privát logolást is)
my_card.deposit(500)
my_card.deposit(-50) # Hibás, negatív érték tesztelése

print("\n--- Balance Check ---") #  Egyenleg lekérdezése hibás, majd helyes PIN kóddal
my_card.get_balance("0000") 
current_balance = my_card.get_balance("4321") 
print(f"Final Balance: {current_balance}")
```

### **4. Adatabsztrakció**

Az absztrakció elrejti a belső megvalósítási részleteket, miközben csak a szükséges funkciókat teszi láthatóvá. 

Segít abban, hogy a **„mit kell csinálni”** kérdésre koncentráljunk a **„hogyan” helyett**.

pl. **Amikor vezetsz egy autót**, az autó egy absztrakt felületet biztosít számodra (kormány, gázpedál, fékpedál), és nem a hogyan részen van a hangsúly (pl. hogyan számolja ki a pontos üzemanyag-befecskendezést, hogyan nyitnak a szelepek)


#### Absztrakt alaposztályok (ABC)

Az **absztrakt alaposztály (Abstract Base Class - ABC)** az adatabsztrakció elérésének hivatalos eszköze Pythonban. Segítségével egy közös interfészt (tervrajzot) határozhatunk meg az alosztályok számára.

##### Főbb jellemzői:
* **Nem példányosítható közvetlenül:** Nem hozható létre belőle objektum.
* **Kikényszeríti a megvalósítást:** Ha egy osztály örököl egy absztrakt osztályból, kötelező megvalósítania annak absztrakt metódusait, különben hibát kapunk.
* **Modul:** A Python beépített `abc` modulját és az `@abstractmethod` dekorátort használja.


```python
from abc import ABC, abstractmethod

# Absztrakt alaposztály
class Greeting(ABC):
    @abstractmethod
    def say_hello(self):
        pass   # Absztrakt metódus: nincs benne megvalósított kód

# Alosztály, amely kötelezően implementálja a fenti metódust
class English(Greeting):
    def say_hello(self):
        return "Hello!"

english_visitor = English()
print(english_visitor.say_hello())
```

#### Az absztrakció fő összetevői

Az absztrakt alaposztályok négy fontos szerkezeti elemből épülnek fel, amelyek biztosítják a konzisztens osztálystruktúrát.

-  1. Absztrakt metódus (Abstract Method)
    - Olyan metódusdeklaráció az absztrakt osztályon belül, amelynek **nincs törzse** (csak egy `pass` áll belőle). Helyőrzőként működik, és kötelezi a származtatott osztályokat a saját verziójuk kidolgozására.


```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def make_sound(self):
        pass   # Absztrakt metódus, itt nincs belső logika
```

- 2. Konkrét metódus (Concrete Method)
    - Az absztrakt osztályon belül **teljesen implementált (megírt) metódus**. Az alosztályok ezt automatikusan megöröklik, és közvetlenül használhatják anélkül, hogy újra meg kellene írniuk. Ez segíti a kód újrafelhasználását.


```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def make_sound(self):
        pass

    # Konkrét metódus: van kész törzse és működése
    def move(self):
        return "Moving forward..."

class Dog(Animal):
    def make_sound(self):
        return "Bark!"

my_dog = Dog()
print(my_dog.move())  # Közvetlenül a szülőosztály konkrét metódusát hívja meg
```

- 3. Absztrakt tulajdonságok (Abstract Properties)
    - Úgy működnek, mint az absztrakt metódusok, de a Python `@property` dekorátorával kötelezően megvalósítandó attribútum-lekérdezéseket (gettereket) írnak elő az alosztályok számára.


```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @property
    @abstractmethod
    def species(self):
        pass   # Absztrakt tulajdonság, az alosztályban kötelező a @property használata

class Dog(Animal):
    @property
    def species(self):
        return "Canine"

# Konkrét alosztály példányosítása
my_dog = Dog()
print(my_dog.species)  # Tulajdonságként (zárójelek nélkül) érhető el
```

- 4. Absztrakt osztály példányosítása (Hibaüzenet)
    - Ha közvetlenül próbálunk meg létrehozni egy objektumot egy olyan absztrakt osztályból, amelynek befejezetlen (absztrakt) metódusai vannak, a Python **`TypeError`** hibát dob.


```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def make_sound(self):
        pass

# Ez a sor szándékosan hibát fog okozni, mert az Animal absztrakt!
try:
    invalid_animal = Animal()
except TypeError as error:
    print(f"TypeError caught successfully: {error}")
```

### **Iterators (Iterátorok)**

A Pythonban az **iterátor** egy olyan objektum, amely egy gyűjtemény (például listák, tuple-ök vagy szótárak) elemein képes egyenként végighaladni. Az iterátorok az úgynevezett **iterátor protokollt** követik, amely két fő metódusból áll:
* `__iter__()`: Visszaadja magát az iterátor objektumot.
* `__next__()`: Visszaadja a sorozat következő értékét. Ha a sorozat végére ér, egy `StopIteration` kivételt vált ki.

#### Miért hasznosak az iterátorok?
* **Lusta kiértékelés (Lazy evaluation):** Csak akkor dolgozza fel az elemeket, amikor ténylegesen szükség van rájuk, ezzel jelentős memóriát takarít meg.
* **Állapotalapú bejárás:** Pontosan megjegyzi és nyomon követi, hol hagyta abba a bejárást.
* **Egységes kezelés:** Ugyanazzal a logikával járhatók be a listák, karakterláncok, fájlok vagy egyéni adatfolyamok.

**1. Beépített iterátorok használata**

A Python automatikusan biztosít iterátorokat a beépített kollekciókhoz (szövegek, listák stb.). Az `iter()` függvénnyel kérhetünk le egy iterátort, a `next()` függvénnyel pedig egyesével lépkedhetünk előre.


```python
text = "Python"
text_iterator = iter(text)

for char in text:
    print(next(text_iterator))
```

**2. Egyéni iterátor létrehozása**

Egyéni osztályok esetén saját magunknak kell megírnunk az `__iter__()` és `__next__()` metódusokat. Az alábbi példában egy olyan `EvenNumbers` osztályt hozunk létre, amely a páros számokat adja vissza egy megadott határértékig.


```python
class EvenNumbers:
    def __init__(self, limit):
        self.limit = limit  # A felső határérték
        self.current = 2    # Az első páros szám, ahonnan indulunk

    def __iter__(self):
        return self         # Az iterátor önmagát adja vissza

    def __next__(self):
        # Ha elértük a határt, leállítjuk az iterációt
        if self.current > self.limit:
            raise StopIteration

        # Elmentjük az aktuális értéket, növeljük a lépzközzel, majd visszaadjuk
        value_to_return = self.current
        self.current += 2
        return value_to_return

# Iterátor példányosítása 10-es határértékkel
even_sequence = EvenNumbers(10)

# automatikusan hívja a next()-et
for number in even_sequence:
    print(number)
```

**3. A `StopIteration` kivétel kézi kezelése**

Amikor az iterátor kiürül, a `StopIteration` hiba jelzi a ciklusoknak, hogy nincs több elem. Ezt a folyamatot egy `while` ciklusban manuálisan is elkaphatjuk egy `try-except` blokk segítségével.


```python
numbers_list = [100, 200, 300]
list_iterator = iter(numbers_list)

# Iteráció kézi vezérlése a StopIteration elkapásáig
while True:
    try:
        item = next(list_iterator)
        print(f"Item: {item}")
    except StopIteration:
        print("End of Iterator reached.")
        break
```

#### 4.  Iterable vs. Iterator

Bár a két fogalom hasonlónak tűnik, fontos különbség van köztük:
1. **Iterálható (Iterable):** Bármilyen objektum, amiből iterátort lehet készíteni (megvalósítja az `__iter__()` metódust). Önmagában nincs belső állapota arra nézve, hogy hol tart a bejárásban.
2. **Iterátor (Iterator):** Az az aktuális objektum, ami a bejárást végzi, ismeri a jelenlegi pozícióját, és megvalósítja a `__next__()` metódust is.

| Jellemző | Iterálható (Iterable) | Iterátor (Iterator) |
| :--- | :--- | :--- |
| **Meghatározás** | Olyan objektum, amely képes visszaadni egy iterátort. | Olyan objektum, amely ténylegesen végzi a léptetést. |
| **Szükséges metódusok** | Csak az `__iter__()` metódus. | Mind az `__iter__()`, mind a `__next__()` metódusok. |
| **Példák** | `list`, `tuple`, `str`, `dict`, `set` | Az `iter()` függvény által visszaadott egyedi objektumok. |
| **Hívható rá közvetlenül a `next()`?** | **Nem** | **Igen** |


```python
numbers_iterable = [1, 2, 3]
numbers_iterator = iter(numbers_iterable)

# iterátornál működik a next(), az iterálhatónál hibát adna
print(f"Using next() on iterator: {next(numbers_iterator)}")

# Az iterátor hívható a next-tel, de a lista nem:
print(f"Is numbers_iterator an iterator? {'__next__' in dir(numbers_iterator)}")
print(f"Is numbers_iterable an iterator? {'__next__' in dir(numbers_iterable)}")
```
