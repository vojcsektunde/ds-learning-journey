# **OOP (Object Oriented Programming)**

Az objektumorientált programozás a kód szervezésének egy olyan módja, amely **objektumokat (object)** és **osztályokat (class)** használ a valós világbeli entitások és azok viselkedésének ábrázolására. Segíti, hogy moduláris, karbantartható és skálázható alkalmazásokat készíthessünk. 

- Az OOP-ban az **objektumok attribútumokkal** rendelkeznek – olyan dolgok, amelyek meghatározott adatokkal rendelkeznek, és metódusok segítségével bizonyos műveleteket hajthatnak végre.

    - A kódot osztályokba és objektumokba rendezi.
    - Támogatja az enkapszulációt az adatok és metódusok csoportosításához.
    - Lehetővé teszi az **öröklődést** az **újrafelhasználhatóság** és a hierarchia érdekében.
    - Lehetővé teszi a **polimorfizmust** a rugalmas módszermegvalósítás érdekében.


## Class - Osztály

Objektumok gyűjteménye. Összefogja az adatokat és a függvényeket, így könnyebben kezelhetők és használhatók. Amikor új osztályt hozunk létre, egy új típusú objektumot definiálunk. Ezután több példányt is létrehozhatunk ebből az objektumtípusból.

- `class` kulcsszó segítségével hozhatjuk lére
- Attribútumok olyan változók, amelyek az osztályhoz tartoznak.  `. operátorral` érhető el. pl. `car.brand`

## Object - Objektum

- Egy osztály egy **példánya**. Az osztály egy adott implementációját reprezentálja, és saját adatokat tárol. 

    -  **Pl. Tesla egy Car típúsú objektum**

- Egy objektum a következőkből áll:
    - **Állapot (state)** : az objektum attribútumai által képviselt állapot, amely egy objektum tulajdonságait tükrözi. `Pl. autó márkája, modelle, színe, sebessége, üzemanyagszintje`

    - **Viselkedés (behavior)**: egy objektum metódusai reprezentálják, és egy objektum más objektumokra adott válaszát tükrözi. `Pl. autó fékezés, gyorsítás, kanyarodás, beindítás / leállítás, tankolás`
    
    - **Azonosság (identity):** egyedi nevet ad egy objektumnak, és lehetővé teszi, hogy egy objektum más objektumokkal interakcióba lépjen. `Pl. autó alvázszáma, rendszáma`

Egy objektum létrehozása magában foglalja egy osztály példányosítását, létrehozzuk az adott osztály egy új példányát. Ezt a folyamatot **objektum példányosításnak** is nevezik.


```python
class Car:
    wheels = 4

    def __init__(self, brand, model):
        self.brand = brand  
        self.model = model  

    def __str__(self):
        return f"{self.brand} is a(n) {self.model} model."


car1 = Car("Audi", "A8")
car2 = Car("Tesla", "Model 3")
car3 = Car("Ford", "Mustang")

print(car1)
print(car2)
print(car3)

print(f"The {car2.brand} has {car2.wheels} wheels.")
```

Magyarázat:

- **Car osztály**, autó objektumok számára. 
- A kerék (wheels) egy **osztályattribútum**, ami azt jelenti, hogy az osztály minden példányára igaz.
- Az` __init__()` egy **konstruktor metódus**, amely automatikusan lefut egy új objektum létrehozásakor. Az objektumadatok **inicializálására** szolgál.

- `__str__() metódus` lehetővé teszi egy objektum egyéni karakterlánc definiálását


- A `self` az aktuális objektumra utal, lehetővé téve minden objektum számára a saját adatainak tárolását és elérését.
- A `self.brand` és a `self.model` példányattribútumok, amelyek az osztályból létrehozott minden Car objektumra egyediek. (Autónként változó, egyedi értékek)



`__new__() metódus`
Ez a metódus felelős egy osztály új példányának létrehozásáért. Memóriát foglal le és visszaadja az új objektumot. A metódus meghívása az __init__ metódus előtt történik.

- Megváltoztathatatlan (Immutable) beépített típusok származtatásakor
- (Singleton) tervezési minta megvalósításakor


## **OOP 4 pillére**

### **1. Öröklés**

Az öröklődés lehetővé teszi egy osztály *(child, gyermekosztály)* számára, hogy egy másik osztály *(parent, szülőosztály)* tulajdonságait és metódusait átvegye. Támogatja a **hierarchikus osztályozást, azok modellezését** és elősegíti a kód újrafelhasználását.

```python
class Vehicle:
    def __init__(self, color):
        self.color = color

class Car(Vehicle):
    def __init__(self, color, doors):
        super().__init__(color)
        self.doors = doors

class Motorcycle(Vehicle):
    def __init__(self, color, helmet):
        super().__init__(color)
        self.helmet = helmet

my_car = Car("red", 4)
print("Car color:", my_car.color, "- Doors:", my_car.doors)

my_bike = Motorcycle("black", True)
print("Motorcycle color:", my_bike.color, "- Helmet:", my_bike.helmet)
```

Magyarázat:

- `A super() függvény` a Python metódusfeloldási sorrendjét (MRO) követő szuperosztály metódusainak meghívására szolgál. Különösen gyakran használják a gyermekosztály `__init__()` metódusában az örökölt attribútumok inicializálására. Így a **gyermekosztály kihasználhatja a szülőosztály funkcionalitását.**

### **2. Polimorfizmus**
A polimorfizmus jelentése **„ugyanaz a művelet, eltérő viselkedés”**. Lehetővé teszi, hogy az azonos nevű függvények vagy metódusok **eltérően működjenek** attól függően, hogy milyen típusú objektummal vagy adattal dolgozik.

*Pl. minden hangszer kiad valamilyen hangot, de más módszerrel: pengetéssel, ütéssel, billentyűleütéssel, fújással stb.* A hangkiadás mindegyikre igaz, csak a módszere tér el.


#### Polimorfizmus típusai:

1. **Fordítási idejű polimorfizmus**
    - A Pythonban nincs hagyományos, fordítási idejű (statikus) polimorfizmus (például metódustúlterhelés – method overloading – mint a C++-ban) de hasonló viselkedést ér el alapértelmezett argumentumokkal, `*args` és `**kwargs` paraméterekkel, mivel a metódusok feloldása futásidőben történik.

    - Lehetővé teszi több, azonos nevű, de eltérő paraméterekkel rendelkező metódus használatát, a megfelelő metódus végrehajtás előtti kiválasztásával. 




2. **Futásidejű polimorfizmus** (felülírás)
    - Egy metódus viselkedését a program futása közben döntik el, a hívó objektum alapján. Ez a metódusok felülbírálásával történik. Egy gyermekosztály metódusának felülbírálása a szülőosztályban már definiált metódus saját verzióját biztosítja.

```python
class Employee:
    def work(self):
        return "Doing general office work"

class Manager(Employee):
        def work(self):
            return "Managing the team and planning meetings"

class Developer(Employee):
    def work(self):
        return "Writing and debugging code"

employees = [Manager(), Developer(), Employee()]
for emp in employees:
    print(emp.work())
```

> `A work()` metódus minden alosztályban *(Manager, Developer)* más-más üzenetet ad vissza. 

>  Python futás közben **(runtime)** dönti el, hogy az adott objektumhoz tartozó melyik work() metódust kell meghívni, attól függően, hogy az employees lista aktuális eleme pontosan melyik osztályból jött létre.

#### **Beépített függvények**
A Python beépített függvényei, mint például a `len()` és a `max()` polimorfikusak, mivel különböző adattípusokkal dolgoznak, és az átadott objektum típusa alapján adnak vissza eredményeket. 

Ez jól mutatja dinamikus természetét, ahol ugyanaz a függvénynév a bemenettől függően módosítja a viselkedését.

```python
print(len("Polymorphism")) 
print(len([4,6,10,35]))  

print(max(30,44,2, 71,64,38,21,10)) 
print(max("y", "z", "x"))  
```

#### **Polimorfizmus operátoroknál**
Ugyanaz az `operátor +` különböző feladatokat hajthat végre az operandus típusától függően. Ezt **operátor-túlterhelésnek**nevezik. Ez a rugalmasság a polimorfizmus egyik kulcsfontosságú aspektusa.

```python
print(10 + 10) 
print("Hello " + "World!")  
print([2, 4] + [6, 8])  
```

#### **Polimorfizmus függvényekben**
A polimorfizmus lehetővé teszi a függvények számára, hogy különböző objektumtípusokkal működjenek, amennyiben azok támogatják a kívánt viselkedést. 

A kacsa típuskezelés **(duck tipization)** használata arra összpontosít, hogy egy objektum rendelkezik-e a szükséges metódusokkal, nem pedig a típusára, így **rugalmas és újrafelhasználható kódot tesz lehetővé**.



```python
class PDFExporter:
    def save(self, data):
        return f"Saving data in PDF format: '{data}'"

class CSVExporter:
    def save(self, data):
        return f"Saving data in CSV format: '{data}'"

def export_data(exporter, data):
    print(exporter.save(data))

export_data(PDFExporter(), "Report") 
export_data(CSVExporter(), "Report") 
```

> - 1. **Nincs közös ős:** A `PDFExporter` és a `CSVExporter` két teljesen **független osztály.** Nem örökölnek semmilyen közös Exporter nevű osztályból. 
> - 2. **A közös interfész (a duck tipization):** Mindkét osztálynak van egy pontosan ugyanúgy kinéző metódusa: save(self, data). Ez a közös viselkedésük.
> - 3. **A export_data függvény:** Amikor meghívod az `export_data(exporter, data)` függvényt, **nem ellenőrzi**, hogy az átadott exporter objektum egy PDFExporter-e vagy egy CSVExporter-e.  A függvénynek **csak egyetlen elvárása van:** a kapott objektumnak legyen egy `.save()` nevű metódusa, amit meg lehet hívni egy paraméterrel.
