# Adatvizualizáció Matplotlib használatával

A **Matplotlib** egy Python könyvtár statikus, animált és interaktív adatvizualizációk készítéséhez. A NumPy-ra épül, és könnyen kezeli a nagy adathalmazokat különféle diagramok létrehozásához. 

A **Pyplot** egy olyan modul, amely egyszerű felületet biztosít a diagramok minimális kóddal történő generálásához.

## 1. Alapvető diagramtípusok

### Vonaldiagram (Line Chart)

A vonaldiagram két adatsor (X és Y) közötti kapcsolatot ábrázol. `plot()` függvénnyel hozható létre

Paraméterek: 

- **Szín (color)**: A vonal színének módosítása
- **Vonalszélesség (linewidth)** : Állítsa be a vonal szélességét
- **Jelölő (marker)** : A kirajzolt pontok stílusának módosítása
- **Jelölőméret (markersize)** : A jelölők méretének módosítása
- **Vonalstílus (linestyle)** : Adja meg a vonal stílusát, például folytonos, szaggatott stb.

```python
import matplotlib.pyplot as plt

x = [10, 20, 30, 40]
y = [20, 25, 35, 55]

plt.plot(x, y, color='blue', linewidth=3, marker='D', markersize=15, linestyle='--')
plt.title("Line Chart")
plt.ylabel('Y-Axis')
plt.xlabel('X-Axis')
plt.show()
```

### Oszlopdiagram (Bar Chart)
Kategorikus adatokat jelenít meg téglalap alakú oszlopokkal, amelyek hossza arányos az értékekkel. Függőlegesen vagy vízszintesen ábrázolható a különböző **kategóriák összehasonlítása érdekében**.

 `pyplot.bar(x, height)` 
 
- `x`: kategóriák vagy pozíciók az x tengelyen 
- `height`: az oszlopok magassága (y tengely értékei)
- `color` : Sávkitöltés színe
- `edgecolor` : Sáv élének színe
- `linewidth` : Az élek vastagsága

```python
import matplotlib.pyplot as plt

x = ['Thur', 'Fri', 'Sat', 'Sun']
y = [170, 120, 250, 190]

plt.bar(x, y, color='green', edgecolor='black', linewidth=2)
plt.title("Bar Chart")
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.show()
```

### Hisztogram (Histogram)

A hisztogram az **adatok eloszlását mutatja** az értékek tárolókba csoportosításával. A hisztogram létrehozásához a `hist()` függvényt használjuk, ahol az X tengely a tárolókat, az Y tengely pedig a gyakoriságokat mutatja.

`matplotlib.pyplot.hist(x, bins)`

- `x`: Bemeneti adatok.
- `bins`: Az adatok felosztásához szükséges csoportok (rekeszek) száma
- `color` : Sávkitöltés színe
- `edgecolor` : Sáv élének színe
- `linestyle` : Az élek stílusa, például folytonos, szaggatott stb.
- `alpha` : Átlátszósági szint (0 = átlátszó, 1 = átlátszatlan)

```python
import matplotlib.pyplot as plt

x = [7, 8, 9, 10, 10, 12, 12, 12, 13, 14, 14, 15, 16, 16, 17, 18, 18, 19, 20, 20,
     21, 22, 23, 24, 25, 25, 26, 28, 30, 32, 35, 36, 38, 40, 42, 44, 48, 50]

plt.hist(x, bins=10, color='steelblue', edgecolor='black',linestyle='--', alpha=0.5)
plt.title("Histogram")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")
plt.show()
```

## Szórásdiagram (Scatter Plot)

A szóródási diagramokat a **változók közötti kapcsolatok megfigyelésére** használjuk. A matplotlib könyvtár `scatter()` metódusa használható szóródási diagramok rajzolására.

Szintaxis:

`matplotlib.pyplot.scatter(x, y)`

Paraméter: x, y A pontok koordinátái.
- `S`: Jelölő mérete (egyetlen érték vagy tömb)
- `C`: A jelölők színe vagy színsorrend
- `marker`: Jelölő stílusa, például kör, rombusz stb.
- `linewidths`: A jelölő szegélyeinek szélessége
- `edgecolor`: A jelölők szegélyének színe
- `Alpha`: Keverési érték 0 (átlátszó) és 1 (átlátszatlan) között

```python
import matplotlib.pyplot as plt

x = ['Thur', 'Fri', 'Sat', 'Sun', 'Thur', 'Fri', 'Sat', 'Sun']
y = [170, 120, 250, 190, 160, 130, 240, 200]

size = [2, 3, 4, 2, 3, 2, 4, 3]  
bill = [170, 120, 250, 190, 180, 130, 260, 200] 

plt.scatter(x, y, c=size, s=bill, marker='D', alpha=0.5)
plt.title("Scatter Plot")
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.show()
```

#  Kördiagram (Pie Chart)

A kördiagram egy kör alakú diagram, amely az adatokat **arányok vagy százalékok formájában** jeleníti meg. `A pie()` függvénnyel hozható létre, ahol minden szelet (ék) az egész egy részét jelöli.

Szintaxis:

`matplotlib.pyplot.pie(x, labels, autopct)`

Paraméter:

- `x:` A kördiagram szeleteinek adatértékei.
- `labels:` Az egyes szeletek nevei.
- `autopct`: Százalék megjelenítésének formátuma (pl. '%1.1f%%').
- `color` : Szeletek színe
- `shadow`: árnyék effect

```python
import matplotlib.pyplot as plt
import pandas as pd

cars = ['AUDI', 'BMW', 'FORD','TESLA', 'JAGUAR',]
data = [23, 10, 35, 15, 12]

explode = [0.1, 0.5, 0, 0, 0]
colors = ( "red", "blue", "yellow","grey", "green",)

plt.pie(data, labels=cars,  explode=explode, autopct='%1.2f%%',colors=colors, shadow=True)
plt.title("")
plt.show()
```

##  Dobozdiagram (Box Plot)

A dobozdiagram egy egyszerű grafikon, amely az adatok eloszlását mutatja. Megjeleníti **a minimumot, a maximumot, a mediánt és a kvartiliseket, valamint segít a kiugró értékek** könnyű kiszűrésében.

Szintaxis:

`matplotlib.pyplot.boxplot(x, notch, vert)`

Paraméter:

- `x`: Az az adat, amelyhez a dobozdiagramot meg kell rajzolni (általában egy lista vagy tömb).
- `notch`: Ha igaz, akkor egy bevágást rajzol a medián körüli konfidenciaintervallum megjelenítésére.
- `vert`: Ha igaz, a dobozok függőlegesek. Ha hamis, akkor vízszintesek.

```python
import matplotlib.pyplot as plt

data = [ [10, 12, 14, 15, 18, 20, 22],
         [8, 9, 11, 13, 17, 19, 21],
         [14, 16, 18, 20, 23, 25, 27] ]

plt.boxplot(data)
plt.xlabel("Groups")
plt.ylabel("Values")
plt.title("Box Plot")
plt.show()
```

> Ez a kód egy dobozdiagramot hoz létre az adateloszlás bemutatására és három csoport összehasonlítására a matplotlib használatával.

 ## Hőtérkép (Heat Map)
A hőtérkép az adatok grafikus ábrázolása, ahol az értékek **színekként jelennek** meg. Segít a minták, korrelációk vagy intenzitás **mátrixszerű formátumban történő vizualizálásában.** A Matplotlib `imshow()` metódusával hozható létre.

Szintaxis:

`matplotlib.pyplot.imshow(X, cmap='viridis')`

Paraméter:

- `X`: 2D tömb (képként vagy hőtérképként megjelenítendő adatok).
- `cmap`: Beállítja a színtérképet.

**Példa**: Ez a kód véletlenszerű 10×10-es adatokból készít hőtérképet az `imshow()` függvény segítségével. A 'viridis' színtérképet használja, a `colorbar()` pedig egy színskálát ad hozzá.

```python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(0)
data = np.random.rand(10, 10)

plt.imshow(data, cmap='viridis', interpolation='nearest')

plt.colorbar()
plt.xlabel('X-axis Label')
plt.ylabel('Y-axis Label')
plt.title('Heatmap')
plt.show()
```

> - `np.random.seed(0)`: Biztosítja, hogy minden alkalommal ugyanazok a véletlenszerű értékek jelenjenek meg (reprodukálhatóság).
> - `np.random.rand(10, 10):` 0 és 1 közötti véletlenszámokból álló 10×10-es tömböt generál.
