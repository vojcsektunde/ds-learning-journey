## Series 

**Egy dimenziós, cimkézett tömb** amelyben **bármilyen típus**ú adatot tárolhatunk. Létrehozható Python list-ből, NumPy tömbökből, szótárak, skaláris értékekből vagy CSV-ből betöltött adatok, Excel fájlok és adatbázisok. 

A series-hez társított cimkéket **indexnek** nevezzük. 

```python
import pandas as pd
import numpy as np

data = np.array(['P', 'y', 't', 'h', 'o', 'n'])
s = pd.Series(data)
print("Pandas Series:")
print(s)
```

> NumPy segítségével létrehozunk egy karaktereket tartalmazó tömböt, majd a `pd.Series()` parancsot használjuk hogy Panda Series-sé konvertáljuk. Ezután minden egyes elemhez egy alapértelmezett egész index tartozik, nullától kezdődően. 


## DataFrame

**Két-dimenziós adatszerkezet cimkézett tengelyekkel** *(sor, oszlop)*. Az adatok betöltésekor (már meglévő tárból) jön létre ami lehet SQL adatbázis vagy egy CSV / Excel fájl is. 

Létrehozható listákból, szótárakból, vagy szótárak listájából is. 

```python
import pandas as pd 
   
df = pd.DataFrame() 
print(df)
lst = ['Have', 'a', 'great', 'day'] 
  
df = pd.DataFrame(lst) 
print(df)
```

- `head()`: A DataFrame **első n sorának** megtekintése *(alapértelmezett érték 5 sor).*
- `tail()`: A DataFrame **utolsó n sorának** megtekintése *(alapértelmezett érték 5 sor).*
- `info()`: A DataFrame **összefoglalását** jeleníti meg, beleértve az **oszlopneveket**, az **adattípusokat** és a **nem null** értékeket.
- `describe()`: Összefoglaló statisztikákat biztosít, például **darabszámot**, **átlagot** és **szórást** numerikus oszlopokhoz.
    - Használd a `describe(include="all") `függvényt**nem numerikus adatok megadásához.**

```python
import pandas as pd

df = pd.read_csv("files/data.csv")
# df = pd.read_json('example.json') JSON fájl beolvasása esetén

print("First 3 rows using head():")
print(df.head(3))

print("\nLast 2 rows using tail():")
print(df.tail(2))

print("\nDataFrame summary using info():")
df.info()
```


### Az index elérése és módosítása
Lehetővé teszi **a sorok címkézésének megértését** és szükség szerinti **testreszabását**. A meglévő index az` .index` attribútum segítségével megtekinthető, és később az igényeknek megfelelően frissíthető.

### Egyéni index beállítása
A `set_index()` metódus egy DataFrame indexének módosítására szolgál egy vagy több oszlop új indexként való beállításával.

```python
import pandas as pd

data = {'Name': ['Jake', 'Mike'],
        'Age': [25, 30],
        'Salary': [50000, 55000]}
df = pd.DataFrame(data)
res = df.set_index('Name')
print(res)
```

### Az index visszaállítása
Ha vissza kell állítani az indexet az alapértelmezett egész indexre, használd a `reset_index()` metódust. Ez az **aktuális indexet egy normál oszloppá alakítja**, és létrehoz egy új alapértelmezett indexet.

```python
import pandas as pd

data = {'Name': ['Jake', 'Maria', 'Sam'],
        'Age': [25, 30, 22] }
df = pd.DataFrame(data)

res = df.reset_index(drop=True)
print(res)
```

 **Oszlopok elérése a DataFrame-ből**

A DataFrame **oszlopai egyenként is elérhetők** zárójeles jelölés `[ ]` használatával. 
Az adott oszlopot **sorozatként** kéri le, amely ezután **tovább manipulálható**.

**Sorok elérése index alapján**

Egy DataFrame adott sorainak *- indexpozíció vagy cimke alapján -* eléréséhez 2 metódust használunk:
 - `iloc` *(pozíció szerinti indexeléshez)* 
 -  `loc` *(címke alapú indexeléshez)*

 **Több sor vagy oszlop elérése egyszerre**
- Oszlopnevek vagy indexpozíciók listájának átadásával. Ez akkor hasznos, ha több oszlopot vagy sort kell kiválasztanod a további elemzéshez.

 **Sorok elérése feltételek alapján**
- A Pandas lehetővé teszi a sorok feltételek szerinti szűrését, ami nagyon hatékony lehet az adott kritériumoknak megfelelő adathalmazok feltárásához.

**Meghatározott cellák elérése**

- Cimke alapú indexeléshez, egy adott cella hozzáféréséhez: `.at[]` metódus
- egész szám pozíció alapú indexeléshez: `.iat[]` metódus
    - Ezek az egyes értékekhez való gyors hozzáférésre vannak optimalizálva.

```python
import pandas as pd

data = {'Name': ['John', 'Alice', 'Bob', 'Eve', 'Charlie'], 
        'Age': [25, 30, 22, 35, 28], 
        'Gender': ['Male', 'Female', 'Male', 'Female', 'Male'], 
        'Salary': [50000, 55000, 40000, 70000, 48000]}

df = pd.DataFrame(data)

# Oszlopok elérése
age_column = df['Age']
print(age_column)

# Sorok elérése index alapján
second_row = df.iloc[1]
print("\n",second_row)

# Több sor vagy oszlop elérése
subset = df.loc[0:2, ['Name', 'Age']]
print("\n",subset)

# Sorok elérése feltételek alapján
filtered_data = df[df['Age'] > 25]
print("\n",filtered_data)

# Meghatározott cellák elérése at használatával
salary_at_index_2 = df.at[2, 'Salary']
print("\n", salary_at_index_2)
```

## Adatok kiválasztása és szűrése feltételek alapján

Az adatok szűrése azt jelenti, hogy **csak azokat a sorokat vagy oszlopokat választjuk ki, amelyek megfelelnek bizonyos feltételeknek**, a teljes adathalmaz kezelése helyett. Ez segít a releváns adatok kinyerésében az elemzéshez logikai feltételek használatával.

```python
import pandas as pd

data = {'Name': ['Alice', 'Bob', 'Charlie'],'Age': [25, 30, 35]}
df = pd.DataFrame(data)
filtered_df = df[df['Age'] > 28]

print(filtered_df)
```

## Hiányzó adatok kezelése 

**Az adatkezelés egyik leggyakoribb feladata**. A Pandas számos függvényt kínál a hiányzó értékek hatékony azonosítására, kitöltésére és eltávolítására. Ez segít **biztosítani az adatminőséget** az elemzés elvégzése előtt.

- `isnull()`: Hiányzó értékeket ad vissza *(NaN)*, és hamis értéket, ahol értékek vannak jelen.True 
- `notnull()`: Igaz értéket ad vissza, ha van érték, és Hamis értéket, ha hiányzik.

Valós adathalmazokkal való munka során **gyakoriak a hiányzó értékek.** Általánosságban két fő módja van a hiányzó adatok kezelésének:

- Hiányzó értékek pótlása  - `.fillna()` *pl. hiányzó érték 0-val való helyettesítése.* Ez a metódus közvetlenül módosítja az adathalmazt, ha az `inplace=True` beállítást használjuk.


```python
import pandas as pd
import numpy as np

df = pd.DataFrame({'Col1': [1, 2, np.nan],'Col2': [3, np.nan, np.nan]})
df.fillna(0,inplace=True)

print(df)
```

- hiányzó értékekek törlése - a `dropna() `
    - Hiányzó értékeket tartalmazó **oszlopok** eltávolítása a `df.dropna(axis=1)` függvénnyel
    - Hiányzó értékeket tartalmazó **sorok** törlése a  `df.dropna(subset=['Column1', 'Column2'])` függvénnyel


```python
import pandas as pd

data = {'Name': ['Alice', 'Bob', None], 'Age': [25, None, 35]}
df = pd.DataFrame(data)
df_dropped = df.dropna()

print(df_dropped)
```
