## Series 

**Egy dimenziós, cimkézett tömb** amelyben **bármilyen típus**ú adatot tárolhatunk. Létrehozható Python list-ből, NumPy tömbökből, szótárak, skaláris értékekből vagy CSV-ből betöltött adatok, Excel fájlok és adatbázisok. 

A series-hez társított cimkéket **indexnek** nevezzük. 

```python
import pandas as pd
import numpy as np

data = np.array(['P', 'y', 't', 'h', 'o', 'n'])
s = pd.Series(data)
print("Pandas Series:")
print(s)
```

> NumPy segítségével létrehozunk egy karaktereket tartalmazó tömböt, majd a `pd.Series()` parancsot használjuk hogy Panda Series-sé konvertáljuk. Ezután minden egyes elemhez egy alapértelmezett egész index tartozik, nullától kezdődően. 


## DataFrame

**Két-dimenziós adatszerkezet cimkézett tengelyekkel** *(sor, oszlop)*. Az adatok betöltésekor (már meglévő tárból) jön létre ami lehet SQL adatbázis vagy egy CSV / Excel fájl is. 

Létrehozható listákból, szótárakból, vagy szótárak listájából is. 

```python
import pandas as pd 
   
df = pd.DataFrame() 
print(df)
lst = ['Have', 'a', 'great', 'day'] 
  
df = pd.DataFrame(lst) 
print(df)
```

DataFrame **indexe az egyes sorokhoz rendelt címkéket jelöli.** Segít az adatok hatékony azonosításában és elérésében, és lehet alapértelmezett numerikus érték vagy egyéni címkék.

### Az index elérése és módosítása
Lehetővé teszi **a sorok címkézésének megértését** és szükség szerinti **testreszabását**. A meglévő index az` .index` attribútum segítségével megtekinthető, és később az igényeknek megfelelően frissíthető.

```python
import pandas as pd

data = {'Name': ['Jake', 'Eve', 'Charlie'],
        'Age': [ 22, 35, 28],
        'Gender': [ 'Male', 'Female', 'Male'],
        'Salary': [40000, 70000, 48000]}
df = pd.DataFrame(data)

print(df.index)
print(df)
```

### Egyéni index beállítása
A `set_index()` metódus egy DataFrame indexének módosítására szolgál egy vagy több oszlop új indexként való beállításával.

```python
import pandas as pd

data = {'Name': ['Jake', 'Mike'],
        'Age': [25, 30],
        'Salary': [50000, 55000]}
df = pd.DataFrame(data)
res = df.set_index('Name')
print(res)
```

### Az index visszaállítása
Ha vissza kell állítani az indexet az alapértelmezett egész indexre, használd a `reset_index()` metódust. Ez az **aktuális indexet egy normál oszloppá alakítja**, és létrehoz egy új alapértelmezett indexet.

```python
import pandas as pd

data = {'Name': ['Jake', 'Maria', 'Sam'],
        'Age': [25, 30, 22] }
df = pd.DataFrame(data)

res = df.reset_index(drop=True)
print(res)
```

 ### Oszlopok elérése a DataFrame-ből
A DataFrame **oszlopai egyenként is elérhetők** zárójeles jelölés használatával. Egy oszlop elérése az adott oszlopot sorozatként kéri le, amely ezután tovább manipulálható.

### Sorok elérése index alapján
Egy DataFrame adott sorainak eléréséhez használhatja az `iloc` metódust *(pozíció szerinti indexeléshez)* vagy a `loc` metódust *(címke alapú indexeléshez)*. Ezek a metódusok lehetővé teszik a sorok indexpozícióik vagy címkéik alapján történő lekérését.

### Több sor vagy oszlop elérése
Egyszerre több sorhoz vagy oszlophoz is hozzáférhetsz oszlopnevek vagy indexpozíciók listájának átadásával. Ez akkor hasznos, ha több oszlopot vagy sort kell kiválasztanod a további elemzéshez.

### Sorok elérése feltételek alapján
A Pandas lehetővé teszi a sorok feltételek szerinti szűrését, ami nagyon hatékony lehet az adott kritériumoknak megfelelő adathalmazok feltárásához.

### Meghatározott cellák elérése az at és az iat használatával
Ha egy adott cellához kell hozzáférnie, használhatja az `.at[]` metódust a címke alapú indexeléshez, és az `.iat[]` metódust az egész szám pozíció alapú indexeléshez. Ezek az egyes értékekhez való gyors hozzáférésre vannak optimalizálva.

```python
import pandas as pd

data = {'Name': ['John', 'Alice', 'Bob', 'Eve', 'Charlie'], 
        'Age': [25, 30, 22, 35, 28], 
        'Gender': ['Male', 'Female', 'Male', 'Female', 'Male'], 
        'Salary': [50000, 55000, 40000, 70000, 48000]}

df = pd.DataFrame(data)

# Oszlopok elérése
age_column = df['Age']
print(age_column)

# Sorok elérése index alapján
second_row = df.iloc[1]
print("\n",second_row)

# Több sor vagy oszlop elérése
subset = df.loc[0:2, ['Name', 'Age']]
print("\n",subset)

# Sorok elérése feltételek alapján
filtered_data = df[df['Age'] > 25]
print("\n",filtered_data)

# Meghatározott cellák elérése at használatával
salary_at_index_2 = df.at[2, 'Salary']
print("\n", salary_at_index_2)
```

## Operátorok

Alapvető műveleteket biztosít, a struktúrált adatokkal való munkához. 

- **Adatbetöltés**: CSV, Excel vagy JSON fájlokból olvas ki adatot egy DataFrame-be. 

- **Adatok megtekintése:**  fontos megérteni azok szerkezetét és tartalmát. Ezek a módszerek lehetővé teszik a sorok, az összesített statisztikák és a metaadatok vizsgálatát. A `df.info()` függvény információkat jelenít meg a DataFrame-ről, beleértve az oszlopneveket, az adattípusokat, a nem null értékek számát és a memóriahasználatot.

- **Hiányzó adatok kezelése** : Az adathalmazok gyakran tartalmaznak üres vagy hiányzó értékeket. A Pandas függvényeket biztosít ezen értékek észlelésére, eltávolítására vagy cseréjére: df.fillna(0) a hiányzó értéket nullával helyettesíti

- **Adatok kiválasztása és szűrése:** Ez a művelet egy feltételnek megfelelő oszlopokat, sorokat vagy rekordokat kér le. Lehetővé teszi a szükséges információk pontos kinyerését.

- **Oszlopok hozzáadása és eltávolítása**: Létrehozhat új oszlopokat a meglévők alapján, vagy törölheti a nem kívánt oszlopokat a DataFrame-ből.

- ** Adatok csoportosítása (GroupBy)**: A csoportosítás lehetővé teszi az adatok kategóriákba rendezését és az egyes csoportok értékeinek kiszámítását, például összegzéseket, darabszámokat vagy átlagokat.
