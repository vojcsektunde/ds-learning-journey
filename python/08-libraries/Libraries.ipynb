# **DSA Libraries** - DSA könyvtárak

Az **adatszerkezetek és algoritmusok (Data Structures and Algorithms  - DSA)** képezik a hatékony problémamegoldás alapját a programozásban. 

A Python számos **beépített és külső könyvtárat biztosít**, amelyek segítenek az olyan gyakori adatszerkezetek megvalósításában, mint a **tömbök, láncolt listák, várólisták, hash térképek, heapek és fák.**

## **Beépített DSA könyvtárak**
Ezek a könyvtárak nem igényelnek telepítést, és széles körben használják őket a valós alkalmazásokban:

### Array modul

A tömb modul tömb adatstruktúrát valósít meg, ahol az **azonos adattípusú elemek összefüggő memóriahelyeken tárolódnak.** Főleg akkor használják, ha a listákhoz képest memóriahatékonyságra és **gyorsabb hozzáférésre** van szükség, különösen **numerikus adatok** esetén.

Fontos módszerek

- `append(x)`: Egyetlen elemet ad hozzá a tömb végéhez

- `extend(iterable)`: Több elemet ad hozzá egy másik iterálható objektumból
- `count(x)`: Visszaadja, hogy egy érték hányszor jelenik meg
- `itemsize()`: Visszaadja egy elem memóriaméretét (bájtban)
- `buffer_info():` Visszaadja a memóriacímet és az elemek számát

---
## **Numpy könyvtár**
A NumPy támogatja a **többdimenziós tömbös adatszerkezeteket**, és nagy teljesítményű numerikus és tudományos számításokra van optimalizálva.

Fontos műveletek

- `np.array():` 1D, 2D vagy N dimenziós tömböket hoz létre
- `np.zeros()` / `np.ones()` : Nullákkal vagy egyesekkel töltött tömböket hoz létre
- `np.arange()` / `np.linspace()` : Numerikus sorozatokat generál

- `np.reshape()` : Megváltoztatja egy tömb alakját

- `np.sum()` / `np.mean()` / `np.max()` / `np.min()` : Matematikai műveleteket hajt végre
- `np.dot()` : Mátrix- és vektorszorzás

Array slicing és indexelés: `arr[start:stop]`, `arr[row, col]`

Broadcasting : Automatikus alakzatigazítás műveletek közben

---
## **Deque modul**

A deque modul egy **duplán láncolt lista** adatstruktúrát valósít meg. Lehetővé teszi a **konstans idejű beszúrást** és törlést **mindkét végről**, így ideális várólisták és veremelemek számára .

Fontos műveletek

- `append(x)`: Elem beszúrása a jobb oldali végre

- `appendleft(x)`: Elem beszúrása a bal oldali végre

- `pop()`: Elem eltávolítása a jobb oldalról

- `popleft()`: Elem eltávolítása a bal oldalról
- `extend(iterable)`: Több elem hozzáadása jobbra
---
## **Queue modul**
A `queue.Queue` modul egy **FIFO sorrendet** követő sor adatstruktúrát biztosít . Szálbiztosra tervezték, és főként többszálú, valamint termelő-fogyasztó alkalmazásokban használják.

Fontos műveletek

- `put(x)`: Beszúr egy elemet a sorba

- `get()`: Elem eltávolítása és visszaadása

- `qsize()`: Visszaadja az elemek aktuális számát

- `empty()`: Ellenőrzi, hogy a sor üres-e
- `full()`: Ellenőrzi, hogy a sor tele van-e

---
## **Collections modul**
Speciális adatszerkezeteket biztosít, amelyek kiterjesztik a Python beépített konténereit. Általában különféle adatszerkezetekkel való munkára használják, főként a következőkkel:

- **Hash mapek :** adatok kulcs-érték párokként történő, gyors hozzáférésű tárolására
- **Stackek (LIFO)**: push és pop műveletekhez
- **Sorok (FIFO)**: sorba helyezési és sorból való kivételi műveletekhez
- **Duplán láncolt listák:** elemek egyszerű hozzáadásához és eltávolításához mindkét végről
- **Strukturált rekordok:** kapcsolódó értékek tárolására nevek használatával indexek helyett
---
- **Gyakori adatszerkezetek gyűjteményekben**

    - `dict` kulcs-érték párokként tárolja az adatokat

    - `collections.Counter` az elemek előfordulásait számlálja.

    - `collections.defaultdict` automatikusan létrehozza az alapértelmezett értékeket.
    - `collections.OrderedDict` a kulcsokat beszúrási sorrendben tárolja.

    - `collections.ChainMap` több szótárat egyesít egyesítés nélkül.

    - `deque`, amelyet  verem felépítésére használnak
    - `namedtuple` objektumokhoz hasonlóan elnevezett mezőkkel tárolja az adatokat

Fontos műveletek

- `keys()`: Visszaadja az összes kulcsot
- `values()`: Visszaadja az összes értéket

- `items()`: Kulcs-érték párokat ad vissza

- `get(key)`: Biztonságosan lekéri az értéket
- `update(dict)`: Több bejegyzés frissítése

- `append(x)`: Elemet ad hozzá a jobb oldali véghez (deque)
- `popleft():` Eltávolítja az elemet a bal oldali végről (deque)
---
## **Heapq modul**
A heap adatstruktúrát, konkrétan a min-heap-et valósítja meg. Lehetővé teszi a **legkisebb elem gyors elérését**, és széles körben használják **prioritási sorokban és ütemező algoritmusokban**.

Fontos műveletek

- `heapify(list) `: egy listát halommá alakít.
- `heappush(heap, x)` : hozzáad egy elemet a heap-hez.
- `heappop(heap)` : eltávolítja és visszaadja a legkisebb elemet.
- `nlargest(n, iterable)` : n legnagyobb elemet ad vissza.
- `nsmallest(n, iterálható) `: n legkisebb elemet ad vissza.

---
## **Felező modul**
**Támogatja a bináris keresési műveleteket rendezett listákon**. Segít fenntartani a rendezett sorrendet, miközben hatékonyan illeszti be az elemeket a lista manuális rendezése nélkül.

Fontos módszerek

- `bisect_left(list, x):` bal szélső beszúrási pont.
- `bisect_right(list, x)`: jobb szélső beszúrási pont.

- `insort_left(list, x):` a bal szélső érvényes pozícióba szúr be.
- `insort_right(list, x):` a legjobb szélső érvényes pozícióba szúr be.

- `bisect(lista, x):` a bisect_right aliasa.

## Külső könyvtárak fejlett adatszerkezetekhez
Ezek alapértelmezés szerint nem részei a Pythonnak. Telepíteni kell őket, és csak akkor használjuk őket, ha a problémádnak valóban szüksége van rájuk.

### **treelib** (fa adatszerkezet)
Ez a könyvtár fa adatszerkezetek megvalósítására szolgál , ahol az adatok **szülő-gyermek hierarchiában tárolódnak**. Olvashatóvá teszi a **fák létrehozását, kezelését és megjelenítését**, ami hasznos hierarchikus adatokkal való munka során.

> pip install treelib

Fontos módszerek

- `create_node(tag, id, parent)`: Új csomópontot ad a fához
- `remove_node(id)`: Eltávolít egy csomópontot
- `get_node(id):` Lekér egy adott csomópontot
- `children(id)`: Visszaadja a gyermekcsomópontokat
- `show():` Megjeleníti a fa struktúráját


## **intervaltree** (Intervallumfa)

Ez a függvénykönyvtár numerikus tartományok (intervallumok) hatékony tárolására és keresésére szolgál. Hasznos, ha gyorsan meg kell találni az átfedő intervallumokat vagy tartományokat.

> pip install intervaltree

- `add(interval)`: hozzáad egy intervallumot.
- `remove(interval)`: eltávolít egy intervallumot.
- `search(start, end)`: egy tartományt átfedő intervallumokat keres.
- `at(point)`: megkeresi azokat az intervallumokat, amelyek egy adott pontot tartalmaznak.
- `clear()`: eltávolítja az összes intervallumot

## **pygtrie** (Trie/Előtagfa)
Ez a könyvtár a Trie (előtagfa) adatstruktúra egy implementációját biztosítja . **Karakterenként tárolja a karakterláncokat, így az előtag alapú keresések nagyon gyorsak.**

> pip install pygtrie

Fontos módszerek

- `insert(word)`: hozzáad egy szót a trie-hoz.
- `search(word)`: ellenőrzi, hogy létezik-e egy szó.
- `startswith(prefix)`: ellenőrzi, hogy van-e olyan szó, amelyik tartalmazza az előtagot.
- `delete(word)`: eltávolít egy szót.
- `words(prefix)`: előtaggal kezdődő szavakat ad vissza
---

- Használj külső DSA könyvtárat, ha:

    - A nulláról történő megvalósítás időigényes
    -  Speciális struktúrákra van szükséged (próbálkozások, intervallumfák, egyedi fák)
    - A könyvtár karbantartott és megbízható
    - Javítani szeretnéd a fejlesztési sebessége

- Kerüld a használatukat, ha:

    - A beépített szerkezetek is képesek a feladatra
    - Maximális teljesítményt szeretnél (a beépített modulok C-re optimalizáltak)

# **GUI Library and Packages**

A **grafikus felhasználói felületek (GUI-k)** lehetővé teszik az alkalmazásokkal való interakciót vizuális elemeken, például **gombokon, menükön, szövegmezőkön és ablakokon** keresztül. A parancssori bevitel interaktív komponensekkel való helyettesítésével a GUI-k intuitívabbá és könnyebben hozzáférhetővé teszik az alkalmazásokat.

A Python számos grafikus felhasználói felületű könyvtárat kínál egyszerű asztali eszközök és teljes körű alkalmazások fejlesztéséhez:

## **Tkinter**

 Alapvető **widgeteket** és struktúrát biztosít **asztali alkalmazások készítéséhez.** A Tkinter többplatformos, és kezdőktől a közepes méretű eszközökig alkalmas.

- Pythonnal együtt érkezik, telepítést nem igényel
- Windows, macOS és Linux rendszereken is működik
- Tiszta és kezelhető kódstruktúra
- Jó dokumentáció és hosszú távú közösségi támogatás

## **Kivy**
Többplatformos keretrendszer asztali és mobil alkalmazások fejlesztéséhez. Támogatja a többérintéses vezérlést, a gesztusokat, és különösen alkalmas **érintőképernyős vagy mobil felületekhez.**

- Működik Windows, Linux, macOS, Android és iOS rendszereken
- Támogatja a többérintéses és gesztusalapú interakciókat
- Egységes megjelenés minden platformon
- Nyílt forráskódú és aktívan karbantartott

## **Streamlit**

 Interaktív webes alkalmazások közvetlenül Python szkriptekből történő létrehozására szolgál. Különösen népszerű adatelemzési irányítópultokhoz, **vizualizációkhoz és gépi tanulási demókhoz**.

- Webalkalmazásokat készít HTML, CSS vagy JavaScript nélkül
- Valós idejű frissítések minimális kóddal
- Ideális adatelemzéshez és modelldemókhoz
- Erős közösségi és bővítmény-támogatás

## **PyQt**
 Python kötéseket biztosít a Qt keretrendszerhez, amely egy eszközkészlet **professzionális grafikus felhasználói felületű alkalmazásokhoz. Nagyméretű widgeteket** és eszközöket tartalmaz többplatformos alkalmazások fejlesztéséhez.

- Átfogó widgetek és eszközök készlete
- Kiváló nagy és fejlett alkalmazásokhoz
- Minden főbb operációs rendszer platformon működik
- Elrendezéstervezőket és fejlesztőeszközöket kínál

## **PyGUI**
A minimalizmusra összpontosít, és közvetlenül integrálódik a natív operációs rendszer grafikus felhasználói felületének elemeivel. Letisztult absztrakciókat biztosít, amelyek alkalmasak a könnyűsúlyú alkalmazásokhoz.

- Kicsi és könnyű
- Natív rendszermegjelenésű alkalmazásokat hoz létre
- Könnyen érthető és használható
- Minimális rezsiköltség

## **wxPython**
Az egyes platformok natív grafikus felhasználói felületének komponenseit veszi körül. Valódi natív megjelenésű alkalmazásokat hoz létre, és széles választékban kínál widgeteket.

- Natív operációs rendszer megjelenést és érzetet biztosít
- Számos fejlett widgetet kínál
- Érett és stabil könyvtár
- Közepes és nagy alkalmazásokhoz alkalmas


## **PySimpleGUI**
**A Tkinter, a Qt és a wxPython csomagokat egy egyszerűbb felületbe foglalja.**Gyors prototípusok és kisebb eszközök minimális kódolási erőfeszítéssel történő létrehozásához használják.

- Rendkívül egyszerű szintaxis
- Ideális gyors grafikus felhasználói felület létrehozásához
- Több háttérrendszerrel integrálható
- Csökkenti a sablonkód mennyiségét
