# **Pathlib**

(a 3.4-es verzióban bevezetett) **objektumorientált módot** kínál a fájlrendszer-elérési utakkal való munkára.

 A hagyományos `os.path`-sal ellentétben, amely az elérési utakat egyszerű karakterláncokként kezeli, a **pathlib objektumokként ábrázolja azokat**, így az olyan műveletek, mint az elérési utak összekapcsolása, a fájlok olvasása/írása és a létezés ellenőrzése, sokkal **letisztultabbak és platformfüggetlenebbek**.

A modulban található `Path` osztály a **fő osztály**, amely a fájl- és könyvtárelérési utak ábrázolására és azokkal való interakcióra szolgál a tényleges fájlrendszeren.

- **Platformfüggetlen kompatibilitás:** ugyanaz a kód működik Windows, Linux és macOS rendszereken.
- **Biztonságosabb műveletek:** metódusok és operátorok túlterhelésének (/) felajánlásával elkerülhető a manuális karakterlánc-összefűzés.

```python
from pathlib import Path

current_dir = Path.cwd()
print("Current directory:", current_dir)

# Útvonal összekapcsolása a / operátorral
file_path = current_dir / "assets" / "first.txt"
print("Full path:", file_path)
```

## Fájl- és könyvtárműveletek
Leegyszerűsíti a mindennapi feladatokat, mint például a mappák listázása, a fájlok adatainak ellenőrzése, valamint a fájlok olvasása és írása. 

### Alkönyvtárak listázása
Egy adott könyvtár összes alkönyvtárának megtekintéséhez használd az iterdir() metódust, és szűrd az eredményeket az `.is_dir() függvénnyel.` Ez segít a mappák egyszerű azonosításában fájlok vizsgálata nélkül.


```python
from pathlib import Path

# A 06-file-handling mappára mutató objektum
base_path = Path(".")

print("Subdirectories:")
for subdir in base_path.iterdir():
    if subdir.is_dir():
        print(subdir.name)
```

> felsorolja a root könyvtárban található összes alkönyvtárat (/ Linuxon vagy C:/ Windowson).

> - ` Path('.'):` Létrehoz egy Path objektumot, amely az alkönyvtárakra mutat (/ Unix/Linux rendszereken vagy C:/ Windows rendszereken) `/` jel esetén a gyökérkönyvtárra.

> - `iterdir():` Visszaadja a p elérési út összes közvetlen bejegyzését (fájlokat és könyvtárakat).

> - `is_dir():` Ellenőrzi, hogy az aktuális bejegyzés egy könyvtár-e (a fájlokat figyelmen kívül hagyja).

### Fájlok keresése a könyvtárfában

Az `rglob()` metódus rekurzív keresést hajt végre a könyvtárfában egy adott mintázat alapján (pl. az `assets` mappában lévő összes szöveges fájl).

```python
from pathlib import Path

# mappa kijelölése
assets_path = Path("assets")

# Rekurzív keresés .txt kiterjesztésű fájlokra
txt_files = assets_path.rglob("*.txt")

print("Founded .txt files' path:")
for file in txt_files:
    print("-",file)
```

### Útvonal tulajdonságainak lekérdezése
A `Path` objektumok különféle **tulajdonságokkal rendelkeznek,** amelyekkel könnyen lekérdezhető az elérési út neve, kiterjesztése, szülőkönyvtára vagy az, hogy abszolút útvonalról van-e szó.

```python
from pathlib import Path

target_file = Path("assets") / "students.csv"

print("Is absolute:", target_file.is_absolute())
print("File name:", target_file.name)
print("Extension:", target_file.suffix)
print("Parent directory:", target_file.parent)
```

### Fájl olvasása és írása

Megkönnyíti a fájlok kezelését azáltal, hogy az `.open()` metódussal közvetlenül nyithatunk meg fájlokat olvasásra vagy írásra.

```python
from pathlib import Path

# Fájl olvasása az assets mappából
file_to_read = Path("assets") / "world.txt"

with file_to_read.open("r", encoding="utf-8") as f:
    content = f.read()
    print("Opened file's content(s): ",content)

# Új fájl írása
output_file = Path("assets") / "pathlib_example.txt"
with output_file.open("w", encoding="utf-8") as f:
    f.write("Hello, Pathlib!")

print("New file successfully created and written.")
```
