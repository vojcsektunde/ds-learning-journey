# **File Handling** (fájlkezelés)

Fájlokon végzett műveletek - **létrehozás, megnyitás, olvasás, írás és bezárás**-  folyamatát jelenti. 

- Az **adatokat véglegesen tárolja,** még a program befejezése után is.
- Hozzáférés külső fájlokhoz, például **.txt, .csv, .json** stb.
- Nagy fájlok hatékony feldolgozása **sok memória használata** nélkül.
- Automatizáljon olyan feladatokat, mint a konfigurációk olvasása vagy a kimenetek mentése.

## Fájl megnyitása, beolvasása

Egy fájl **megnyitásához** használhatjuk az `open()` függvényt, amely **argumentumként a fájl elérési útját** és a módot kéri.

A `close() `metódus bezárja a fájlt és felszabadítja a rendszer erőforrásait. Ha a fájlt írási vagy hozzáfűzési módban nyitották meg, a **bezárás biztosítja, hogy minden módosítás megfelelően mentésre kerüljön.**

Szintaxis: `file = open('filename.txt', 'mode', 'encoding')`

Az alábbi metódusok használatával beolvasható:

- `read()` (összes sor, egyetlen stringként, nagy memóriaigény)
- `readline()` (egyetlen sor, stringként, alacsony memóriaigény)  
- `readlines()` (összes sor, listként ha indexelni szeretnéd, közepes vagy nagy memóriaigény)


```python
file = open("assets/first.txt", "r", encoding="utf-8")  
rows = file.read()
file.close()

print(rows)
```


## Fájltulajdonságok ellenőrzése:

## Fájlba írás: 
- `write()` metódussal adhatunk új szöveget a fájlhoz˙
- Teljesen átírja az eddigi fájlt, **mode argumentumként** `w` adjuk meg. 

## `open()` mode-ok:

- `r` - fájl olvasása, hibát dob, ha nem létezik
- `w` write mode - létrehozza a fájlt, ha az már létezik, teljesen törli  a korábbi tartalmát
- `a` append mode - hiányzó fájl létrehozása, az adatokat mindig a végére írja
- `x` exclusive create - új fájlt hoz létre, de `FileExistsError` hibával hibát jelez, ha már létezik.
- `b` binary flag - más módokkal (pl. "wb", "ab") használatos bináris fájlok esetén
- `+`  read/write flag -  (pl. "r+", "w+") kombinálható mind olvasás, mind írás esetén

 - `"encoding="` - szövegkódolás megadása pl utf-8
 - `"newline="` - új sor karakterek fordításának szabályozása szöveges módban (pl. "\n")

## with statement használata:
- A fájl manuális megnyitása és bezárása helyett használhatja a `with` utasítást , amely **automatikusan kezeli a bezárást.**
- Ez csökkenti a **fájlok sérülésének** és az erőforrás-szivárgás **kockázatát.**


```python
with open("assets/first.txt", "w") as file:
    file.write("Hello Python!\n")
    file.write("File handling is easy with Python.")

print("File written successfully")
```

```python
with open("assets/first.txt", "a", encoding="utf-8") as f:
    f.write("\nAppended a new line.")

with open("assets/first.txt", "r", encoding="utf-8") as f:
    print(f.read())
```

### "Csak akkor hozd létre, ha nem létezik"

Az „x” módban a Python új fájlt hoz létre, de FileExistsError hibát generál, ha a fájl már létezik. Ez megakadályozza a véletlen felülírásokat.

```python
try:
    with open("assets/file.txt", "x", encoding="utf-8") as f:
        f.write("Created using exclusive mode.\n")
except FileExistsError:
    print("file.txt already exists, exclusive creation aborted.")
```

### Több sor írása
Néha több sort kell egyszerre írni egyenként ahelyett, hogy egyesével kellene írni. 

A Python két gyakori megközelítést kínál: a `writelines()-t` karakterláncok listájához és a `join()-t` egyetlen szövegblokk létrehozásához.

```python
lines = ["\nFirst line\n", "Second line\n", "Third line\n"]
with open("assets/multiple_lines.txt", "w", encoding="utf-8") as f:
    f.writelines(lines)

with open("assets/multiple_lines.txt", "r", encoding="utf-8") as f:
    print(f.read())
```

```python
lines = ["Line A", "Line B", "Line C"]
text = "\n".join(lines) + "\n"
with open("assets/multiple_lines.txt", "w", encoding="utf-8") as f:
    f.write(text)

with open("assets/multiple_lines.txt", "r", encoding="utf-8") as f:
    print(f.read())
```

## Kivételek kezelése fájl bezárásakor:

- Fontos a kivételek kezelése annak érdekében, hogy a fájlok **megfelelően bezáródjanak,** még akkor is, ha hiba történik a fájlműveletek során. 
- A `finally` blokk biztosítja, hogy a fájl bezáródjon, **még akkor is, ha hiba történik.**

```python
try:
    with open("firstt.txt", "r") as file:
        content = file.read()
        print(content)
except FileNotFoundError as fnfe:
    print("Error:", fnfe)
```

>  `try` blokk olyan kódot tartalmaz, amely hibát okozhat, az `except` blokk bizonyos hibákat, például hiányzó fájlokat kezel.

## CV fájlok olvasása

A CSV *(vesszővel elválasztott értékek)* formátumot széles körben használják táblázatos adatok tárolására. 
CSV könyvtárbeli olvasás: `csv.reader()` paranccsal

```python
import csv

with open("assets/students.csv", "r") as csv_file:
    header = csv_file.readline()
    rows = csv.reader(csv_file)
    for row in rows:
        print(row)
```

## **JSON fájlok olvasása**

A JSON *(JavaScript Object Notation)* fájlok olvasása széles körben elterjedt **API-kban** és webes alkalmazásokban adatátvitelhez.

 A Python modulja lehetővé teszi számunkra, hogy a JSON-t egyszerűen betöltsük szótárakba további feldolgozás céljából.

```python
import json
with open("assets/employee_data.json", "r") as jsonfile:
    data = json.load(jsonfile)
    #print(data) - az egész JSON fájlt kiirná
    
    print(data["company"],"'s employees:\n")   
    
    for employee in data["employees"]:
        name = employee["firstName"] + " " + employee["lastName"]
        position = employee["position"]
        city = employee["contact"]["city"]
        print(f"{name} - {position} ({city})")
```

## **os modul**

Hordozható **(platformfüggetlen)** módot biztosít az **operációs rendszerrel** való interakcióhoz. A Python standard könyvtárának része, így nem kell külön telepíteni.

Az Aktuális munkakönyvtár **(CWD)** az a mappa, ahol a Python jelenleg fut. Ez a könyvtár eltérhet attól a helytől, ahol a Python szkript tárolva van. **Lekéréséhez** az `os.getcw()` függvényt használjuk.


Az aktuális munkakönyvtár **módosítása** az `os.chdir(path)` használatával módosítható

```python
import os

cwd = os.getcwd()
print(f"Current working directory: {cwd}")

os.chdir('../')
print(f"New working directory: {os.getcwd()}")

# Visszalépés az eredeti helyre 
os.chdir(cwd)
```

 ### Könyvtár létrehozása
 
 Az `os.mkdir()` függvény egy könyvtár **létrehozására** szolgál a megadott elérési úton, és opcionálisan beállítható hozzá egy jogosultsági mód is. `FileExistsError` hibát generál, ha a könyvtár már létezik. 

 Az `os.makedirs()` függvény **rekurzív** könyvtárakat hoz létre, ami azt jelenti, hogy a **hiányzó szülőkönyvtárakat is létrehozza.** 

```python
import os

initial_directory = os.getcwd()
print(f"Initial working directory: {initial_directory}")


project_folder = "created_folder"
os.mkdir(project_folder)
print(f"The folder '{project_folder}' has been successfully created.")

# Az aktuális munkakönyvtár módosítása 
os.chdir(project_folder)
updated_directory = os.getcwd()
print(f"The new working directory: {updated_directory}")


nested_path = os.path.join("src", "data")
os.makedirs(nested_path)
print(f"The entire folder structure '{nested_path}' has been created (src -> data).")
```

### Fájlengedélyek és metaadatok

Az operációs rendszer modul **hozzáférést biztosít az alacsonyabb szintű fájlrendszer metaadatokhoz** és jogosultságkezeléshez, amelyek hasznosak szkriptelés, adminisztráció és rendszerszintű feladatokhoz. 

Három fontos metódus ebben a kategóriában:

- **`os.chmod()`** - fájl vagy könyvtár jogosultságainak (olvasás, írás, végrehajtás) módosítására szolgál. A jogosultságokat oktális formátumban kell átadni *(pl. 0o600).*

- **`os.chown()`** - lehetővé teszi egy fájl tulajdonosának és csoportazonosítójának módosítását. Ezt jellemzően rendszerparancsfájlokban használják, és megfelelő jogosultságokat igényel. Csak Unix/Linux rendszereken használatos.

- **`os.stat()`** - egy fájl metaadatainak, például méretének, jogosultságainak és időbélyegeinek lekérésére szolgál.
    - st_size : Méret bájtban
    - st_mtime : Utolsó módosítás időbélyege (UNIX idő)
    - st_mode : Fájlmód bitek; az utolsó 3 számjegyet kinyerjük az engedélyinformációkhoz
    - Hasonló az ls -l parancs kimenetéhez Unix/Linux alatt

## Gyakran használt függvények


#### 1. Rendszerinformációk és hibakezelés

- `os.name:` Megadja az importált operációs rendszertől függő modul nevét.
 Lehetséges értékek: 'posix' (Linux/macOS), 'nt' (Windows), 'os2', 'ce', 'java', 'riscos'.
- `os.error:` A beépített OSError kivétel álneve. Érvénytelen/elérhetetlen fájlnevek vagy hibás OS-szintű argumentumok esetén váltódik ki. Modern Pythonban ajánlott közvetlenül az OSError-t (vagy specifikusabb utódait, pl. FileNotFoundError) használni.

#### 2. Fájlok megnyitása és bezárása

- `os.popen(parancs, mode='r', bufsize=-1):` Megnyit egy csővezetéket (pipe) a megadott parancsból vagy parancsba. A visszatérési érték egy olvasható vagy írható fájlobjektum (módtól függően).

- `os.close(fd):` Alacsony szintű fájlleírók (egész számok, pl. amit az `os.open()` ad vissza) bezárására szolgál.
	- A sima `open()` függvénnyel megnyitott fájlobjektumokon meghívva `TypeError`-t dob, mert azokhoz a beépített .`close()` metódust kell használni.

#### 3. Fájlrendszer műveletek

- `os.rename(régi_név, új_név):` Átnevezi a megadott fájlt vagy könyvtárat. Csak akkor működik, ha a fájl létezik és van hozzá jogosultságunk, különben `FileNotFoundError`-t (vagy engedélyezési hibát) eredményez.

#### 4. Fájlok tulajdonságainak vizsgálata (os.path)

-  `os.path.exists(útvonal):` Ellenőrzi, hogy a megadott fájl vagy mappa létezik-e. Visszatérési értéke logikai True vagy False.
- `os.path.getsize(útvonal):` Megadja a paraméterként átadott fájl méretét bájtban.


```python
import os 

file ="Files.ipynb"

result = os.path.exists("file")
print(result)

size = os.path.getsize(file)
print("Size of the file is", size,"bytes.")
```
