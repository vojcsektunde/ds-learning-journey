# **File Handling** (fájlkezelés)

Fájlokon végzett műveletek - **létrehozás, megnyitás, olvasás, írás és bezárás**-  folyamatát jelenti. 

- Az **adatokat véglegesen tárolja,** még a program befejezése után is.
- Hozzáférés külső fájlokhoz, például **.txt, .csv, .json** stb.
- Nagy fájlok hatékony feldolgozása **sok memória használata** nélkül.
- Automatizáljon olyan feladatokat, mint a konfigurációk olvasása vagy a kimenetek mentése.

## Fájl megnyitása, beolvasása

Egy fájl **megnyitásához** használhatjuk az `open()` függvényt, amely **argumentumként a fájl elérési útját** és a módot kéri.

A `close() `metódus bezárja a fájlt és felszabadítja a rendszer erőforrásait. Ha a fájlt írási vagy hozzáfűzési módban nyitották meg, a **bezárás biztosítja, hogy minden módosítás megfelelően mentésre kerüljön.**

Szintaxis: `file = open('filename.txt', 'mode', 'encoding')`

Az alábbi metódusok használatával beolvasható:

- `read()` (összes sor, egyetlen stringként, nagy memóriaigény)
- `readline()` (egyetlen sor, stringként, alacsony memóriaigény)  
- `readlines()` (összes sor, listként ha indexelni szeretnéd, közepes vagy nagy memóriaigény)


```python
file = open("first.txt", "r", encoding="utf-8")  
rows = file.read()
file.close()

print(rows)
```


## Fájltulajdonságok ellenőrzése:

```python
f = open("first.txt", "r")

print("Filename:", f.name)
print("Mode:", f.mode)
print("Is Closed?", f.closed)

f.close()
print("Is Closed?", f.closed)
```

---

## Fájlba írás: 
- Teljesen átírja az eddigi fájlt, **mode argumentumként** `w` adjuk meg. 
- `write()` metódussal adhatunk új szöveget a fájlhoz

## with statement használata:
- A fájl manuális megnyitása és bezárása helyett használhatja a `with` utasítást , amely **automatikusan kezeli a bezárást.**
- Ez csökkenti a **fájlok sérülésének** és az erőforrás-szivárgás **kockázatát.**


```python
with open("first.txt", "w") as file:
    file.write("Hello Python!\n")
    file.write("File handling is easy with Python.")

print("File written successfully")
```

## Kivételek kezelése fájl bezárásakor:

- Fontos a kivételek kezelése annak érdekében, hogy a fájlok **megfelelően bezáródjanak,** még akkor is, ha hiba történik a fájlműveletek során. 
- A `finally` blokk biztosítja, hogy a fájl bezáródjon, **még akkor is, ha hiba történik.**

```python
try:
    with open("firstt.txt", "r") as file:
        content = file.read()
        print(content)
except FileNotFoundError as fnfe:
    print("Error:", fnfe)
```

>  `try` blokk olyan kódot tartalmaz, amely hibát okozhat, az `except` blokk bizonyos hibákat, például hiányzó fájlokat kezel.
