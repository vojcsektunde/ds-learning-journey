# **Data structures** (Adatszerkezetek)

## **String** (karakterlánc)

- Idézőjelek (egyszeres `''` vagy dupla `""`) között írt karaktersorozatok. 
- Tartalmazhat betűket, számokat, szóközöket és speciális karaktereket is. 
- Pythonban nincsen külön karaktertípus *(char)*
- Többsoros karakterláncok létrehozásához `"""` vagy `'''`-t használunk

#### **Karakterek elérése a Stringben:**

A karakterláncok **indexelt sorozatok**. A pozitív indexek balról 0-ról, míg a negatívak jobbról -1-gyel kezdődnek. 


```python
# Pozitív indexelés:
x = "Positive"
print(x[0])   
print(x[4])

# Negatív indexelés: 
x = "Negative"
print(x[-2])  
print(x[-6])
```

## **String Slicing** (Szeletelés)

- A karakterlánc bizonyos részeit "szeleteléssel" lehet kinyerni. 
- Szükséges a kezdő és záró indexek megadása. *Szintaxis: `string[start:end]`*


```python
x = "Hello World"

print(x[0:5])    
print(x[:2]) # első 2     
print(x[6:])  # utolsó 6  
print(x[::-1]) #visszafele
```

### **String Immutability**

A karakterláncok **megváltoztathatatlanok**, ami azt jelenti, hogy az értékük a létrehozásuk után nem módosítható. 

Egy karakterlánc bármilyen **módosítása új karakterláncot hoz létre** az eredeti helyett.

```python
x = "hello"
x = "H" + x[1:]  
print(x)
```

az eredeti "hello" karakterlánc nem módosult. Egy új karakterlánc jön létre az "H" és az x[1:] (a karakterlánc többi része) kombinálásával.

### **Karakterlánc törlése**

Egy karakterlánc egyes karakterei nem törölhetők, mivel a karakterláncok megváltoztathatatlanok. Egy teljes karakterláncváltozó azonban eltávolítható a `del` kulcsszó használatával.

```python
x = "Hi"
del x

# print(x) 
```

> Ha egy karakterlánc törlése után megpróbálunk hozzáférni, a Python `NameError` hibát generál, mivel a változó már nem létezik a memóriában.


### **Karakterlánc frissítése**
A karakterláncok nem módosíthatók közvetlenül a létrehozás után. Tehát bármilyen módosítás egy új karakterlánc létrehozását eredményezi szeleteléssel vagy olyan metódusokkal, mint a `replace()`

```python
s = "Almafa"
s1 = "a" + s[1:] 
s2 = s.replace("Alma", "Barack")

print(s1)  
print(s2)  
```

## String metódusok

- `len():` visszaadja a karakterláncban lévő karakterek teljes számát (beleértve a szóközöket és az írásjeleket).
- `upper()` és `lower()`: az összes karaktert nagybetűvé / kisbetűvé alakítja.
- `strip()` és `replace()`: a strip() eltávolítja a karakterlánc elejéről és végéről a szóközöket, a replace() egy megadott alkarakterlánc összes előfordulását egy másikkal helyettesíti.


## Stringek összefűzése és ismétlése

- összefűzés: A karakterláncok a `+ operátor` használatával kombinálhatók.
- ismétlés: Egy karakterlánc többször is megismételhető a `*` használatával.

## String végigiterálása karakterenként

```python
word = "Python"
for char in word:
    print(char)
```

## Stringek formázása

**f-string használatával:** lehetővé teszik változók és kifejezések közvetlen beszúrását egy karakterláncba `{}` zárójelek használatával.


```python
name = "Alice"
age = 20
print(f"Name: {name}, Age: {age}")
```

**format() metódus használatával:** lehetővé teszi az értékek beszúrását a karakterláncon belüli `{} helyőrzőkbe`.

```python
introduction = "My name is {} and I'm {} years old".format("Bob", 35)
print(introduction)
```

```python
text = "apple"
print("apple" in text)
print("ae" in text)
```
