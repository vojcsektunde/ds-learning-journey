# **Collections** (gyűjtemények)

Beépített konténer-adatszerkezeteket kínál, amelyek a hagyományos típusok (például dict, list, set, tuple) funkcionalitását bővítik ki.

## Counters 

a Python dict osztályának egy alosztálya a collections modulból. Főleg egy iterálható objektum *(például listák, karakterláncok vagy tuple-k)* vagy egy leképezés (szótár) elemeinek gyakoriságának számlálására szolgál. 

Tiszta és hatékony módot biztosít az elemek számlálására extra ciklusok írása nélkül, és hasznos beépített metódusokkal

```python
from collections import Counter

colors = ["yellow", "red", "red", "yellow", "green", "blue", "green", "green", "blue", "purple", "green","red"]

cnt = Counter(colors)
print(cnt)
```

### Miért használjuk a `Counter()` függvényt egy normál szótár helyett?
A számláló és a szótár hasonló, de a számláló további funkciókat biztosít kifejezetten az elemek számlálásához.

- Hiányzó kulcsok esetén 0-t ad vissza, míg egy szótár `KeyError` hibát generál.
- Kifejezetten **hashable** elemek számlálására tervezték, míg a szótár egy általános célú adatstruktúra.
- **Beépített metódusokat** biztosít, mint például a `most_common()`, `elements()`, `update()` és `subtract()` a gyakorisági műveletek egyszerűsítéséhez.
- Támogatja a számtani műveleteket, például az összeadást (+), kivonást (-), metszetet (&) és egyesítést (|).
- Kevesebb manuális kódot igényel a számlálás, mint egy szótárban lévő értékfrissítés.



```python
from collections import Counter
ctr = Counter([1, 1, 2, 2, 2, 3])
ctr2 = Counter([0, 2, 4, 6, 8, 10])
ctr3 = Counter([0, 1, 3, 5, 7, 9])

items = list(ctr.elements())
print(items)

#n leggyakoribb elem listáját és azok számát a leggyakoribbtól a legkisebbig. 
# Ha n nincs megadva, akkor a számláló összes elemét visszaadja.
common = ctr.most_common(1)
print(common)

# Kivonja az elemek számát egy másik iterálható elemből vagy leképezésből. A számok negatívak is lehetnek.
ctr.subtract([2, 2, 3])
print(ctr, "\n")


print("Összeadás:", ctr2 + ctr3)   
print("Kivonás:", ctr2 - ctr3) 
print("Metszet:", ctr2 & ctr3) 
print("Unió:", ctr2 | ctr3) 
```

## **Heap queue** (halom)

Olyan **fa-alapú adatszerkezet**, amely hatékonyan valósítja meg a prioritásos sort *(priority queue)*, ahol a legkisebb (vagy max-heap esetén a legnagyobb) elem mindig azonnal elérhető.

A legkisebb elem mindig a heap gyökércsomópontjai között, a legnagyobb elem pedig a levélcsomópontok között található.

A Python egy beépített, **`heapq`** nevű modult biztosít, amely lehetővé teszi heap sorok létrehozását és kezelését.

```python
import heapq

nums = [35, 30, 25, 40, 50, 20]
heapq.heapify(nums)
print("Heap queue:", nums)
```

### Miért van szükségünk Heap sorra?

- Hatékony módszert kínál a **prioritási sorok megvalósítására** és az elemek heap sorrendben tartására minimális kóddal és nagy teljesítménnyel.
- Hasznos olyan algoritmusokban, mint a Dijkstra- vagy a Huffman-kódolás, vagy bármilyen olyan feladatban, amely a **legkisebb elemhez való gyors hozzáférést** igényli.

- Ideális a rendezett adatok dinamikus kezelésére anélkül, hogy minden művelet után teljes rendezés történne.

### Főbb műveletek 

| Művelet | Függvény  | Leírás |
| :--- | :--- | :--- |
| **Create** | `heapq.heapify()` | Egy sima listát érvényes min-heappé  alakít át helyben. |
| **Push** | `heapq.heappush()` | Új elemet ad a heaphez, miközben fenntartja a heap tulajdonságot. |
| **Pop** | `heapq.heappop()` | Eltávolítja és visszaadja a legkisebb elemet a heapből. |
| **Peek** | `heap[0]` | Eléri a legkisebb elemet anélkül, hogy eltávolítaná azt. |
| **Push and Pop** | `heapq.heappushpop()` | Egyetlen lépésben hozzáad egy új elemet, majd eltávolítja a legkisebbet. |
| **Replace** | `heapq.heapreplace()` | Eltávolítja a legkisebb elemet, és egy új elemet szúr be egyetlen műveletben. |


Legkisebb és legnagyobb elemek megtalálása: 


```python
import heapq

nums = [35, 30, 25, 40, 50, 20]

heapq.heappush(nums, 55)
print("Smallest=",nums)

heapq.heapify(nums)

min = heapq.heappop(nums)
print(min)

maxi = heapq.nlargest(2, nums)
print("2 largest elements:", maxi)

min = heapq.nsmallest(2, nums)
print("2 smallest elements:", min)
```
