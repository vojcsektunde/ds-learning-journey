# **Collections** (gyűjtemények)

Beépített konténer-adatszerkezeteket kínál, amelyek a hagyományos típusok (például dict, list, set, tuple) funkcionalitását bővítik ki.

## Counters 

a Python dict osztályának egy alosztálya a collections modulból. Főleg egy iterálható objektum *(például listák, karakterláncok vagy tuple-k)* vagy egy leképezés (szótár) elemeinek gyakoriságának számlálására szolgál. 

Tiszta és hatékony módot biztosít az elemek számlálására extra ciklusok írása nélkül, és hasznos beépített metódusokkal

```python
from collections import Counter

colors = ["yellow", "red", "red", "yellow", "green", "blue", "green", "green", "blue", "purple", "green","red"]

cnt = Counter(colors)
print(cnt)
```

### Miért használjuk a `Counter()` függvényt egy normál szótár helyett?
A számláló és a szótár hasonló, de a számláló további funkciókat biztosít kifejezetten az elemek számlálásához.

- Hiányzó kulcsok esetén 0-t ad vissza, míg egy szótár `KeyError` hibát generál.
- Kifejezetten **hashable** elemek számlálására tervezték, míg a szótár egy általános célú adatstruktúra.
- **Beépített metódusokat** biztosít, mint például a `most_common()`, `elements()`, `update()` és `subtract()` a gyakorisági műveletek egyszerűsítéséhez.
- Támogatja a számtani műveleteket, például az összeadást (+), kivonást (-), metszetet (&) és egyesítést (|).
- Kevesebb manuális kódot igényel a számlálás, mint egy szótárban lévő értékfrissítés.



```python
from collections import Counter
ctr = Counter([1, 1, 2, 2, 2, 3])
ctr2 = Counter([0, 2, 4, 6, 8, 10])
ctr3 = Counter([0, 1, 3, 5, 7, 9])

items = list(ctr.elements())
print(items)

#n leggyakoribb elem listáját és azok számát a leggyakoribbtól a legkisebbig. 
# Ha n nincs megadva, akkor a számláló összes elemét visszaadja.
common = ctr.most_common(1)
print(common)

# Kivonja az elemek számát egy másik iterálható elemből vagy leképezésből. A számok negatívak is lehetnek.
ctr.subtract([2, 2, 3])
print(ctr, "\n")


print("Összeadás:", ctr2 + ctr3)   
print("Kivonás:", ctr2 - ctr3) 
print("Metszet:", ctr2 & ctr3) 
print("Unió:", ctr2 | ctr3) 
```

## **Heap queue** (halom)

Olyan **fa-alapú adatszerkezet**, amely hatékonyan valósítja meg a prioritásos sort *(priority queue)*, ahol a legkisebb (vagy max-heap esetén a legnagyobb) elem mindig azonnal elérhető.

A legkisebb elem mindig a heap gyökércsomópontjai között, a legnagyobb elem pedig a levélcsomópontok között található.

A Python egy beépített, **`heapq`** nevű modult biztosít, amely lehetővé teszi heap sorok létrehozását és kezelését.

```python
import heapq

nums = [35, 30, 25, 40, 50, 20]
heapq.heapify(nums)
print("Heap queue:", nums)
```

### Miért van szükségünk Heap sorra?

- Hatékony módszert kínál a **prioritási sorok megvalósítására** és az elemek heap sorrendben tartására minimális kóddal és nagy teljesítménnyel.
- Hasznos olyan algoritmusokban, mint a Dijkstra- vagy a Huffman-kódolás, vagy bármilyen olyan feladatban, amely a **legkisebb elemhez való gyors hozzáférést** igényli.

- Ideális a rendezett adatok dinamikus kezelésére anélkül, hogy minden művelet után teljes rendezés történne.

### Főbb műveletek 

| Művelet | Függvény  | Leírás |
| :--- | :--- | :--- |
| **Create** | `heapq.heapify()` | Egy sima listát érvényes min-heappé  alakít át helyben. |
| **Push** | `heapq.heappush()` | Új elemet ad a heaphez, miközben fenntartja a heap tulajdonságot. |
| **Pop** | `heapq.heappop()` | Eltávolítja és visszaadja a legkisebb elemet a heapből. |
| **Peek** | `heap[0]` | Eléri a legkisebb elemet anélkül, hogy eltávolítaná azt. |
| **Push and Pop** | `heapq.heappushpop()` | Egyetlen lépésben hozzáad egy új elemet, majd eltávolítja a legkisebbet. |
| **Replace** | `heapq.heapreplace()` | Eltávolítja a legkisebb elemet, és egy új elemet szúr be egyetlen műveletben. |


Legkisebb és legnagyobb elemek megtalálása: 


```python
import heapq

nums = [35, 30, 25, 40, 50, 20]

heapq.heappush(nums, 55)
print("Smallest=",nums)

heapq.heapify(nums)

min = heapq.heappop(nums)
print(min)

maxi = heapq.nlargest(2, nums)
print("2 largest elements:", maxi)

min = heapq.nsmallest(2, nums)
print("2 smallest elements:", min)
```

Előnyök:

- Gyors behelyezés és eltávolítás prioritással.
- Kevesebb memóriát használ, mint néhány más adattípus.
- Egyszerűen használható a heapq modullal.
- Sok esetben működik, például halomkezelésben és prioritási sorokban.


Hátrányok:
- Nem alkalmas összetett adatkezelésekre
- Nincs közvetlen hozzáférés a középső elemekhez.
- Nem lehet teljesen automatikusan rendezni az elemeket.
- Nem biztonságos több szál egyidejű használata esetén.

## **Deque** 

A deque a **Double-Ended Queue** (kétvégű sor) rövidítése. Ez egy olyan adatszerkezet, amely lehetővé teszi elemek hatékony hozzáadását és eltávolítását mindkét végről.

- Gyors behelyezést és törlést tesz lehetővé **mind az elölről, mind a hátulról.**

- Működik sorban **(queue , FIFO)** és veremként **(stack, LIFO)** is.
- Hasznos ütemezésben, csúszóablakos problémák megoldásában és valós idejű adatfeldolgozásban.

```python
from collections import deque
dq = deque(['first_name', 'last_name', 'gender', 'email'])

print(dq[0])
print(dq[-1])

dq.append('phone_num') # egy elemet ad hozzá a deque jobb végéhez.
print(dq)

dq.appendleft("age") # egy elemet ad hozzá a deque bal végéhez
print(dq)

dq.extend(["address"]) # több elemet ad hozzá a deque jobb végéhez
print(dq)

dq.extendleft(["status", "id"]) # több elemet ad hozzá a deque bal végéhez
print(dq)

print(dq.count("email")) # visszaadja, hogy egy adott elem hányszor szerepel a
print(len(dq))

dq.remove("status") # eltávolítja a megadott érték első előfordulását
print(dq)

dq.pop() # eltávolítja a jobb oldali elem értékét 
print(dq)

dq.popleft() # eltávolítja  a bal oldali elemet.
print(dq)

dq.clear() # összes elemet eltávolítja
print(dq)

dq.extend(['phone_num', 'email', 'gender','age', 'last_name', 'first_name', 'id'])
dq.reverse() # megfordítja az elemek sorrendjét 
print(dq)

dq.rotate(1) #elforgatja a deque elemeit
print(dq)
```

## **OrderedDict**

Az `OrderedDict` a Python beépített szótárának egy alosztálya, amely megjegyzi a kulcsok beszúrásának sorrendjét. 

A Python 3.7 előtt **a szótárak nem őrizték meg a beszúrási sorrendet**, ezért az OrderedDict-et használták. Még most is hasznos, mert további funkciókat biztosít:

- Sorrendfüggő egyenlőség-ellenőrzések (két azonos elemmel, de eltérő sorrendben szereplő OrderedDict nem egyenlő).
- Adatstruktúrák, például sorok, vermek vagy LRU gyorsítótárak egyszerű megvalósítása.

Példa: Ez a példa bemutatja, hogyan hozható létre egy OrderedDict, hogyan szúrhatók be elemek, és hogyan ellenőrizhető, hogy megőrzi-e a beszúrási sorrendet.

```python
from collections import OrderedDict

od = OrderedDict()
od['apple'] = 1
od['banana'] = 2
od['cherry'] = 3

print(list(od.items()))
```

### Dict VS OrderedDict

Dict: 
- Megtartja a beszúrási sorrendet
- Nem teszi lehetővé a kulcsok átrendezését
- Elemek etávolítása a végéről -  nem, csak a popitem() távolítja az utolsó elemet
- Sorrend nem számít
- Gyorsabb


OrderedDict: 
- Megtartja a beszúrási sorrendet
- Lehetővé teszi a kulcsok átrendezését
- Elemek etávolítása a végéről - popitem(last=True/False)  - mindkét oldalról
- Sorrend számít
- Lassabb egy picit


### **főbb jellemzői**

#### 1. Beszúrási sorrend megőrzése

- Az OrderedDict pontosan úgy tartja meg a sorrendet, ahogyan az elemeket hozzáadták. 

- Ez különösen hasznos olyan alkalmazásokban, mint a **JSON szerializálás** (például objektumot, tömböt vagy hashtáblát szöveges JSON formátummá alakítunk át), az űrlapmezők feldolgozása vagy a naplók megjelenítése, **ahol az elemek sorrendje szemantikai jelentéssel bír.**

#### 2. Az érték megváltoztatása nem befolyásolja a sorrendet

 - Egy meglévő **kulcs értékének módosítása nem változtatja meg a sorrendben elfoglalt helyét.** 
 
 - Ez azt jelenti, hogy az **értékek frissíthetők az eredeti kulcssorrend befolyásolása nélkül**.

#### 3. Az egyenlőségellenőrzések a sorrendet veszik figyelembe
    
- A hagyományos szócikkekkel ellentétben az OrderedDict mind **a tartalmat, mind a sorrendet ellenőrzi az egyenlőség szempontjából**, így az eltérő sorrendek egyenlőtlenné teszik őket. 
    
- Ez akkor **hasznos, ha a sorrend számít.**

#### 4. Rendezett szófordulat megfordítása

- **Nincs beépített `.reverse()` metódusa**, de a sorrendjét **meg lehet fordítani** a Python **`reversed()` függvényével** a list(od.items()) függvényen. 

- Egy új OrderedDict létrehozása ebből a fordított listából megőrzi a fordított sorrendet.    
 
 #### 5. Utolsó vagy az első elem törlése

- A `popitem()` függvény eltávolíthatja az **utolsó elemet** (last=True, alapértelmezett) **vagy az első elemet** (last=False). 

- Ezzel szemben **egy normál dict popitem()** függvénye **mindig csak az utolsó elemet** távolítja el.


#### 6. Mozgassa a kulcsokat előre vagy végére

- `A move_to_end()` metódussal az OrderedDict **rugalmasan áthelyezheti a kulcsokat.** 

- Egy adott kulcsot a szótár elejére vagy végére helyezhet anélkül, hogy törölni és újra beilleszteni kellene.


#### 7. Kulcsok törlése és újbóli behelyezése
  
  - Egy kulcs törlése és újbóli beillesztése egy OrderedDict-ben a **végére helyezi azt**, megőrizve a beillesztési sorrendet. 
  
  - Ez **hasznos a legutóbbi műveletek nyomon követéséhez** vagy a kiemelt elemek frissítéséhez.

   


```python
from collections import OrderedDict

# 1. Beszúrási sorrend megőrzése (Insertion order) - az elemek pontosan a megadott sorrendben tárolódnak el
user_tasks = OrderedDict()
user_tasks['login'] = 'Completed'
user_tasks['fetch_data'] = 'Pending'
user_tasks['render_page'] = 'Waiting'

print("1. Eredeti sorrend:")
for key, value in user_tasks.items():
    print(f"  {key}: {value}")


# 2. Érték megváltoztatása - A 'fetch_data' értékét módosítjuk, de a szótárban elfoglalt helye változatlan marad
user_tasks['fetch_data'] = 'Completed'
print("\n2. Sorrend a 'fetch_data' frissítése után (a helye nem változott):")
print(f"  {list(user_tasks.keys())}")


# 3. Egyenlőségellenőrzés sorrend alapján -  Két szótár azonos elemekkel, de eltérő sorrendben
tasks_ordered_1 = OrderedDict([('task_a', 1), ('task_b', 2)])
tasks_ordered_2 = OrderedDict([('task_b', 2), ('task_a', 1)])

print("\n3. Egyenlőségellenőrzés:")
print(f"  Megegyezik a két OrderedDict? {tasks_ordered_1 == tasks_ordered_2}") 
# Két sima dict esetén ez True lenne, de OrderedDict-nél False!


# 4. Sorrend megfordítása - A reversed() függvénnyel visszafelé is bejárhatjuk a kulcsokat
print("\n4. Kulcsok fordított sorrendben:")
for key in reversed(user_tasks):
    print(f"  {key}")


# 5. Első vagy utolsó elem törlése (Popitem) - A last=False paraméterrel a legelsőként beszúrt elemet távolítjuk el
print("\n5. Elemeltávolítás:")
first_item = user_tasks.popitem(last=False)
print(f"  Eltávolított legelső elem: {first_item}")
print(f"  Megmaradt elemek: {list(user_tasks.keys())}")


# 6. Kulcsok mozgatása az elejére vagy a végére 
# Visszahelyezzük a 'login'-t, majd a 'fetch_data'-t a sor elejére mozgatjuk
user_tasks['login'] = 'Done'  # Ez a végére kerül alapból
user_tasks.move_to_end('fetch_data', last=False)  # Átmozgatjuk a legeslegelejére

print("\n6. Sorrend a 'fetch_data' elejére mozgatása után:")
print(f"  {list(user_tasks.keys())}")


# 7. Kulcs törlése és újbóli behelyezése 
# Ha törlünk egy kulcsot, majd újra hozzáadjuk, a lista legvégére fog kerülni
del user_tasks['fetch_data']
user_tasks['fetch_data'] = 'Restarted'

print("\n7. Sorrend a 'fetch_data' törlése és újbóli beszúrása után:")
print(f"  {list(user_tasks.keys())}")
```

## Defaultdict

- a collections modul beépített dict osztályának egy alosztálya. 
- Automatikusan hozzárendel egy alapértelmezett értéket a nem létező kulcsokhoz, ami azt jelenti, hogy nem kell manuálisan ellenőrizni a hiányzó kulcsokat, és elkerülni a KeyError hibát.

### Miért van szükségünk a defaultdict() függvényre?

Egy normál szótárban egy hiányzó kulcs elérése `KeyError` hibát okoz. A defaultdict ezt a következőképpen oldja meg:

- **Hiányzó kulcsok automatikus létrehozása** alapértelmezett értékkel.
- Az ismétlődő, *„nincs kulcs a szótárban”* ellenőrzések csökkentése.
- Megkönnyíti az olyan feladatokat, mint a számolás vagy csoportosítás.
- Különösen hasznos **hisztogramokhoz, grafikonkészítéshez, szövegcsoportosításhoz és gyorsítótárazáshoz.**

### Hogyan működik? 
Amikor létrehozol egy defaultdict-et egy default:factory-t specifikálsz. (a callable).

- **Ha a kulcs létezik:** az érték returned.
- **Ha a kulcs  nem létezik:** default_factory generál egy alapértelmezett értéket.

`int: returns 0`

`list: returns []`

`str: returns ""`

Ez a mechanizmus elkerüli a hibákat éd egyszerűbbé teszi a kódnak a hiányzó kulcsok kezelését.


#### 1. Lista (list) mint alapértelmezett gyár (Default Factory)

Ha a list osztályt adjuk meg argumentumként, minden új kulcs automatikusan egy üres listával inicializálódik.

```python
from collections import defaultdict

students_grades = defaultdict(list)

# Új kulcsokhoz közvetlenül hozzáfűzhetünk (append), nem kapunk KeyError-t
students_grades['Alice'].append(5)
students_grades['Alice'].append(3)

students_grades['Bob'].append(4)

students_grades['Charlie'].append(3)
students_grades['Charlie'].append(4)

print("Students grades: ")
print(dict(students_grades))
```

>  Mivel a kulcsok hiányakor automatikusan létrejön egy üres lista [], a .append() metódus azonnal hívható anélkül, hogy ellenőriznénk a diák nevének létezését.



#### 2. Egész szám (int) mint alapértelmezett gyár

Az int() hívás alapértelmezetten 0-át ad vissza. Ez kiválóan alkalmas elemek számlálására (gyakoriság számításra).


```python
from collections import defaultdict

products = ['apple', 'milk', 'bread', 'apple', 'milk', 'apple']
cart = defaultdict(int)

for product in products:
    cart[product] += 1

print("In the cart:")
print(dict(cart))
```

> Amikor egy termék először szerepel a ciklusban, a kosar[termek] értéke automatikusan 0 lesz, amihez a += 1 rögtön hozzáad egyet.


#### 3. Karakterlánc (str) mint alapértelmezett gyár

Az str() egy üres szöveget ('') ad vissza alapértelmezett értékként, így közvetlenül végezhetünk szöveges összefűzést.

```python
from collections import defaultdict

diary = defaultdict(str)

# Szövegek hozzáfűzése napok szerint
diary['Monday'] += "It was raining."
diary['Monday'] += " It was cold."
diary['Tuesday'] += " It was sunny."

print("Weekly post:")
print(dict(diary))
```

> A hiányzó kulcsoknál nem kell KeyError-tól tartani, az üres karakterlánchoz azonnal hozzáadódik az új szövegrészlet.


#### 4. Adatok csoportosítása 

A defaultdict(list) egyik leggyakoribb gyakorlati alkalmazása az adatok kategóriák vagy tulajdonságok szerinti csoportosítása.

```python
from collections import defaultdict

cities = ["Pécs", "Győr", "Eger", "Sopron", "Baja", "Érd"]
length_of_city_names= defaultdict(list)

# Városok csoportosítása a nevük hossza alapján
for city in cities:
    length_of_city_names[len(city)].append(city)

print("Városok hossza szerint csoportosítva:")
print(dict(length_of_city_names))
```

> A szótár kulcsai a szóhosszak lesznek. Minden egyes hosszúsághoz automatikusan létrejön egy lista, amibe bekerülnek a megfelelő városok.



#### 5. A __missing__() metódus

A defaultdict a háttérben a beépített `__missing__(key)` metódust hívja meg, amikor egy kulcs nem található. Ha megadtunk egy függvényt (gyárat), ez a metódus fut le és generálja le az értéket.


```python
from collections import defaultdict

stock = defaultdict(lambda: "Out of stock")

stock["A1"] = 15 

# A __missing__ manuális tesztelése (belső működés demonstrálása)
print("Missing key B2:", stock.__missing__("B2"))

# Normal access
print("A1 shelf content:", stock["A1"])
print("X5 shelf content (automatic initialization):", stock["X5"])
```

> A __missing__() metódus belső használatra való. Éles kódban mindig a megszokott szotar[kulcs] formát vagy a szotar.get() metódust használjuk!
