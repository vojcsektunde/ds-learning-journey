# **Operators** (Operátorok)

## Arithmetic operators - Aritmetikai operátorok:

- `+`  összeadás
- `-` kivonás
- `*` szorzás
- `/` osztás
- `%` moduló (maradékos osztás)
- `//` egész osztás
- `**` hatványozás

```python
a = 10
b = 3

print("Osztás:", a / b) # 3.3333333333333335
print("Moduló:", a % b)  # 1
print("Egész osztás:", a // b) #3
```

## Assignment operators - Hozzárendelő operátorok:

- `=` sima értékadás
- `+=` értékadás összeadással `(x+=3 -> x=x+3)`
- `-=` értékadás kivonással `(x-=3 -> x=x-3)`
- `*=` értékadás szorzással
- `/=` értékadás osztással
- `%=` értékadás maradékosztásal
- `//=` értékadás egészosztással
- `**=` értékadás hatványozással

```python
x = 5
print("Kezdeti x:", x)

x += 3  #  x = x + 3
print("x += 3 ->", x)  # 8

x -= 2  #x = x - 2
print("x -= 2 ->", x)  # 6

x *= 4  #x = x * 4
print("x *= 4 ->", x)  # 24
```

## Comparison operators - Összehasonlító operátorok:

- `==` - egyenlőség
- `!=` - nem egyenlő
- `>` - nagyobb mint
- `<` - kisebb mint
- `>=` - nagyobb vagy egyenelő
- `<=` - kisebb vagy egyenlő

```python
print("5 == 5:", 5 == 5)    # True
print("5 != 5:", 5 != 5)    # False
print("10 > 5:", 10 > 5)    # True
print("3 > 5:", 3 > 5)      # False
```

## Logical operators - Logikai operátorok:

- `and` - ÉS - Csak akkor IGAZ, ha mindegyik IGAZ, egyébként HAMIS 
- `or` - VAGY - Akkor IGAZ, ha legalább az egyik IGAZ, egyébként HAMIS
- `not` - NEM (tagadás / negate) - IGAZ-ból HAMIS; HAMIS-ból  IGAZ 


> - Logikai műveletek: 
>   - Kizáró vagy (!= vagy ^),      True ^ False -> True
>   - Következtetés (not A or B),   (not True) or False -> False
>   - Ekvivalencia (==),            True == True -> True

```python
igaz_e = True
hamis_e = False

print("True and False:", igaz_e and hamis_e)  # False
print("True or False:", igaz_e or hamis_e)    # True
print("not True:", not igaz_e)                # False
```

## Identity operators - Identitás operátorok (azonosságot vizsgál):

- `is` - értéke `True, if x is y`
- `is not` - értéke `True, if x is not y`

```python
cart = ["apple", "lemon", "pear"]
otherCart = ["apple", "lemon", "pear"]
favoriteCart = cart

print("Same items?", cart == otherCart)        
print("Same cart?", cart is otherCart)  # False, mert másik memóriacím 
print("Different cart?", cart is not otherCart)   

print("Same as the fav cart?", cart is favoriteCart)        
```


# Membership operators - Tagsági operátor:

- `in` - valami része-e a másiknak, értéke `True, if x in y`
- `not in` - valami nem része-e a másiknak, tagadás, értéke `True if x not in y`

```python
text = "Hello Python!"
numbers = [2, 4, 6, 8, 10]

print("'Python' in text:", "Python" in text)      # True
print("'Java' not in text:", "Java" not in text)  # True
print("3 in numbers' list:", 3 in numbers)   # False
```


# Bitwise operators - Bitműveleti operátorok:
    Jogosultságkezelésnél (bitmaszkolás), mikrovezérlők programozásánál, adattömörítésnél, tömörített fájlok vagy képek beolvasásakor.

- `&` - bitenkétni ÉS művelet - Az eredmény bit csak akkor lesz 1, ha mindkét eredeti bit 1-es.
- `|` - bitenkénti VAGY művelet - Az eredmény bit 1 lesz, ha a két bit közül bármelyik (vagy mindkettő) 1-es.
- `^` - bitenkénti XOR művelet - Az eredmény bit akkor lesz 1, ha a két bit különbözik egymástól. Egyezésnél 0.
- `~` - bitenkénti NEM művelet - 0-ból 1-et csinál, az 1-ből pedig 0-t.
- `>>` - jobbra léptetés - A szám összes bitjét eltolja jobbra a megadott mértékkel. A szélső jobb oldali bit ekkor „leesik” és elveszik.
- `<<` - balra léptetés - A szám összes bitjét eltolja balra a megadott mértékkel. A megüresedett helyre jobbról egy 0 kerül.

```python
# 5 = 0101, 3 = 0011
n1 = 5 
n2 = 3 

print("Bitenkénti ÉS (5 & 3):", n1 & n2)         # 1  (0001)
print("Bitenkénti VAGY (5 | 3):", n1 | n2)       # 7  (0111)
print("Bitenkénti XOR (5 ^ 3):", n1 ^ n2)       # 6  (0110)

# A bitenkénti balra léptetés megszorozza a számot kettő hatványával (mivel 1-gyel léptettünk, 2-vel szoroz). Ezért lesz az 5-ből
print("Balra léptetés (5 << 1):", n1 << 1)     # 10 (1010 - a bitek balra tolódtak, értéke megduplázódott) 

print("Jobbra léptetés (5 >> 1):", n1 >> 1)    # 2  (0010 - a bitek jobbra tolódtak)
```

# **Selection** (Szelekció, elágazás)

Az **elágazások** olyan **vezérlési struktúrák** a programozásban amelyek lehetővé teszik, hogy bizonyos feltételek alapján utasításokat hajtsanak végre a programok. Ezeket a feltételeket az **“if”, “elif” és “else”** kulcsszavak segítségével hozzuk létre a Pythonban.  Segítségével adatokat tudunk ellenőrizni, kezelni a hibajelzéseket stb..

* Az **“if”** (ha akkor) kulcsszó egy feltételt vizsgál, és **ha az igaz, akkor a blokkját hajtja végre, ellenkező esetben átugorja.** 

* Az **“elif”** kulcsszó hasonlóan működik, mint az “if”, azonban abban az esetben használjuk, **ha az előző feltétel hamis volt. Többszörös elágazásoknál használjuk.**

* Az **“else”** (különben) kulcsszó akkor hajtódik végre, **ha az összes korábbi feltétel hamis.**

> Túlzott használatuk nehezebbé tehetik a programok olvashatóságát és karbantarthatóságát ezért érdemes ésszerűen alkalmazni. 

Példa, többszörös elágazásra:

```python
# Kérjünk be egy számot a billentyűzetről, és vizsgáljuk meg, hogy kisebb, nagyobb, vagy egyenlő-e öttel. 
x = int(input("Give me a number: "))
if x > 5:
    print(str(x)+" > 5") 
elif x < 5:
    print(str(x)+" < 5")
else:
    print(str(x)+" = 5")
```

Elágazások logikai feltétellel:

```python
# Csak akkor írjuk ki a szöveget ha a beolvasott szám 1-10 között található, máskülönben rossz input!
x = int(input("Give me a number between 1 and 10: "))
if (x>=1 and x<=10):
    print("Right input!")
else:
    print("Wrong input!")
```

### "Short hand", Rövidített if
 Ha csak egy sorból áll a feltétel parancsa, akkor írhatjuk így is:

```python
age = 20 
if age >= 18: print("Eligible to Vote.")
```

### "Short hand", Rövidített if ...else:
Ilyenkor először végrehajtjuk az if parancsát, majd a vizsgálat után jöhet az else

```python
age = 16
status = "Eligible" if age >= 18 else "Not Eligible"
print(status + " to Vote.")
```

###  Nested (Egymásba ágyazott) if ... else:

```python
age = 12
is_member = True

if age < 18:
    if is_member:
        print("30% minor discount!")
    else:
        print("10% minor discount.")
else:
    print("Not eligible for a minor discount.")
```

###  Match-Case Statement 

Hasonlít a más programozási nyelvekben található switch-case utasításhoz. Akkor érdemes használni, amikor a kód összetett adatstruktúrák formáját (mintáját) kell ellenőrizni, és nem csak egyszerű logikai igaz/hamis feltételeket akarsz kiértékelni.

Olvashatóbb, ha sok különböző eset van, vagy egymásba ágyazott listákat/szótárakat kell vizsgálni.

- `match:` Megadja azt a kifejezést (értéket, változót), amit meg akarunk vizsgálni.
- `case:` Megadja a mintát, amihez hasonlítjuk.
- `case _:` Az úgynevezett "wildcard" (mindent lefedő) ág. Akkor fut le, ha egyik előző minta sem volt igaz (olyan, mint az else)

```python
status = 404

match status:
    case 200:
        print("Success")
    case 404:
        print("Page not found")
    case 500:
        print("Server error")
    case _:
        print("Unknown status code")
```

## Iteráló ciklus: For
Egy iterálható objektumon megy végig (iterál), például egy lista elemein, egy szöveg karakterein vagy egy számsorozaton. Az **iteráció számát a lista elemeinek száma határozza meg**, így a ciklus addig fog futni, amíg az összes elemet be nem járta.

Tehát akkor használjuk, **ha tudjuk, hányszor fut le**, vagy egy listán halad végig.


Pythonban leggyakrabban a `range(start, stop, step)` függvényt használják a megvalósításához. A `start` és `step paraméterek` opcionálisak. 

A `len()` függvény visszaadja a lista elemeinek teljes számát, hosszát.


```python
n = 5
for i in range(1, n+1):
    print(i)
```

```python
list(range(5)) # Végérték, 0-tól indul az indexelés alapértelmezetten. 
```

```python
list(range(1, 11)) # Kezdő- és végérték megadása.

#Az utolsó elem 10, mert a végpont nincs a sorozatban.
```

```python
list(range(0, 11, 2)) # Kezdő- és végérték megadása lépésközzel 
```
