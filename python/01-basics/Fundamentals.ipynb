# **Variables** (Változók)

## 1. Változók elnevezése



* **NEM kezdődhet számmal!**
* **CSAK kis- és nagybetűket, valamint számokat és _ karaktert** tartalmazhat!
* **Case-sensitive** (megkülönbözteti a kis- és nagybetűt): **a != A**
* Utaljon a változó **feladatára**! (Clean Code)
* **Definiált kulcsszavak (if, else, for...) nem lehetnek** változó nevek!
* **Kezdődjön kisbetűvel!** Igaz, a nagybetűs változónevek is hiba nélkül lefutnak, de **PEP8 - Style Guide** szerint ez számít egységesnek.
> Tartalmazhat ékezetes betűt, de globális vagy hosszútávú projektekben **angol elnevezéseket célszerű** használni.

### **Példák helytelen változó elnevezésre:**

```python
1myvar = "First wrong variable name"
```

```python
my-var = "Second wrong variable name"
```


### **Több szóból álló változók elnevezése:**

```python
# Az első szó kisbetűvel kezdődik, minden további szó pedig nagybetűvel (camelCase)
firstNumber = 1
```

```python
# Minden szó nagybetűvel kezdődik (PascalCase)
FirstNumber = 1
```

```python
# Minden szót aláhúzás jellel választok el (snake_case)
first_number = 1
```

### **Több változó létrehozása, egyszerre:**

```python
# Több értékkel
x, y, z = 2, 3, 4 
print(x, y, z) 
```

```python
# Ugyanazzal az értékkel
x = y = z = 5 
print(x, y, z) 
```
