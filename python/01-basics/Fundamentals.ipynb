# **Data Types** (Adattípusok)

A Python az alábbi főbb adattípusokat kezeli:

* **Számértékek (Numerikus):** 
    * `int` (egész)
    * `float` (lebegőpontos) 
    * `complex` (komplex)

* **Sorozat típusok (Szekvenciák):** 
    * `list []` (rendezett és módosítható elemekből álló lista) 
    * `tuple ()` (nem módosítható - immutable)
    * `range` (memóriahatékony, immutable, függvényként használják cikusokban `pl. range(start, stop, step)`)
    * **Szöveg:** `str` (karakterláncok, szöveg)

* **Logikai típus (Bool):** `boolean` (False)

* **Szótár/Könyvtár (Dictionary):** `dict` (Kulcs-érték pár alapú adatszerkezet, a kulcsok egyediek.)
    - Struktúrált adatok, objektumok vagy rekordok tárolására alkalmas. 
    - Számlálásra, statisztika készíéshez is használják, JSON formátum

* **Halmazok:** (rendezetlen, duplikált elem nélküli egyedi elemek gyűjteménye )
    * `set` (változtatható (mutable): pl. add() és remove() metódusokkal)
    * `frozenset`  (fix értékeknél, amelyek nem változhatnak)

* **Bináris típusok:** 
    * `bytes` - Bájtsorozat, immutable, nyers bájtokat tárol. Bináris adatok beolvasásakor, ha biztosan nem kell módosítani őket. 
    * `bytearray` - Bájttömb, mutable, bytes testvére, de ennek a tartalmát szabadon módosíthatod, pl. alacsony szintű adatfeldolgozás vagy képmanipuláció során
    * `memoryview` - Memórianézet -Nem tárol saját adatot; ehelyett egy ablakot nyit egy már létező bináris objektum (pl. bytes vagy bytearray) memóriájára. Slicing-nél hasznos, rendkívül gyors és memóriatakarékos nagy fájlok feldolgozásánál.

* **Típus nélküli típus:** `None` (üres érték **NEM 0 ÉS NEM ""**) 
    * Változók helyzőrzöjeként
    * Hianyzó adatok jelölésére pl .API-k esetén 
    * Függvények implicit visszatérési értékeként.

# **Operators** (Operátorok)

## Arithmetic operators - Aritmetikai operátorok:

- `+`  összeadás
- `-` kivonás
- `*` szorzás
- `/` osztás
- `%` moduló (maradékos osztás)
- `//` egész osztás
- `**` hatványozás

```python
a = 10
b = 3

print("Osztás:", a / b) # 3.3333333333333335
print("Moduló:", a % b)  # 1
print("Egész osztás:", a // b) #3
```

## Assignment operators - Hozzárendelő operátorok:

- `=` sima értékadás
- `+=` értékadás összeadással `(x+=3 -> x=x+3)`
- `-=` értékadás kivonással `(x-=3 -> x=x-3)`
- `*=` értékadás szorzással
- `/=` értékadás osztással
- `%=` értékadás maradékosztásal
- `//=` értékadás egészosztással
- `**=` értékadás hatványozással

```python
x = 5
print("Kezdeti x:", x)

x += 3  #  x = x + 3
print("x += 3 ->", x)  # 8

x -= 2  #x = x - 2
print("x -= 2 ->", x)  # 6

x *= 4  #x = x * 4
print("x *= 4 ->", x)  # 24
```

## Comparison operators - Összehasonlító operátorok:

- `==` - egyenlőség
- `!=` - nem egyenlő
- `>` - nagyobb mint
- `<` - kisebb mint
- `>=` - nagyobb vagy egyenelő
- `<=` - kisebb vagy egyenlő

```python
print("5 == 5:", 5 == 5)    # True
print("5 != 5:", 5 != 5)    # False
print("10 > 5:", 10 > 5)    # True
print("3 > 5:", 3 > 5)      # False
```

## Logical operators - Logikai operátorok:

- `and` - ÉS - Csak akkor IGAZ, ha mindegyik IGAZ, egyébként HAMIS 
- `or` - VAGY - Akkor IGAZ, ha legalább az egyik IGAZ, egyébként HAMIS
- `not` - NEM (tagadás / negate) - IGAZ-ból HAMIS; HAMIS-ból  IGAZ 


> - Logikai műveletek: 
>   - Kizáró vagy (!= vagy ^),      True ^ False -> True
>   - Következtetés (not A or B),   (not True) or False -> False
>   - Ekvivalencia (==),            True == True -> True

```python
igaz_e = True
hamis_e = False

print("True and False:", igaz_e and hamis_e)  # False
print("True or False:", igaz_e or hamis_e)    # True
print("not True:", not igaz_e)                # False
```

## Identity operators - Identitás operátorok (azonosságot vizsgál):

- `is` - értéke `True, if x is y`
- `is not` - értéke `True, if x is not y`

```python
cart = ["apple", "lemon", "pear"]
otherCart = ["apple", "lemon", "pear"]
favoriteCart = cart

print("Same items?", cart == otherCart)        
print("Same cart?", cart is otherCart)  # False, mert másik memóriacím 
print("Different cart?", cart is not otherCart)   

print("Same as the fav cart?", cart is favoriteCart)        
```


# Membership operators - Tagsági operátor:

- `in` - valami része-e a másiknak, értéke `True, if x in y`
- `not in` - valami nem része-e a másiknak, tagadás, értéke `True if x not in y`

```python
text = "Hello Python!"
numbers = [2, 4, 6, 8, 10]

print("'Python' in text:", "Python" in text)      # True
print("'Java' not in text:", "Java" not in text)  # True
print("3 in numbers' list:", 3 in numbers)   # False
```


# Bitwise operators - Bitműveleti operátorok:
    Jogosultságkezelésnél (bitmaszkolás), mikrovezérlők programozásánál, adattömörítésnél, tömörített fájlok vagy képek beolvasásakor.

- `&` - bitenkétni ÉS művelet - Az eredmény bit csak akkor lesz 1, ha mindkét eredeti bit 1-es.
- `|` - bitenkénti VAGY művelet - Az eredmény bit 1 lesz, ha a két bit közül bármelyik (vagy mindkettő) 1-es.
- `^` - bitenkénti XOR művelet - Az eredmény bit akkor lesz 1, ha a két bit különbözik egymástól. Egyezésnél 0.
- `~` - bitenkénti NEM művelet - 0-ból 1-et csinál, az 1-ből pedig 0-t.
- `>>` - jobbra léptetés - A szám összes bitjét eltolja jobbra a megadott mértékkel. A szélső jobb oldali bit ekkor „leesik” és elveszik.
- `<<` - balra léptetés - A szám összes bitjét eltolja balra a megadott mértékkel. A megüresedett helyre jobbról egy 0 kerül.

```python
# 5 = 0101, 3 = 0011
n1 = 5 
n2 = 3 

print("Bitenkénti ÉS (5 & 3):", n1 & n2)         # 1  (0001)
print("Bitenkénti VAGY (5 | 3):", n1 | n2)       # 7  (0111)
print("Bitenkénti XOR (5 ^ 3):", n1 ^ n2)       # 6  (0110)

# A bitenkénti balra léptetés megszorozza a számot kettő hatványával (mivel 1-gyel léptettünk, 2-vel szoroz). Ezért lesz az 5-ből
print("Balra léptetés (5 << 1):", n1 << 1)     # 10 (1010 - a bitek balra tolódtak, értéke megduplázódott) 

print("Jobbra léptetés (5 >> 1):", n1 >> 1)    # 2  (0010 - a bitek jobbra tolódtak)
```
