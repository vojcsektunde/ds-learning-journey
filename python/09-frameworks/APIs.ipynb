# API-kérelmek készítése Pythonban

Az **API**-k határozzák meg azokat a módszereket és adatformátumokat, amelyeket az alkalmazások használhatnak információk kérésére és cseréjére. Adatok webszerverről való lekéréséhez egy kliensalkalmazás **kezdeményez egy kérést, és a szerver válaszol a kért adatokkal.**

Az API-kkal való munkához néhány eszközre van szükség, például a `request` modulra, amelyeket először telepítenünk kell a rendszerünkbe.
> `pip install requests`

**Státuszkódok**: 

- `200 OK:` Sikeres kérés (pl. GET vagy PUT esetén, visszaadott tartalommal).
- `201 Created:` Sikeres létrehozás (gyakran POST után, jelzi az új erőforrás elérhetőségét).
- `304 Not Modified:` Az adat nem változott a legutóbbi lekérés óta; 
- `400 Bad Request:` Hibás szintaxis, érvénytelen JSON vagy hiányzó paraméterek.
- `401 Unauthorized:` Hiányzó vagy érvénytelen hitelesítési adatok (Authentication).
- `403 Forbidden:` A kliens azonosítva van, de nincs joga az adott erőforráshoz (Authorization).
- `404 Not Found:` A keresett erőforrás nem található ezen a címen.
- `405 Method Not Allowed:` Az adott HTTP metódus (pl. DELETE) nem engedélyezett az erőforráshoz.
- `500 Internal Server Error: `Általános, váratlan szerverhiba.
- `503 Service Unavailable:` A szerver ideiglenesen elérhetetlen (pl. túlterheltség vagy karbantartás miatt).
- `504 Gateway Timeout:` Az átjáró (gateway) időtúllépés miatt nem kapott választ a háttérszervertő


```python
import requests

baseURL = 'https://dummyjson.com/products'

def getProductInfo(name: str) -> dict:
    
    search_url = f"{baseURL}/search"
    params = {'q': name}
    
    response = requests.get(search_url, params=params)
    response.raise_for_status()
    return response.json()

 #Segédfüggvény egy adott tulajdonság (pl. price, rating) alapján történő szűréshez
def _getProductByMetric(name: str, metric: str, selector_func) -> dict:
   
    data = getProductInfo(name)
    products = data.get('products', [])
    if not products:
        raise ValueError(f"Nincs találat a következőre: '{name}'")
    return selector_func(products, key=lambda x: x[metric])



def getMostExpensiveProduct(name: str) -> dict:
    return _getProductByMetric(name, metric='price', selector_func=max)

def getWorstRatedProduct(name: str) -> dict:
    return _getProductByMetric(name, metric='rating', selector_func=min)


try:
    query = "phone"
    
    most_expensive = getMostExpensiveProduct(query)
    worst_rated = getWorstRatedProduct(query)
    all_results = getProductInfo(query)
    
    print(f"Összes találat száma ({query}): {all_results.get('total')} db.")
    print(f"Legdrágább: {most_expensive['title']} - ${most_expensive['price']}")
    print(f"Legrosszabb értékelés: {worst_rated['title']} - Rating: {worst_rated['rating']}")
    
except requests.exceptions.RequestException as e:
    print(f"Hiba történt a lekérdezés során: {e}")
```

> A példában felhasznált API nem az én tulajdomonban áll! Forrás: https://dummyjson.com/docs/products
