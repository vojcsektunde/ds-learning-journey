#### REST API FastAPI használatával.

Az **API**-k határozzák meg azokat a módszereket és adatformátumokat, amelyeket az alkalmazások használhatnak információk kérésére és cseréjére. Adatok webszerverről való lekéréséhez egy kliensalkalmazás **kezdeményez egy kérést, és a szerver válaszol a kért adatokkal.**

 **REST** *(Representational State Transfer)* egy architektúrális stílus webszolgáltatások tervezéséhez. Olyan alapelveket határoz meg, amelyek egyszerűvé, szabványosítottá és skálázhatóvá teszik **a kliensek és a szerverek közötti kommunikációt.** minden egyes klienstől érkező kérés tartalmazza a feldolgozásához szükséges összes információt, anélkül, hogy a korábbi interakciókra támaszkodna. 

- **Nagy teljesítmény:** A webes kérések kezeléséhez a Starlette-re, az adatellenőrzéshez pedig a Pydanticra épül, így rendkívül gyors.

- **Fejlesztőbarát:** Automatikusan ellenőrzi az adattípusokat és validálja a bemeneteket, csökkentve a hibákat és időt takarítva meg.

- **Interaktív dokumentáció**: Automatikusan generálja az API dokumentációt a `/docs` (Swagger UI) és a `/redoc` könyvtárban , így a fejlesztők **extra beállítás nélkül tesztelhetik és felfedezhetik a végpontokat.**


 - **Egyedi URL-címek (végpontok):** Minden erőforrás egy adott URL-címen keresztül azonosítható és érhető el.
- **Szabványos HTTP metódusok (CRUD):** Az erőforrásokon végzett műveleteket jól definiált metódusok segítségével hajtják végre
    - **POST**: Új adatok létrehozása - *Create*
    - **GET**: Adatok lekérése - *Read*
    - **PUT**: Meglévő adatok frissítése - *Update*
    - **DELETE** : Adatok eltávolítása - *Delete*



```python
from fastapi import FastAPI
from fastapi.testclient import TestClient

app = FastAPI()

@app.post("/endpoint")
def function_name(data: dict):
    return {"message": "Success", "data": data}


client = TestClient(app)

response = client.post("/endpoint", json={"uzenet": "Hello world!"})

print("Státuszkód:", response.status_code)
print("Válasz JSON:", response.json())
```

#### REST API-k típusai
Csoportosíthatók aszerint, hogy kik használhatják őket, és hogyan vannak kialakítva:

- **Nyílt (Public):** Bárki által szabadon elérhető. Pl. Időjárás-alkalmazások által használt nyilvános időjárás-előrejelző API.

- **Belső (Internal):** Csak egy cégen vagy szervezeten belül használják a belső folyamatokhoz. Pl. Egy vállalat saját HR- és bérszámfejtő rendszere.

- **Partner API:** Kizárólag szerződött partnerek érhetik el, szigorú hitelesítéssel. Pl. Stripe fizetési kapu integrálása egy webshopba.

- **Összetett (Composite):** Több különböző adatforrást vagy kérést von össze egyetlen hívásba. Pl. Egy utazási oldal, ami egy gombnyomásra egyszerre hozza le a repülőjegyet, a szállodát és az autóbérlési ajánlatokat.

### Más API-k

Bár a REST az egyik legnépszerűbb stílus, más megközelítések is léteznek az API-k építésére. Mindegyiknek megvannak a maga erősségei a felhasználási esettől függően:

- **SOAP**: Régebbi, nagyon szigorú, **XML-alapú és biztonságos szabvány.** Pl. Bankközi tranzakciók és vállalati pénzügyi rendszerek.

- **GraphQL**: Modern lekérdezési nyelv, amivel **pontosan annyi adatot kérsz le, amennyi kell** (se többet, se kevesebbet). Pl. Közösségi média feedek (pl. Facebook/Instagram), ahol a mobilapp pontosan a képernyőn látható posztok adatait kéri el.

- **gRPC**: Google által fejlesztett, rendkívül gyors, **bináris alapú** protokoll. Mikroszolgáltatások közötti **valós idejű kommunikációra tervezték.** pl. Nagy rendszerek mikroszolgáltatásai *(streaming szolgáltatók háttérfolyamatai)* közötti villámgyors kommunikáció.
