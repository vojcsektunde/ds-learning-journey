# Python frameworks

A keretrendszerek újrafelhasználható eszközöket és koponenseket kínálnak amelyek gyorsítják és leegyszerűsítik az alkalmazásfejlesztést.

- **Webfejlesztésre, API fejlesztésre** és **adatvezérelt alkalmazásokhoz** (adatok tárolása, megjelenítése manipulálása) használják. 
- Csökkenti az ismétlődő fejlesztési munkát
- Egyszerűbb tesztelés, telepítés, skálázás
- Jobban karbantartható


## Top 3 keretrendszer

### **Django**

Webfejlesztésen alapuló keretrendszer, gyors, biztonságos és karbantartható webapplikációk készítésére. Nagy, vállalati és E-kereskedelmi webes alkalmazásokhoz, tartalomkezelő rendszerek építésére használják. 

- **Built-in ORM** (objektum-relációs leképzés) - objektum-orientált nyelven, relációs adatbázisokkal kapcsolatban írt kód.
- **Admin interface**
- **Authentication, authorization** - hitelesítés
- **URL** routing
- **Biztonsági** funkciók
- **Adatbázis migrációk**


### **Flask**

A **WSGI** *(Web Server Gateway Interface)* szabvány a webalkalmazások / **keretrendszerek** és a **webszerverek közötti kommunikációhoz** szükséges. Bővítményeken keresztül új funkciókat vezethetünk be. 


### **FastAPI**
Egy modern Python keretrendszer, amelyet**elsősorban API-k fejlesztésére** terveztek. Python type-hinteket használ validációhoz, szerializációhoz és automatikus API-dokumentációhoz, támogatja az aszinkron alkalmazásfejlesztést is.

> Lehetővé teszi a függvényargumentumok várható típusainak megadását. 
> A type-hintek javítják a kód olvashatóságát, futási idő előtti hibák észletlését biztosítják. 

Változó és függvény Type Hint-ek:


```python
age: int = 25
def greet(name: str) -> str:
    return f"Hello, {name}!"
```

- **Nagy teljesítményű API-k fejlesztése**
- Python **type-hint támogatás** 
- Automatikus API dokumentáció
- **Adatvalidáció**
- **Async** és **await** támogatás
- **OpenAPI** és **JSON** séma integráció
