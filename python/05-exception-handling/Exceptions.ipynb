# **Exception Handling** (Kivételkezelés)

A kivételkezelés lehetővé teszi, hogy a programunk **ellenőrzött módon kezelje a futás közbeni váratlan hibákat**, ahelyett, hogy hirtelen összeomlana. Ezzel növelhetjük az alkalmazások megbízhatóságát, és biztosíthatjuk, hogy a program futása ne szakadjon meg váratlanul.

Segítségével olyan futásidejű hibákat kezelhetünk, mint a **hibás felhasználói bemenet**, **hiányzó fájlok**, **nullával való osztás vagy a típuseltérések**.

---

## 1. Alapvető Példa
Az alábbi egyszerű példa bemutatja, hogyan kaphatunk el egy kivételt és hogyan kezelhetjük azt elegánsan:

```python
number = 10
try:
    result = number / 0
except ZeroDivisionError:
    print("Can't be divided by zero!")
```

**Magyarázat:** A szám 0-val való osztása egy `ZeroDivisionError` kivételt vált ki. A `try` blokk tartalmazza a kockázatos kódot, az `except` blokk pedig elkapja a hibát, így a program leállása helyett egy biztonságos hibaüzenet jelenik meg.

---

## 2. Szintaxis és Kulcsszavak
A Python négy fő kulcsszót biztosít a kivételek kezelésére, melyek mindegyike egyedi szerepet játszik:
* **`try`**: A kockázatos kódot futtatja, ami hibát okozhat.
* **`except`**: Elkapja és kezeli a hibát, ha az bekövetkezik.
* **`else`**: Csak akkor fut le, ha a `try` blokkban nem történt kivétel.
* **`finally`**: Mindenképpen lefut, függetlenül a hibáktól (hasznos erőforrások lezárásakor).

**Példa, mind a négy kulcsszó használatára:**


```python
try:
    denominator = 0
    result = 100 / denominator
except ZeroDivisionError:
    print("You can't divide by zero!")
except ValueError:
    print("Enter a valid number!")
else:
    print("Result is", result)
finally:
    print("Execution complete.")
```

---


## 3. Kivételek elkapása
Hatékonyabban kezelhetjük a hibákat, ha pontosan megadjuk a várt kivételek típusait. Ez biztonságosabbá teszi a kódot és megkönnyíti a hibakeresést (debugging).

### Különböző hibák egyedi kezelése
Ez a kód a `ValueError` és a `ZeroDivisionError` hibákat eltérő üzenetekkel kezeli:


```python
try:
    # Ez ValueError-t fog okozni, mert a "str" nem alakítható egésszé
    user_input = int("str")
    inverse_value = 1 / user_input
except ValueError:
    print("Not Valid!")
except ZeroDivisionError:
    print("Zero has no inverse!")
```

### Több kivétel kezelése egy blokkban
Ha több hibát is ugyanúgy szeretnénk kezelni, csoportosíthatjuk őket egy zárójeles felsorolásban:


```python
items = ["10", "twenty", 30]
try:
    # A "twenty" szöveg nem alakítható számmá
    total_sum = int(items[0]) + int(items[1])
except (ValueError, TypeError) as error:
    print("Error:", error)
except IndexError:
    print("Index out of range.")
```

### A Catch-All és veszélyei
Ha csak magát az `except` kulcsszót használjuk hibaosztály nélkül, az minden hibát elkap. 

**Kockázat:** Ez **megnehezíti a hibakeresést!!**, mivel elrejti a hiba valódi típusát és okát. Csak végső biztonsági hálóként szabad használni!

```python
try:
    # Szöveg osztása számmal TypeError-t okoz
    calculation = "100" / 20
except ArithmeticError:
    print("Arithmetic problem.")
except:
    print("Something went wrong!")
```

---

## 4. Kivételek 'kiváltása' 
A `raise` kulcsszó használatával mi magunk is kiválthatunk egy adott típusú kivételt, ha a program logikája azt megköveteli. 

- Olyan esetekben használjuk, ha matematikailag bár helyes, de a sajátos logika alapján nem. 
- Pl. *Banki átutalás:* 100 - 500 = - 400, de **negatív értéket nem lehet átutalni.** Így matematikailag helyes, de logikailag nem. 

Ugyanez érvényes az életkorra is: 


```python
def set_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative.")
    print(f"Age set to {age}")

try:
    set_age(-5)
except ValueError as error:
    print(error)
```

---



## 5. **Errors VS Exceptions** 
Bár mindkettő problémát jelent a programban, súlyosságuk és kezelésük időpontja eltér:

* **Error (Hiba):** A program logikájának vagy szintaxisának hibái (pl. `SyntaxError`). Jellemzően a kód **beolvasásakor/fordításakor** (compile time) jelentkeznek, és megakadályozzák a program elindulását.

* **Exception (Kivétel):** Olyan problémák, amelyek a program **futása közben** (runtime) jelentkeznek (pl. hibás bemenet, hiányzó fájl). Ezek kezelhetők a `try-except` blokkok segítségével, így a program tovább futhat.


# Built-in Exceptions
A Python számos beépített kivételt biztosít, amelyek mindegyike egy-egy konkrét hibatípus jelzésére szolgál, segítve ezzel a hatékonyabb hibakeresést. A beépített kivételek listáját a `locals()` függvénnyel is megtekinthetjük:


```python
locals()['__builtins__']
```

## 1. BaseException

Ez az osztály a Python kivétel-hierarchiájának a gyökere. Minden más kivétel közvetlenül vagy közvetve ebből származik. Bár közvetlenül ritkán használjuk a kódban, fontos alapját képezi a Python hibakezelési rendszerének.

```python
try:
    raise BaseException("This is a BaseException")
except BaseException as error:
    print(error)
```

## 2. Exception

Az `Exception` osztály az alapja minden olyan kivételnek, amely nem a programból való kilépést szolgálja. Legtöbbször ezt az osztályt kapjuk el az általános hibakezelő kódokban, amikor nem egy konkrét hibatípust célzunk meg.

```python
try:
    raise Exception("This is a generic exception")
except Exception as error:
    print(error)
```

## 3. ArithmeticError

Az `ArithmeticError` az alaposztálya minden olyan hibának, amely matematikai műveletekkel kapcsolatos. Általában nem közvetlenül váltjuk ki, de lehetőséget ad arra, hogy egyetlen blokkban kezeljünk minden matekkal kapcsolatos hibát.

```python
try:
    raise ArithmeticError("Arithmetic error occurred")
except ArithmeticError as error:
    print(error)
```

## 4. ZeroDivisionError
Akkor fordul elő, amikor egy számot nullával próbálunk elosztani. Mivel a nullával való osztás a matematikában nem értelmezett, a Python ezzel a kivétellel jelzi a hibát.

```python
try:
    result = 10 / 0
except ZeroDivisionError as error:
    print(error)
```

 ## 5. OverflowError

 Akkor jelentkezik, ha egy numerikus művelet eredménye túl nagy ahhoz, hogy a Python ábrázolni tudja. Bár a Python a nagy egészeket rugalmasan kezeli, bizonyos lebegőpontos műveletek kiválthatják ezt a hibát.

```python
import math

try:
    result = math.exp(1000)
except OverflowError as error:
    print(error)
```

## 6. FloatingPointError

Akkor fordul elő, ha egy lebegőpontos számítás meghiúsul. Alapértelmezés szerint a Python a legtöbb lebegőpontos problémát csendben kezeli. A hibajelzést azonban explicit módon engedélyezhetjük olyan külső könyvtárakkal, mint a NumPy.

```python
import numpy as np
np.seterr(all='raise')

try:
    np.divide(1, 0)
except FloatingPointError as error:
    print("FloatingPointError caught:", error)
```

## 7. AssertionError

Akkor váltódik ki, ha egy `assert` állítás hamisnak bizonyul. Az `assert` kulcsszót leggyakrabban hibakeresésre vagy feltételek tesztelésére használjuk a kódban.


```python
try:
    assert 1 == 2, "Assertion failed"
except AssertionError as error:
    print(error)
```

## 8. AttributeError

 Akkor jelentkezik, amikor egy objektum olyan attribútumára vagy metódusára próbálunk hivatkozni, amely az adott objektumnál nem létezik.

```python
class MyClass:
    pass

obj = MyClass()

try:
    obj.some_attribute
except AttributeError as error:
    print(error)
```

## 9. IndexError

Akkor fordul elő, ha egy listát vagy bármilyen más sorozatot olyan indexszel próbálunk elérni, ami a megengedett tartományon kívül esik.


```python
my_list = [1, 2, 3]

try:
    element = my_list[5]
except IndexError as error:
    print(error)
```

## 10. KeyError

Akkor jelentkezik, amikor egy szótárban olyan kulcsra próbálunk hivatkozni, amely nem található meg benne.


```python
my_dict = {"key1": "value1"}

try:
    val = my_dict["key2"]
except KeyError as error:
    print(error)
```

## 11. MemoryError

Akkor fordul elő, ha a Python nem tud elegendő memóriát lefoglalni egy művelethez. Ez általában akkor történik, ha extrém méretű adatszerkezeteket próbálunk létrehozni.


```python
try:
    my_large_list = [1] * (10**10)
except MemoryError as error:
    print("MemoryError caught!")
```

## 12. NameError

Akkor jelentkezik, amikor olyan változó- vagy függvénynévre hivatkozunk, amely még nem lett definiálva a kódban.

```python
try:
    print(undefined_variable)
except NameError as error:
    print(error)
```

## 13. OSError, IOError & FileNotFoundError
Rendszerszintű műveletek meghiúsulásakor váltódnak ki. A `FileNotFoundError` az `OSError` egy alosztálya, amely kifejezetten akkor jelentkezik, ha a megadott fájl nem létezik.


```python
try:
    open("non_existent_file.txt")
except FileNotFoundError as error:
    print("FileNotFoundError caught:", error)
except OSError as error:
    print("OSError caught:", error)
```

| :--- | :--- |
| **TypeError** | Akkor váltódik ki, ha egy műveletet vagy függvényt nem megfelelő típusú objektumra alkalmazunk (pl. karakterlánc és egész szám összeadása). |

| **ImportError** | Akkor jelentkezik, ha probléma adódik egy importálási állítással. |

| **IOError** | Az `OSError` aliasa (modern Pythonban ugyanarra az alaphibára utalnak). |

| **TabError** | Akkor jelenik meg, ha a tabulátorok és szóközök használata a behúzásban következetlen. |
