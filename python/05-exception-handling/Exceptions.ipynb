# **Exception Handling** (Kivételkezelés)

A kivételkezelés lehetővé teszi, hogy a programunk **ellenőrzött módon kezelje a futás közbeni váratlan hibákat**, ahelyett, hogy hirtelen összeomlana. Ezzel növelhetjük az alkalmazások megbízhatóságát, és biztosíthatjuk, hogy a program futása ne szakadjon meg váratlanul.

Segítségével olyan futásidejű hibákat kezelhetünk, mint a **hibás felhasználói bemenet**, **hiányzó fájlok**, **nullával való osztás vagy a típuseltérések**.

---

## 1. Alapvető Példa
Az alábbi egyszerű példa bemutatja, hogyan kaphatunk el egy kivételt és hogyan kezelhetjük azt elegánsan:

```python
number = 10
try:
    result = number / 0
except ZeroDivisionError:
    print("Can't be divided by zero!")
```

**Magyarázat:** A szám 0-val való osztása egy `ZeroDivisionError` kivételt vált ki. A `try` blokk tartalmazza a kockázatos kódot, az `except` blokk pedig elkapja a hibát, így a program leállása helyett egy biztonságos hibaüzenet jelenik meg.

---

## 2. Szintaxis és Kulcsszavak
A Python négy fő kulcsszót biztosít a kivételek kezelésére, melyek mindegyike egyedi szerepet játszik:
* **`try`**: A kockázatos kódot futtatja, ami hibát okozhat.
* **`except`**: Elkapja és kezeli a hibát, ha az bekövetkezik.
* **`else`**: Csak akkor fut le, ha a `try` blokkban nem történt kivétel.
* **`finally`**: Mindenképpen lefut, függetlenül a hibáktól (hasznos erőforrások lezárásakor).

**Példa, mind a négy kulcsszó használatára:**


```python
try:
    denominator = 0
    result = 100 / denominator
except ZeroDivisionError:
    print("You can't divide by zero!")
except ValueError:
    print("Enter a valid number!")
else:
    print("Result is", result)
finally:
    print("Execution complete.")
```

---


## 3. Kivételek elkapása
Hatékonyabban kezelhetjük a hibákat, ha pontosan megadjuk a várt kivételek típusait. Ez biztonságosabbá teszi a kódot és megkönnyíti a hibakeresést (debugging).

### Különböző hibák egyedi kezelése
Ez a kód a `ValueError` és a `ZeroDivisionError` hibákat eltérő üzenetekkel kezeli:


```python
try:
    # Ez ValueError-t fog okozni, mert a "str" nem alakítható egésszé
    user_input = int("str")
    inverse_value = 1 / user_input
except ValueError:
    print("Not Valid!")
except ZeroDivisionError:
    print("Zero has no inverse!")
```

### Több kivétel kezelése egy blokkban
Ha több hibát is ugyanúgy szeretnénk kezelni, csoportosíthatjuk őket egy zárójeles felsorolásban:


```python
items = ["10", "twenty", 30]
try:
    # A "twenty" szöveg nem alakítható számmá
    total_sum = int(items[0]) + int(items[1])
except (ValueError, TypeError) as error:
    print("Error:", error)
except IndexError:
    print("Index out of range.")
```

### A Catch-All és veszélyei
Ha csak magát az `except` kulcsszót használjuk hibaosztály nélkül, az minden hibát elkap. 

**Kockázat:** Ez **megnehezíti a hibakeresést!!**, mivel elrejti a hiba valódi típusát és okát. Csak végső biztonsági hálóként szabad használni!

```python
try:
    # Szöveg osztása számmal TypeError-t okoz
    calculation = "100" / 20
except ArithmeticError:
    print("Arithmetic problem.")
except:
    print("Something went wrong!")
```

## 4. Kivételek 'kiváltása' 
A `raise` kulcsszó használatával mi magunk is kiválthatunk egy adott típusú kivételt, ha a program logikája azt megköveteli. 

- Olyan esetekben használjuk, ha matematikailag bár helyes, de a sajátos logika alapján nem. 
- Pl. *Banki átutalás:* 100 - 500 = - 400, de **negatív értéket nem lehet átutalni.** Így matematikailag helyes, de logikailag nem. 

Ugyanez érvényes az életkorra is: 


```python
def set_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative.")
    print(f"Age set to {age}")

try:
    set_age(-5)
except ValueError as error:
    print(error)
```

---



## 5. **Errors VS Exceptions** 
Bár mindkettő problémát jelent a programban, súlyosságuk és kezelésük időpontja eltér:

* **Error (Hiba):** A program logikájának vagy szintaxisának hibái (pl. `SyntaxError`). Jellemzően a kód **beolvasásakor/fordításakor** (compile time) jelentkeznek, és megakadályozzák a program elindulását.

* **Exception (Kivétel):** Olyan problémák, amelyek a program **futása közben** (runtime) jelentkeznek (pl. hibás bemenet, hiányzó fájl). Ezek kezelhetők a `try-except` blokkok segítségével, így a program tovább futhat.
