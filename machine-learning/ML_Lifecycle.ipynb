# **ML Lifecycle** (Gépi tanulási életciklus)

- Olyan strukturált folyamat, amelyet gépi tanulási modellek hatékony fejlesztésére, betanítására, telepítésére és karbantartására használnak. Több szakaszt foglal magában, mint például az adatgyűjtés, az előfeldolgozás, a modell betanítása, az értékelés és a monitorozás a pontos és megbízható előrejelzések biztosítása érdekében.

## 1. Problémameghatározás
Az első lépés a megoldandó probléma világos meghatározása. Egy jól megfogalmazott probléma alapot teremt a projekt céljainak, a várható eredményeknek és a szükséges megoldás típusának meghatározásához.

- Biztosítja az **üzleti igények** és a **technikai megoldások** összhangját
- Határozza meg a **projekt céljait, hatókörét és sikerkritériumait**

## 2. Adatgyűjtés
Olyan adathalmazok szisztematikus gyűjtését foglalja magában, amelyek **nyers adatként használhatók a modell betanításához**. Az adatok minősége és változatossága közvetlenül befolyásolja a modell teljesítményét.

### Jellemzője:

- **Relevancia**: A gyűjtött adatoknak relevánsnak kell lenniük a meghatározott problémához, és tartalmazniuk kell a szükséges jellemzőket.

- **Minőség**: Az adatok minőségének biztosítása olyan tényezők figyelembevételével, mint a pontosság és az etikus felhasználás.

- **Mennyiség**: Gyűjtsön össze elegendő adatmennyiséget egy robusztus modell betanításához.

- **Sokszínűség**: Különböző adathalmazok használata a forgatókönyvek és minták széles skálájának rögzítésére.


## 3. Adattisztítás, előfeldolgozás
A nyers adatok gyakran **rendezetlenek és strukturálatlanok**, és ha ezeket az adatokat közvetlenül használjuk a betanításhoz, az **gyenge pontossághoz vezethet**. Adattisztítást és előfeldolgozást kell végeznünk, ami a következőket foglalja magában:

- **Adattisztítás**: Olyan problémák kezelése, mint a **hiányzó értékek, kiugró értékek** és az adatokban található inkonzisztenciák.

- **Adatelőfeldolgozás**: Formátumok szabványosítása, **értékek skálázása** és kategorikus változók kódolása a konzisztencia érdekében.

- **Adatminőség**: Győződjön meg arról, hogy az adatok **jól rendszerezettek és előkészítettek** az érdemi elemzésre.


## 4. Feltáró adatelemzés 
Az adatokban rejtett mintázatok és jellemzők megtalálásához **feltáró adatelemzést** ( Exploratory Data Analysis - EDA) alkalmazunk, hogy betekintést nyerjünk az adatokba és**megértsük az adathalmaz szerkezetét**. Az EDA során olyan **mintázatok** és **betekintések jelennek meg**, amelyek **szabad szemmel nem feltétlenül láthatók**. Ez az értékes betekintés felhasználható a megalapozott döntéshozatalhoz.

### Jellemzői:

- **Feltárás**: Statisztikai és vizuális eszközök használata az adatokban található mintázatok feltárására. **Pl. diagramok használata.**

- **Minták és trendek:** Azonosítsa az adathalmazon belüli mögöttes **mintákat, trendeket** és lehetséges kihívásokat. **Pl. korosztálytól függőek lehetnek a szokások, trendek.**

- **Betekintések:** Értékes betekintést nyerhet a későbbi szakaszokban hozott megalapozott döntésekhez. **Pl. levonod a következtéteseket**

- **Döntéshozatal:** EDA használata **jellemzőtervezéshez, modellválasztáshoz**.
