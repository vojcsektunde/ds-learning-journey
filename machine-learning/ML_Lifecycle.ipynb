# **ML Lifecycle** (Gépi tanulási életciklus)

- Olyan strukturált folyamat, amelyet gépi tanulási modellek hatékony fejlesztésére, betanítására, telepítésére és karbantartására használnak. Több szakaszt foglal magában, mint például az adatgyűjtés, az előfeldolgozás, a modell betanítása, az értékelés és a monitorozás a pontos és megbízható előrejelzések biztosítása érdekében.

## 1. Problémameghatározás
Az első lépés a megoldandó probléma világos meghatározása. Egy jól megfogalmazott probléma alapot teremt a projekt céljainak, a várható eredményeknek és a szükséges megoldás típusának meghatározásához.

- Biztosítja az **üzleti igények** és a **technikai megoldások** összhangját
- Határozza meg a **projekt céljait, hatókörét és sikerkritériumait**

---
## 2. Adatgyűjtés
Olyan adathalmazok szisztematikus gyűjtését foglalja magában, amelyek **nyers adatként használhatók a modell betanításához**. Az adatok minősége és változatossága közvetlenül befolyásolja a modell teljesítményét.

### Jellemzője:

- **Relevancia**: A gyűjtött adatoknak relevánsnak kell lenniük a meghatározott problémához, és tartalmazniuk kell a szükséges jellemzőket.

- **Minőség**: Az adatok minőségének biztosítása olyan tényezők figyelembevételével, mint a pontosság és az etikus felhasználás.

- **Mennyiség**: Gyűjtsön össze elegendő adatmennyiséget egy robusztus modell betanításához.

- **Sokszínűség**: Különböző adathalmazok használata a forgatókönyvek és minták széles skálájának rögzítésére.

---
## 3. Adattisztítás, előfeldolgozás
A nyers adatok gyakran **rendezetlenek és strukturálatlanok**, és ha ezeket az adatokat közvetlenül használjuk a betanításhoz, az **gyenge pontossághoz vezethet**. Adattisztítást és előfeldolgozást kell végeznünk, ami a következőket foglalja magában:

- **Adattisztítás**: Olyan problémák kezelése, mint a **hiányzó értékek, kiugró értékek** és az adatokban található inkonzisztenciák.

- **Adatelőfeldolgozás**: Formátumok szabványosítása, **értékek skálázása** és kategorikus változók kódolása a konzisztencia érdekében.

- **Adatminőség**: Győződjön meg arról, hogy az adatok **jól rendszerezettek és előkészítettek** az érdemi elemzésre.

---
## 4. Feltáró adatelemzés 
Az adatokban rejtett mintázatok és jellemzők megtalálásához **feltáró adatelemzést** ( Exploratory Data Analysis - EDA) alkalmazunk, hogy betekintést nyerjünk az adatokba és**megértsük az adathalmaz szerkezetét**. Az EDA során olyan **mintázatok** és **betekintések jelennek meg**, amelyek **szabad szemmel nem feltétlenül láthatók**. Ez az értékes betekintés felhasználható a megalapozott döntéshozatalhoz.

### Jellemzői:

- **Feltárás**: Statisztikai és vizuális eszközök használata az adatokban található mintázatok feltárására. **Pl. diagramok használata.**

- **Minták és trendek:** Azonosítsa az adathalmazon belüli mögöttes **mintákat, trendeket** és lehetséges kihívásokat. **Pl. korosztálytól függőek lehetnek a szokások, trendek.**

- **Betekintések:** Értékes betekintést nyerhet a későbbi szakaszokban hozott megalapozott döntésekhez. **Pl. levonod a következtéteseket**

- **Döntéshozatal:** EDA használata **jellemzőtervezéshez, modellválasztáshoz**.


---
## 5. Jellemzők tervezése és kiválasztása
Olyan átalakító folyamat, amely magában foglalja a**releváns jellemzők kiválasztását a modell hatékonyságának és előrejelzésének növelése**, valamint a bonyolultság csökkentése érdekében.

### Jellemzői:

- **Jellemzőmérnökség**: Új jellemzők létrehozása vagy meglévők átalakítása a jobb minták és kapcsolatok rögzítése érdekében.

- **Funkciókiválasztás**: Azonosítsa a modell teljesítményét leginkább befolyásoló funkciók részhalmazát.

- **Szakterületi szakértelem:** Használja a szakterületi ismereteket olyan jellemzők tervezéséhez, amelyek érdemi módon hozzájárulnak az előrejelzéshez .

- **Optimalizálás**: A funkciók halmazának kiegyensúlyozása a pontosság érdekében, miközben minimalizálja a számítási bonyolultságot.

---
## 6. Modell kiválasztása
Egy jó gépi tanulási modellhez **a modellválasztás nagyon fontos** része, mivel olyan modellt kell találnunk, amely **összhangban van a meghatározott problémánkkal, az adatok jellegével**, a probléma összetettségével és a **kívánt eredményekkel.**

### Jellemzői:

- **Komplexitás**: A modell kiválasztásakor vegye figyelembe a probléma összetettségét és az adatok jellegét.

- **Döntési tényezők:** A modell kiválasztásakor olyan tényezőket kell értékelni, mint a teljesítmény, az értelmezhetőség és a skálázhatóság.

- **Kísérletezés:** Kísérletezz különböző modellekkel, hogy megtaláld a problémára legmegfelelőbbet.

---
## 7. Modell betanítása
A kiválasztott modellel a gépi tanulási életciklus a modell**betanítási folyamatába lép.** Ez a folyamat magában foglalja a modell historikus adatokkal való megismerését, amely lehetővé teszi számára, **hogy mintákat, kapcsolatokat és függőségeket tanuljon az adathalmazon belül.**

### Jellemzői:

- **Iteratív folyamat:** A modell iteratív betanítása, a paraméterek módosításával a hibák minimalizálása és a pontosság növelése érdekében.
- **Optimalizálás**: Finomhangolja a modellt a prediktív képességeinek optimalizálása érdekében.
- **Validálás:** A modell szigorú betanítása az új, láthatatlan adatok pontosságának biztosítása érdekében.

---
## 8. Modellértékelés és finomhangolás

A modellértékelés **szigorú tesztelést** foglal magában validációval vagy tesztadatkészletekkel szemben, hogy a modell pontosságát **új, nem látható adatokon** teszteljék. Betekintést nyújt a**modell erősségeibe és gyengeségeibe** 

Ha a modell **nem éri el** a kívánt teljesítményszintet, szükség lehet a **modell újbóli hangolására** és a hiperparaméterek módosítására a prediktív pontosság növelése érdekében.

## Jellemzői:

- **Értékelési mutatók:** Használjon olyan mutatókat, mint a pontosság, a precizitás, a visszahívás és az F1 pontszám a modell teljesítményének értékeléséhez.

- **Erősségek és gyengeségek:** Szigorú teszteléssel azonosítsa a modell erősségeit és gyengeségeit.

- **Iteratív fejlesztés:** Modellhangolás kezdeményezése a hiperparaméterek beállításához és a prediktív pontosság növeléséhez.

- **Modell robusztussága:** Iteratív hangolás a kívánt modell robusztussági és megbízhatósági szintek eléréséhez.

---
## 9. Modell telepítése
A modell most már készen áll a valós **alkalmazásokhoz való telepítésre**. Ez magában foglalja**a prediktív modell integrálását** a meglévő rendszerekkel, lehetővé téve a vállalkozások számára, hogy ezt megalapozott döntéshozatalhoz használják.

### Jellemzői:

- Integráció a **meglévő rendszerekkel**

- **Döntéshozatal engedélyezése** előrejelzések segítségével

- A telepítés skálázhatóságának és biztonságának biztosítása

- **API-k **vagy folyamatfolyamatok biztosítása éles használatra

---
## 10.  Modellfelügyelet és -karbantartás

A telepítés utáni modelleket** monitorozni kell, hogy biztosítsuk a megfelelő teljesítményt** az idő múlásával. A **rendszeres nyomon követés** segít az adateltolódás, a pontosságcsökkenés vagy a változó minták észlelésében, és újratanításra lehet szükség a modell**valós használatban való megbízhatóságának megőrzése** érdekében.

### Jellemzői:

- Modell **teljesítményének nyomon követése** az idő múlásával

- **Adateltolódás** vagy koncepcióeltolódás **észlelése**

- **Frissítse és tanítsa újra**a modellt, ha a pontosság csökken

- **Naplók és riasztások vezetése valós idejű problémák** esetén
