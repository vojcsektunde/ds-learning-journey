# **AI, ML és DL**

##  **Mesterséges intelligencia (MI)**

A mesterséges intelligencia a számítástechnika azon területe, amely olyan rendszerek létrehozására összpontosít, amelyek **emberi intelligenciát igénylő feladatokat** képesek ellátni, például **érvelést**, **problémamegoldást** és nyelvi megértést. A mesterséges intelligencia szabályalapú módszereket és **gépi tanulást is használhat**.

> Példák : Sophia robot, mesterséges intelligencia alapú chatbotok, csalásészlelés, orvosi diagnózis.


## **Gépi tanulás (ML)**

A gépi tanulás a mesterséges intelligencia egy ága, amely lehetővé teszi a rendszerek számára, hogy**mintákat tanuljanak az adatokból, és ezeket előrejelzések vagy döntések meghozatalára** használják. A gyakori megközelítések közé tartozik a felügyelt, a felügyelet nélküli és a megerősítéses tanulás.

> Példák : Amazon-ajánlások, Netflix-javaslatok, spamészlelés.

## **Mélytanulás (DL)**
A mélytanulás a **gépi tanulás** egy olyan **részhalmaza**, amely **többrétegű neurális hálózatokat** használ az adatokból származó összetett **minták megtanulására.** Széles körben használják képeket, hanganyagokat, szöveget és videót tartalmazó feladatokhoz.

> Példák : önvezető autók, arcfelismerés, hangasszisztensek


## Összefoglaló táblázat

| Vonatkozás | Mesterséges intelligencia | Gépi tanulás | Mély tanulás |
| :--- | :--- | :--- | :--- |
| **Meghatározás** | Az intelligens rendszerek létrehozásának széles területe | Az MI azon részterülete, amely adatokból tanul | Mély neurális hálózatokon alapuló gépi tanulási részterület |
| **Hatály** | Legszélesebb | Keskenyebb, mint a mesterséges intelligencia | Legszűkebb |
| **Hogyan működik** | Szabályokat, keresést, gépi tanulást (ML), lekérdezést (DL) és egyéb technikákat használ | Algoritmusok segítségével tanul mintákat az adatokból | Többrétegű neurális hálózatokat használ komplex minták tanulásához |
| **Adatkövetelmények** | A megközelítéstől függ | Általában betanítási adatokat igényel | Gyakran előnyös a nagy adathalmazokból, különösen összetett feladatok esetén |
| **Jellemzőmérnöki** | A megközelítéstől függ | Gyakran manuálisan kiválasztott vagy tervezett funkciókat igényel | Automatikusan képes megtanulni a jellemzőket nyers bemenetekből |
| **Gyakori technikák** | Szakértői rendszerek, keresés, gépi tanulás, lekérdezés és megerősítéses tanulás | Regresszió, döntési fák, klaszterezés, SVM-ek és együttes módszerek | CNN-ek, RNN-ek, LSTM-ek, transzformátorok és egyéb neurális hálózati architektúrák |
| **Alkalmazások** | Robotika, tervezés, NLP, szakértői rendszerek és intelligens asszisztensek | Osztályozás, regresszió, ajánlás és előrejelzés | Számítógépes látás, beszédfelismerés és generatív mesterséges intelligencia |
| **Számítási követelmények** | Széles körben változik | Általában mérsékelt, az algoritmustól és az adathalmaztól függően | Gyakran magas, különösen nagy modellek betanításakor |
