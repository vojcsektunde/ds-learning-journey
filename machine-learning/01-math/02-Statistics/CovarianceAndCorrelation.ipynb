# Kovariancia és korreláció

A kovariancia és a korreláció a statisztika **két kulcsfogalma**, amelyek segítenek **két változó közötti kapcsolat elemzésében.**

A kapcsolat jobb megértéséhez vegyük figyelembe olyan tényezőket, mint a napfény, a víz és a talaj tápanyagai, amelyek független változók, és befolyásolják a növények növekedését, ami a függő változónk.

## Kovariancia

A kovariancia azt méri, hogy **két véletlenszerű változó hogyan változik együtt.** A kovariancia az átlaguktól való eltéréseik szorzatának átlagolásával számítható ki. A **pozitív érték** azt jelenti, hogy u**gyanabba az irányba mozognak**, míg a **negatív érték** azt, hogy **ellentétes irányban mozognak**.

- Bármilyen értéket felvehet - végtelen és + végtelen között, ahol a negatív érték a negatív, míg a pozitív érték a pozitív kapcsolatot jelöli.

- Egy lineáris kapcsolat irányát jelzi, de nem méri annak erősségét szabványosított módon.

- Megadja a változók közötti kapcsolat irányát.

### Minta kovariancia (Sample Covariance)

$$\text{Cov}_S(X, Y) = \frac{1}{n - 1} \sum_{i=1}^{n} (X_i - \bar{X})(Y_i - \bar{Y})$$

- $X_i, Y_i$: a minta $i$-edik adatpárjának értékei.  
-  $\bar{X}, \bar{Y}$: az $X$ és $Y$ változók mintaátlagai.  
-  $n$: a minta elemszáma.   
- $n - 1$: a szabadsági fokok száma (Bessel-féle korrekció a torzítatlan becslés érdekében).   

### Populációs kovariancia (Population Covariance)

-$$\text{Cov}_P(X, Y) = \frac{1}{n} \sum_{i=1}^{n} (X_i - \mu_X)(Y_i - \mu_Y)$$   

A képlet jelölései:

- $\mu_X, \mu_Y$: a teljes populáció valódi átlagai.  
- $n$: a teljes populáció elemszáma.   
- Mivel itt a teljes sokaságot ismerjük, $n - 1$ helyett simán $n$-nel osztunk.  


## Koverencia típusai 
- **Pozitív kovariancia :** Amikor az egyik változó értéke növekszik, a másik változó értéke is növekszik, és fordítva.
- **Negatív kovariancia :** Amikor az egyik változó értéke növekszik, a másik változó értéke csökkenni kezd.
-  **Nulla kovariancia :** A két változó között nincs lineáris kapcsolat, egymástól függetlenül mozognak.


## Alkalmazása 

- **Pénzügyek** : Segít elemezni az eszközök együttes mozgását (kockázatdiverzifikáció)
- **Közgazdaságtan** : A gazdasági mutatók közötti összefüggéseket vizsgálja
- **Jelfeldolgozás** : Jelek mintázatainak elemzésére szolgál
- **Környezettudomány** : Olyan összefüggéseket vizsgál, mint a hőmérséklet és a páratartalom

## Korreláció

A korreláció **két változó közötti lineáris kapcsolat erősségének és irányának** standardizált mérőszáma. A kovarianciából származtatható, **és -1 és 1 között mozog.** A kovarianciával ellentétben, amely csak a kapcsolat irányát jelzi, a **korreláció standardizált mérőszámot biztosít.**

- **Pozitív korreláció (közel +1-hez)**: Ahogy az egyik változó értéke növekszik, a másik változó értéke is növekszik.

- **Negatív korreláció (közel -1-hez):** Ahogy az egyik változó értéke növekszik, a másik változó értéke csökkenni kezd.

- **Nulla korreláció:**A változók között nincs lineáris kapcsolat.

---
A korrelációs együttható ρ , az X és Y változókra a következőképpen van definiálva:

- A korreláció -1 és +1 közötti értékeket vehet fel, ahol a +1-hez közeli értékek erős pozitív korrelációt, a -1-hez közeli értékek pedig erős negatív korrelációt jelentenek.
- A változók negatívan is összefüggésben lehetnek (azaz ellentétes irányba mozoghatnak).
- Megadja a változók közötti kapcsolat irányát és erősségét.


$$\text{Korr}(x, y) = \frac{\sum_{i=1}^{n} (x_i - \bar{x})(y_i - \bar{y})}{\sqrt{\sum_{i=1}^{n} (x_i - \bar{x})^2 \sum_{i=1}^{n} (y_i - \bar{y})^2}}$$

- $\text{Korr}(x, y)$: az $x$ és $y$ változók közötti korrelációs együttható (gyakran $r$-rel jelölve).  
-  $x_i$ és $y_i$ (a képen $x_{én}, y_{én}$): az $x$ és $y$ változók $i$-edik megfigyelt értékei (egyedi minták). 
-  $\bar{x}$ és $\bar{y}$ (a képen $x', y'$): az $x$ és $y$ minták átlagai.   
- $n$: a minták teljes száma (adatpárok száma).   
- $\sum_{i=1}^{n}$: az értékek összegezése $i=1$-től $n$-ig.  



## Alkalmazása

- **Piacelemzés**: Értsd meg az ügyfelek viselkedését az értékesítéssel szemben
- **Egészségügy** : Az egészségügyi változók közötti kapcsolatok elemzése
- **Időjárás-előrejelzés** : Az időjárási tényezők közötti összefüggések vizsgálata
- **Gépi tanulás** : Jellemzők kiválasztása kapcsolatok alapján

## Különbségek

| Kovariancia | Korreláció |
| :--- | :--- |
| A kovariancia azt méri, hogy két véletlen változó mennyire változik együtt | A korreláció egy statisztikai mérőszám, amely megmutatja, hogy két változó milyen erősen kapcsolódik egymáshoz.|
|Két változó vagy adathalmaz közötti kapcsolatot foglalja magában|Több változó közötti kapcsolatot is magában foglal A korreláció (konkrétan a Pearson-korreláció) két változó közötti kapcsolatot méri.| 
|Feküdjön a -végtelen és a +végtelen között|-1 és +1 között helyezkedik el|
|A korreláció mértéke|A kovariancia skálázott változata|
|Irányt ad a kapcsolatnak	|Irányt és erőt ad a kapcsolatnak|
|A változó skálájától függ|Független a változó skálájától|
|Méretekkel rendelkezik|Dimenzió nélküli|

 A legfontosabb különbség az, hogy a **kovariancia a változók közötti kapcsolat irányát** mutatja, míg **a korreláció standardizált formában mind az irányt, mind az erősséget mutatja**.
