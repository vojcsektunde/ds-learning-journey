# Kovariancia és korreláció

A kovariancia és a korreláció a statisztika **két kulcsfogalma**, amelyek segítenek **két változó közötti kapcsolat elemzésében.**

A kapcsolat jobb megértéséhez vegyük figyelembe olyan tényezőket, mint a napfény, a víz és a talaj tápanyagai, amelyek független változók, és befolyásolják a növények növekedését, ami a függő változónk.

## Kovariancia

A kovariancia azt méri, hogy **két véletlenszerű változó hogyan változik együtt.** A kovariancia az átlaguktól való eltéréseik szorzatának átlagolásával számítható ki. A **pozitív érték** azt jelenti, hogy u**gyanabba az irányba mozognak**, míg a **negatív érték** azt, hogy **ellentétes irányban mozognak**.

- Bármilyen értéket felvehet - végtelen és + végtelen között, ahol a negatív érték a negatív, míg a pozitív érték a pozitív kapcsolatot jelöli.

- Egy lineáris kapcsolat irányát jelzi, de nem méri annak erősségét szabványosított módon.

- Megadja a változók közötti kapcsolat irányát.

### Minta kovariancia (Sample Covariance)

$$\text{Cov}_S(X, Y) = \frac{1}{n - 1} \sum_{i=1}^{n} (X_i - \bar{X})(Y_i - \bar{Y})$$

- $X_i, Y_i$: a minta $i$-edik adatpárjának értékei.  
-  $\bar{X}, \bar{Y}$: az $X$ és $Y$ változók mintaátlagai.  
-  $n$: a minta elemszáma.   
- $n - 1$: a szabadsági fokok száma (Bessel-féle korrekció a torzítatlan becslés érdekében).   

### Populációs kovariancia (Population Covariance)

-$$\text{Cov}_P(X, Y) = \frac{1}{n} \sum_{i=1}^{n} (X_i - \mu_X)(Y_i - \mu_Y)$$   

A képlet jelölései:

- $\mu_X, \mu_Y$: a teljes populáció valódi átlagai.  
- $n$: a teljes populáció elemszáma.   
- Mivel itt a teljes sokaságot ismerjük, $n - 1$ helyett simán $n$-nel osztunk.  


## Koverencia típusai 
- **Pozitív kovariancia :** Amikor az egyik változó értéke növekszik, a másik változó értéke is növekszik, és fordítva.
- **Negatív kovariancia :** Amikor az egyik változó értéke növekszik, a másik változó értéke csökkenni kezd.
-  **Nulla kovariancia :** A két változó között nincs lineáris kapcsolat, egymástól függetlenül mozognak.
