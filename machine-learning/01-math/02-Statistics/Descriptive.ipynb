# **Leíró statisztika**

A statisztika az adattudomány alapja. A leíró statisztikák egyszerű eszközök, amelyek segítenek megérteni és összefoglalni az adatokat. Megmutatják egy **adathalmaz alapvető jellemzőit, például az átlagos, a legmagasabb és a legalacsonyabb értékeket, valamint a számok szórását.**



## A leíró statisztikák típusai

A leíró statisztikai módszerek standard osztályozására három kategória létezik, amelyek mindegyike más-más célt szolgál az adatok összefoglalásában és leírásában. Segítenek megérteni:

- Hol találhatók az adatközpontok (a központi tendencia mértékei)
- Az adatok szórtsága (változékonyság mérése)
- Az adatok eloszlása ​​(gyakorisági eloszlás mértékei)


## 1. A központi tendencia mértékei

Statisztikai értékek, amelyek leírják az adathalmazon belüli központi pozíciót. 

A központi tendencia három fő mérőszáma: **átlag, medián, módusz**

### Átlag 
A megfigyelések **összege osztva a megfigyelések teljes számával**. 

> x = Megfigyelések
> n = kifejezések száma

```python
import numpy as np

nums = [5, 6, 11]
mean = np.mean(nums)

print("Mean = ", mean)
```

### Módusz
Az adathalmazban **leggyakrabban előforduló érték**. **Kategóriaalapú** adatok esetén hasznos, és olyan esetekben, amikor a **leggyakoribb választás ismerete** kulcsfontosságú.

```python
import scipy.stats as stats

nums = [1, 2, 2, 3, 2]
result = stats.mode(nums)

print(f"Mode : {result.mode}")
print(f"Count: {result.count}")
```

### Medián 

 A **rendezett adathalmaz középső értéke**. Ha az értékek száma páratlan, akkor a középső érték, **ha páros, akkor a két középső érték átlaga**. Ferde adatok esetén gyakran jobb, mint az átlag.

```python
import numpy as np

nums = [1, 2, 3, 4]
median = np.median(nums)

print("Median = ", median)

nums.append(5)

median = np.median(nums)
print("New median = ", median)
```


## A változékonyság mérése

Nemcsak az adatközpontok helyének ismerete fontos, hanem annak eloszlása ​​is. A változékonyság mértéke, más néven a szóródás mértéke, segít a megfigyelések eloszlásának vagy terjedésének nyomon követésében egy adathalmazban. Segítségükkel azonosíthatók a kiugró értékek, értékelhetők a modellfeltevések, és megérthető az adatok változékonysága az átlaghoz viszonyítva. A változékonyság kulcsfontosságú mértékei a következők:

### Terjedelem 
 Az adathalmazunk **legnagyobb és legkisebb adatpontja közötti különbséget** írja le. Minél nagyobb a terjedelem, annál nagyobb az adatok szórása, és fordítva. Bár könnyen kiszámítható, érzékeny a kiugró értékekre. Ez a mérőszám gyors képet adhat az adatok szórásáról, de más statisztikákkal is ki kell egészíteni.

**Terjedelem** = legnagyobb adatérték - legkisebb adatérték

```python
import numpy as np

arr = [1, 2, 3, 4, 5]

Maximum = max(arr)
Minimum = min(arr)

Range = Maximum-Minimum
print("Maximum = {}, Minimum = {} and Range = {}".format(Maximum, Minimum, Range))
```

 ### Variancia 
 Az átlagtól való átlagos négyzetes eltérés. Kiszámítása úgy történik, hogy megtaláljuk az **egyes adatpontok és az átlag közötti különbséget, négyzetre emeljük őket, összeadjuk az összeset, majd elosztjuk az adathalmazban jelen lévő adatpontok számával.**

$$\sigma^2 = \frac{\sum (x - \mu)^2}{N}$$

 - x -> megfigyelés vizsgálat alatt
 - N -> kifejezések száma 
 - μ -> átlag 


```python
import statistics
arr = [1, 2, 3, 4, 5]
print("Variance =",(statistics.variance(arr)))
```

### Szórás   (Standard deviation)

Azt méri, hogy az adatértékek **mennyire térnek el az átlagtól**. Széles körben használják a statisztikában és a gépi tanulásban az adatszórás és a modell teljesítményének megértésére.

- A **variancia négyzetgyökeként** van definiálva.
- Az alacsony szórás azt jelenti, hogy az értékek közel vannak az átlaghoz.
- A magas szórás nagyobb variációt jelez az adathalmazban

$$\sigma = \sqrt{\frac{\sum (x - \mu)^2}{N}}$$

- x = figyelembe vett megfigyelés
- N = kifejezések száma 
- μ = átlag

```python
import statistics
arr = [1, 2, 3, 4, 5]
print("Std = ", (statistics.stdev(arr)))
```
