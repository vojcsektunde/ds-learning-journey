# **Inferenciális** (következtetéses) statisztika

Lehetővé teszi számunkra, hogy mintavételezett adatok alapján **előrejelzéseket és következtetéseket** tegyünk egy populációról. A leíró statisztikával ellentétben, amely csak az adatokat összegzi, a következtetéses statisztika lehetővé teszi **hipotézisek tesztelését, becslések készítését** és az előrejelzéseinkkel kapcsolatos bizonytalanság mérését.


## Szükségessége
A valós helyzetekben egy **teljes populáció elemzése gyakran lehetetlen**. Ehelyett **mintából gyűjtünk adatokat**, és következtetési statisztikákat használunk a következőkre:

- Vegyük figyelembe a teljes populációt
- Állítások vagy hipotézisek tesztelése
- Számítsuk ki a konfidenciaintervallumokat és a p-értékeket a bizonytalanság mérésére
- Statisztikai modellekkel készítsünk előrejelzéseket

## Következtető statisztikák technikái

A következtetéses statisztika számos kulcsfontosságú módszert kínál a hipotézisek tesztelésére, a populációs paraméterek becslésére és az előrejelzések készítésére.

### Konfidenciaintervallumok

Olyan értéktartományt ad meg, amely valószínűleg tartalmazza a valódi populációs paramétert. Segít számszerűsíteni egy becslés bizonytalanságát. Az átlag konfidenciaintervallumának kiszámítására szolgáló képlet:

$$\text{CI} = \bar{x} \pm Z_{\alpha/2} \times \frac{\sigma}{\sqrt{n}}$$

- $\text{CI}$ (Confidence Interval): konfidenciaintervallum (megbízhatósági intervallum).   
- $\bar{x}$ : a minta átlaga (pontbecslés a populáció átlagára).  
- $Z_{\alpha/2}$: a standard normális eloszlás kritikus értéke, amely a választott megbízhatósági szinttől ($1 - \alpha$) függ

Például, ha 100 ember átlagos magasságát mérjük, egy 95%-os konfidenciaintervallum egy olyan tartományt ad meg, amelybe a valódi populáció átlagos magassága valószínűleg beleesik. Ez segít felmérni a becslésünk pontosságát és összehasonlítani a modelleket (mint az A/B tesztelésben ).

### Hipotézisvizsgálat 

Egy formális eljárás az adatokkal kapcsolatos állítások vagy feltételezések tesztelésére. A következő lépéseket foglalja magában:

-  Nullhipotézis (H₀) : Az alapértelmezett feltételezés, például: „nincs különbség a két modell között”.
- Alternatív hipotézis (H₁) : Az az állítás, amelyet bizonyítani kíván, például: „Az A modell jobban teljesít, mint a B modell.”
Adatokat gyűjtünk és kiszámítunk egy tesztstatisztikát (például Z-t egy Z-próbához vagy t-t egy T-próbához):

$$Z = \frac{\bar{x} - \mu_0}{\frac{\sigma}{\sqrt{n}}}$$

> - $Z$: a számított Z-érték (tesztstatisztika), amely a standard normális eloszlást követi a nullhipotézis teljesülése esetén.   
> - $\bar{x}$ (x-felülvonás): a megfigyelt minta átlaga.   
> - $\mu_0$ (mű null): a nullhipotézisben ($H_0$) feltételezett populációs átlag.   
> - $\sigma$ (szigma): a populáció ismert szórása.   
> - $n$: a minta elemszáma.  
>-  $\frac{\sigma}{\sqrt{n}}$: az átlag standard hibája (Standard Error, SE)


A tesztstatisztika kiszámítása után összehasonlítjuk azt egy kritikus értékkel, vagy p-értéket használunk annak eldöntésére, hogy elvetjük-e a nullhipotézist. Ha a p-érték kisebb, mint az α szignifikanciaszint (általában 0,05), akkor elvetjük a nullhipotézist.


$$p\text{-value} = 2 \cdot P(Z > \vert{}z_{\text{obs}}\vert{})$$

- $p\text{-value}$: a $p$-érték (szignifikanciaszinthez hasonlítandó mutató).  
-  $z_{\text{obs}}$: a megfigyelt, mintából kiszámított Z-statisztika értéke (az előző lépésben kiszámolt $Z$-érték).  
- $\vert{}z_{\text{obs}}\vert{}$: a számított Z-érték abszolút értéke.   
- $Z$: a standard normális eloszlást követő valószínűségi változó ($N(0, 1)$).   
- $2 \cdot$: a szorzó arra utal, hogy kétoldali tesztet végzünk (az eloszlás mindkét szélén / farkában lévő területet figyelembe vesszük).  

 ### Centrális határeloszlás-tétel 

 Kimondja, hogy a mintaátlag eloszlása ​​a mintaelemszám növekedésével normális eloszlást fog megközelíteni, függetlenül az eredeti populációeloszlástól. Ez azért fontos, mert sok statisztikai módszer feltételezi, hogy az adatok normális eloszlásúak. A centrális határeloszlás-tétel matematikailag a következőképpen fejezhető ki:

 $$\bar{X} \sim N\left(\mu, \frac{\sigma}{\sqrt{n}}\right)$$

 - μ - népesség átlaga
- σ  - populációs szórás
- n - minta mérete

Ez a tétel lehetővé teszi számunkra, hogy normális eloszláson alapuló módszereket alkalmazzunk akkor is, ha az eredeti adatok nem normális eloszlásúak, például ferde jövedelmi vagy vásárlási viselkedési adatok esetén.

## Hibák a következtetéses statisztikákban

A hipotézisvizsgálatban az **I. típusú hiba** és a **II. típusú hiba** kulcsfontosságú fogalmak:

- I. típusú hiba akkor fordul elő, amikor **tévesen elutasítunk egy igaz nullhipotézist.** Az I. típusú hiba valószínűségét a következővel jelöljük: `α` *(a szignifikanciaszint).*

- II. típusú hiba akkor fordul elő, **ha nem tudunk elvetni egy hamis nullhipotézist.** A másodfajú hiba valószínűségét a következővel jelöljük: β és a teszt erejét a következő adja meg: `1 − β`

A cél **ezen hibák minimalizálása** a minta méretének és a szignifikanciaszintek gondos megválasztásával.

## Parametrikus és nem paraméteres tesztek

A statisztikai próbák segítenek eldönteni, hogy az adatok **alátámasztanak-e egy hipotézist.**

 Kiszámítanak egy próbastatisztikát, amely megmutatja, hogy az adatok mennyire térnek el a feltételezéstől **(nullhipotézis)**. Ezt **összehasonlítják egy kritikus értékkel vagy p-értékkel**, hogy elfogadják vagy elvethessék a nullhipotézist.

- **Paraméteres tesztek:** Ezek a tesztek azt feltételezik, hogy az adatok egy adott eloszlást (gyakran normális eloszlást) követnek, és konzisztens varianciájuk van. Általában folytonos adatokhoz használják őket. Ilyenek például a Z-teszt , a T-teszt és az ANOVA . Ezek a tesztek hatékonyak modellek összehasonlítására vagy teljesítmény mérésére, amikor a feltételezések teljesülnek.

- **Nem paraméteres tesztek:** A nem paraméteres tesztek nem feltételeznek egy adott eloszlást az adatokra, így ideálisak kis minták vagy nem normális adatok, például kategorikus vagy rangsorolt ​​adatok esetén. Ilyenek például a Chi-négyzet teszt, a Mann-Whitney U teszt és a Kruskal-Wallis teszt. Hasznosak, ha az adatok ferdék vagy kategorikusak, például az ügyfélértékelések vagy viselkedések esetében.

## Példa: Új kézbesítési algoritmus kiértékelése következtetési statisztikák segítségével

Egy gyors kereskedelemmel foglalkozó cég azt szeretné ellenőrizni, hogy egy új kézbesítési algoritmus csökkenti-e a kézbesítési időket a jelenlegi rendszerhez képest.

### **Kísérlet beállítása:**
- 100 megbízás két csoportra oszlott: 50 az új algoritmussal, 50 a jelenlegi rendszerrel.
- Mindkét csoport szállítási idejét rögzítjük.

**Lépések**
**1. Hipotézisek:**
- **Null (H0):** Az új algoritmus nem csökkenti a kézbesítési időt.
- **Alternatíva (H1)**: Az új algoritmus csökkenti a kézbesítési időt.


**2. Jelentőségszint:**

0,05-re állítva (5% a H0 téves elutasításának kockázata).

- **Első típusú hiba:** Azt gondolni, hogy az új rendszer jobb, amikor nem az.
- **Másodtípusú hiba:** Valódi javulás elmaradása.


**3. Tesztstatisztika**: Hasonlítsa össze a két csoport átlagos kézbesítési idejét

**4. Elemzés:**
- Számítsd ki az átlagokat és a különbségeket.
- Ellenőrizd, hogy az adatok nagyjából normálisak-e.

**5. Végezzen el t-test-et vagy z-test-et:**

Ha a p-érték < 0,05, akkor elvetjük a H0-t, és arra a következtetésre jutunk, hogy az új algoritmus jobb. Ellenkező esetben nincs egyértelmű javulás.

**6. Konfidenciaintervallum:** Például a -5 és -2 perc közötti tartomány azt jelenti, hogy az új rendszerrel a kézbesítések 2-5 perccel gyorsabbak.
