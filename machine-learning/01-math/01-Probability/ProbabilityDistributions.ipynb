# **Valószínűségi eloszlások az adattudományban**

Amikor összetett adatokkal, például vevői vásárlásokkal, részvényárfolyamokkal vagy időjárással dolgozunk, a valószínűségeloszlások segítenek megválaszolni a következő kérdéseket:

- Mi a legvalószínűbb, hogy megtörténik?
- Mik a ritka vagy szokatlan eredmények?
- Az értékek közel vannak egymáshoz, vagy szétszórtan helyezkednek el?

Ez segít nekünk jobb előrejelzéseket készíteni és megérteni a bizonytalanságot.

## **Miért fontosak a valószínűségeloszlások?**
- Elmagyarázza, hogyan viselkednek az adatok (fürtözött vagy szétszórt)
- Gépi tanulási modellek alapját képezik
- Statisztikai tesztekben használják (pl. p-érték )
- Segít azonosítani a kiugró értékeket és előrejelzéseket készíteni


## **Főbb összetevői**
Vizsgáljuk meg, hogyan írhatjuk le a valószínűségeiket három kulcsfogalom segítségével:

1. **Valószínűségi tömegfüggvény (PMF)**: **Diszkrét változók** (pl. megvásárolt termékek száma) esetén használatos. Megadja az egyes pontos értékek valószínűségét. *Például a vásárlók 25%-a pontosan 3 terméket vásárol.*

2. **Valószínűségi sűrűségfüggvény (PDF)**: **Folytonos változók** esetén használatos *(pl. elköltött összeg)*. Megmutatja, hogyan oszlanak el a valószínűségek **egy tartományon belül**, de **nem mutatja egyetlen pontos érték** esélyét, mivel az értékek végtelenek lehetnek.

3. **Kumulatív eloszlásfüggvény (CDF)**: **Mindkét típusnál használjuk**, és megmutatja annak valószínűségét, hogy egy érték kisebb vagy egyenlő egy adott számmal. *Például a CDF(3) = 0,75 azt jelenti, hogy a vásárlók 75%-a 3 vagy kevesebb terméket vásárol;* 

a CDF($50) = 0,80 azt jelenti, hogy a vásárlók 80%-a 50 dollárt vagy kevesebbet költ. 
A CDF-et az alábbi képlettel kaphatjuk meg:

> CDF: $F_X(x) = P(X \le x) = \int_{-\infty}^{x} f(t) \, dt$

> Ahol F(x) a CDF, f(t) pedig a PDF


## **Valószínűségeloszlások típusai**
A valószínűségi eloszlások a véletlen változók természete alapján két fő típusra oszthatók: **diszkrét és folytonos**.


## **Diszkrét adateloszlások**
Akkor diszkrét, ha a véletlen változó **megszámlálható**, specifikus értékeket vehet fel. 

Például, amikor egy vásárló egyetlen rendelésben megvásárolható termékek számát jósoljuk meg, a lehetséges eredmények egész számok, például 0, 1, 2, 3 stb. 

**Nem lehet 2,5 terméket vásárolni, így ez egy diszkrét véletlen változó.** Különböző eloszlásokat foglal magában. 

### **Binomiális eloszlás**

Adott számú próbálkozásból **mennyi a siker valószínűsége.**
$$P(X = k) = \binom{n}{k} p^k (1 - p)^{n-k}$$

- n = próbálkozások száma
- k = sikerek száma
- p = a siker valószínűsége


> Példa: feldobunk egy érmét tízszer, és fejeket számolunk.

> - Próbálkozások száma: 10
> - Két kimenetel próbánként: fej (siker) vagy írás (kudarc)
> - Siker valószínűsége (fej): 0,5


### **Bernoulli-eloszlás**
Olyan kísérleteket ír le, amelyekben csak egy kísérlet és két lehetséges kimenetel van: **siker vagy kudarc.** Ez a legegyszerűbb valószínűségi eloszlás.

$$P(X = x) = p^x (1 - p)^{1 - x}, \quad x \in \{0, 1\}$$

> Példa: feldobunk egy érmét egyszer, és megnézzük, hogy fejre esik-e.

> - Csak egy próba
> - Két kimenetel: fej (siker) vagy írás (kudarc)
> - Siker valószínűsége: 0,5
> - A grafikon két oszlopot mutat, amelyek a sikert (1) és a kudarcot (0) egyenlő valószínűséggel jelölik


### **Poisson-eloszlás**

Egy **adott időben vagy területen belül véletlenszerűen bekövetkező események számát** modellezi.
$$P(X = k) = \frac{\lambda^k e^{-\lambda}}{k!}$$

λ = átlagos előfordulási arány

> Példa: Óránként egy kávézóba belépő vendégek számának mérése. Segít megjósolni egy adott számú esemény valószínűségét az átlagos arány alapján.

> - Fix intervallumon belüli eseményeket számol
> - Az átlagos díjszabás (pl. 5 vásárló/óra) ismert
> - Kiszámítja a pontos darabszám valószínűségét (pl. pontosan 3 vásárló)
> - A grafikon egy átlagos arány körüli görbét mutat, amely a kevésbé valószínű számok esetén elkeskenyedik.

## **Geometriai eloszlások**
A geometriai eloszlás modellezi **az első siker eléréséhez szükséges próbálkozások számát** ismételt független próbálkozások során.

$$P(X = k) = (1 - p)^{k - 1} p$$


> Példa: hány e-mailt kell elküldenie, mielőtt egy ügyfél vásárol. Segít megjósolni a siker esélyét az egyes próbák során.

> - Számolja a próbálkozásokat az első sikerig
> - Minden próba független, fix sikervalószínűséggel
> - Hasznos olyan kérdésekhez, mint például a „Hány e-mailt kell még várni az első vásárlásig?”
> - A grafikon csökkenő görbét mutat, kevesebb próbálkozás nagyobb valószínűséggel


# **Folyamatos adateloszlások**

Akkor beszélünk, ha a véletlen változó **egy adott tartományon belül bármilyen értéket felvehet,** például amikor azt vizsgáljuk, hogy egy vásárló **mennyi pénzt költ** egy üzletben, akkor az összeg bármilyen valós szám lehet, beleértve a tizedesjegyeket is, például 25.75$, 50.23$ stb.

Folytonos eloszlásokban **a valószínűségi sűrűségfüggvény** (PDF) megmutatja, hogyan oszlanak el a valószínűségek a lehetséges értékek között. A PDF **görbe alatti területe** annak a valószínűségét jelöli, hogy a véletlen változó **egy bizonyos tartományba esik**. 

Folytonos valószínűségi eloszlások, amelyeket gyakran használnak az adattudományban:

## **Normális eloszlás**
Az egyik leggyakoribb adateloszlás. A **legtöbb érték az átlag körül** helyezkedik el, kevesebb érték van távolabb tőle, szimmetrikus alakzatot alkotva. Tökéletesen alkalmas olyan dolgok modellezésére, mint **az emberek magassága.**

- Az átlag a görbe középpontja
- Szimmetrikus eloszlás (a bal és jobb oldal tükrözi egymást)
- A szórás megmutatja, hogy mennyire szóródnak az adatok
-  A kisebb szórás azt jelenti, hogy az adatok közelebb vannak az átlaghoz


$$f(x) = \frac{1}{\sigma \sqrt{2\pi}} \exp\left(-\frac{(x-\mu)^2}{2\sigma^2}\right)$$

> μ = átlag
> σ = szórás


## **Exponenciális eloszlás**
Modellezi az egymástól függetlenül és folyamatosan bekövetkező események között eltelt időt. 

Például a vásárlók egy **üzletbe érkezése között eltelt időt**. Segít megjósolni, hogy mennyi ideig kell várni a következő eseményre.

Modellek várakozási ideje az események között

- Az átlagos idő (pl. 10 perc két ügyfél között) határozza meg a rátát (λ)
- Az események egymástól függetlenül és folyamatosan történnek
- Hasznos a következő eseményig hátralévő idő előrejelzéséhez

$$f(x) = \lambda e^{-\lambda x}, \quad x \ge 0$$

> Míg az exponenciális eloszlás **a várakozási időkre összpontosít**, néha csak olyan helyzeteket kell modelleznünk, ahol minden kimenetel egyformán valószínű. Ebben az esetben az egyenletes eloszlást használjuk .

## **Egyenletes eloszlás**

Azt jelenti, hogy **egy tartományon belül minden kimenetel egyenlő valószínűséggel** születik. Például egy szabályos hatoldalú kocka dobása vagy egy 0 és 1 közötti véletlenszám kiválasztása. Ez mind diszkrét, mind folytonos esetekre vonatkozik.

- Minden kimenetelnek egyenlő a valószínűsége
- Diszkrét példa: kockadobás (1-től 6-ig)
- Folyamatos példa: véletlenszerű szám 0 és 1 között

$$f(x) = \frac{1}{b - a}, \quad a \le x \le b$$


## **Béta eloszlás**
A valós problémákban a valószínűségek gyakran változnak, ahogy többet megtudunk. A béta-eloszlás segít modellezni ezt a **bizonytalanságot**, és frissíteni a hiedelmeket új adatokkal. Például **megbecsülheti annak az esélyét, hogy egy ügyfél rákattint egy hirdetésre.**

- A valószínűségeket 0 és 1 között változtató modellek
- Az (α és β) paraméterek szabályozzák a megbízhatóságot és az alakot
- Gyakran használják Bayes-statisztikában és A/B tesztelésben

$$f(x) = \frac{x^{\alpha - 1}(1 - x)^{\beta - 1}}{B(\alpha, \beta)}$$

> α, β szabályozza az alakot
> B(α, β) a béta függvény


## **Gamma-eloszlás**
A Gamma eloszlás modellezi **több független esemény bekövetkezéséhez szükséges teljes időt**. Kiterjeszti az exponenciális eloszlást több feladat vagy esemény lefedésére. Például három, különböző időtartamú projektfeladat befejezéséhez szükséges teljes idő becslésére szolgál.

- Modellezi a több esemény teljes idejét
- Az alakparaméter (κ) szabályozza az események számát
- A skálaparaméter (θ) szabályozza az esemény időtartamát

$$f(x) = \frac{x^{k - 1} e^{-x / \theta}}{\theta^k \Gamma(k)}, \quad x \ge 0$$

##  **Khi-négyzet eloszlás**
A khi-négyzet eloszlást **hipotézisvizsgálatokban használják a kategorikus változók közötti kapcsolatok ellenőrzésére**. Például annak vizsgálatára, hogy a nem befolyásolja-e a kávé vagy tea iránti preferenciát. Segít meghatározni, hogy a megfigyelt különbségek a véletlennek köszönhetők-e.

- Kategóriák közötti függetlenség tesztelésére használják
- Működik kontingenciatáblákkal
- A szabadsági fokok a kategóriák számától függenek

$$f(x) = \frac{1}{2^{k/2}\Gamma(k/2)} x^{k/2 - 1} e^{-x/2}, \quad x > 0$$

> k = szabadsági fokok

## **Log-normális eloszlás**
Olyan adatokat modellez, amelyek idővel multiplikatívan (többszörösen) növekednek, például **részvényárakat vagy jövedelmet.** Ha az adatok logaritmusa normális eloszlású, akkor az eredeti adatok log-normális eloszlást követnek. Csak pozitív értékeket modellez.

- Multiplikatív növekedési folyamatokat modellez
- Az adat nem lehet negatív
- Általában részvényárakhoz és jövedelmekhez használják

$$f(x) = \frac{1}{x \sigma \sqrt{2\pi}} \exp\left(-\frac{(\ln x - \mu)^2}{2\sigma^2}\right), \quad x > 0$$
