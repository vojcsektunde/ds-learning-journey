# **Valószínűségi eloszlások az adattudományban**

Amikor összetett adatokkal, például vevői vásárlásokkal, részvényárfolyamokkal vagy időjárással dolgozunk, a valószínűségeloszlások segítenek megválaszolni a következő kérdéseket:

- Mi a legvalószínűbb, hogy megtörténik?
- Mik a ritka vagy szokatlan eredmények?
- Az értékek közel vannak egymáshoz, vagy szétszórtan helyezkednek el?

Ez segít nekünk jobb előrejelzéseket készíteni és megérteni a bizonytalanságot.

## **Miért fontosak a valószínűségeloszlások?**
- Elmagyarázza, hogyan viselkednek az adatok (fürtözött vagy szétszórt)
- Gépi tanulási modellek alapját képezik
- Statisztikai tesztekben használják (pl. p-érték )
- Segít azonosítani a kiugró értékeket és előrejelzéseket készíteni


## **Főbb összetevői**
Vizsgáljuk meg, hogyan írhatjuk le a valószínűségeiket három kulcsfogalom segítségével:

1. **Valószínűségi tömegfüggvény (PMF)**: **Diszkrét változók** (pl. megvásárolt termékek száma) esetén használatos. Megadja az egyes pontos értékek valószínűségét. *Például a vásárlók 25%-a pontosan 3 terméket vásárol.*

2. **Valószínűségi sűrűségfüggvény (PDF)**: **Folytonos változók** esetén használatos *(pl. elköltött összeg)*. Megmutatja, hogyan oszlanak el a valószínűségek **egy tartományon belül**, de **nem mutatja egyetlen pontos érték** esélyét, mivel az értékek végtelenek lehetnek.

3. **Kumulatív eloszlásfüggvény (CDF)**: **Mindkét típusnál használjuk**, és megmutatja annak valószínűségét, hogy egy érték kisebb vagy egyenlő egy adott számmal. *Például a CDF(3) = 0,75 azt jelenti, hogy a vásárlók 75%-a 3 vagy kevesebb terméket vásárol;* 

a CDF($50) = 0,80 azt jelenti, hogy a vásárlók 80%-a 50 dollárt vagy kevesebbet költ. 
A CDF-et az alábbi képlettel kaphatjuk meg:

> CDF: $F_X(x) = P(X \le x) = \int_{-\infty}^{x} f(t) \, dt$

> Ahol F(x) a CDF, f(t) pedig a PDF


## **Valószínűségeloszlások típusai**
A valószínűségi eloszlások a véletlen változók természete alapján két fő típusra oszthatók: **diszkrét és folytonos**.


## **Diszkrét adateloszlások**
Akkor diszkrét, ha a véletlen változó **megszámlálható**, specifikus értékeket vehet fel. 

Például, amikor egy vásárló egyetlen rendelésben megvásárolható termékek számát jósoljuk meg, a lehetséges eredmények egész számok, például 0, 1, 2, 3 stb. 

**Nem lehet 2,5 terméket vásárolni, így ez egy diszkrét véletlen változó.** Különböző eloszlásokat foglal magában. 

### **Binomiális eloszlás**

Adott számú próbálkozásból **mennyi a siker valószínűsége.**
$$P(X = k) = \binom{n}{k} p^k (1 - p)^{n-k}$$

- n = próbálkozások száma
- k = sikerek száma
- p = a siker valószínűsége


> Példa: feldobunk egy érmét tízszer, és fejeket számolunk.

> - Próbálkozások száma: 10
> - Két kimenetel próbánként: fej (siker) vagy írás (kudarc)
> - Siker valószínűsége (fej): 0,5


### **Bernoulli-eloszlás**
Olyan kísérleteket ír le, amelyekben csak egy kísérlet és két lehetséges kimenetel van: **siker vagy kudarc.** Ez a legegyszerűbb valószínűségi eloszlás.

$$P(X = x) = p^x (1 - p)^{1 - x}, \quad x \in \{0, 1\}$$

> Példa: feldobunk egy érmét egyszer, és megnézzük, hogy fejre esik-e.

> - Csak egy próba
> - Két kimenetel: fej (siker) vagy írás (kudarc)
> - Siker valószínűsége: 0,5
> - A grafikon két oszlopot mutat, amelyek a sikert (1) és a kudarcot (0) egyenlő valószínűséggel jelölik


### **Poisson-eloszlás**

Egy **adott időben vagy területen belül véletlenszerűen bekövetkező események számát** modellezi.
$$P(X = k) = \frac{\lambda^k e^{-\lambda}}{k!}$$

λ = átlagos előfordulási arány

> Példa: Óránként egy kávézóba belépő vendégek számának mérése. Segít megjósolni egy adott számú esemény valószínűségét az átlagos arány alapján.

> - Fix intervallumon belüli eseményeket számol
> - Az átlagos díjszabás (pl. 5 vásárló/óra) ismert
> - Kiszámítja a pontos darabszám valószínűségét (pl. pontosan 3 vásárló)
> - A grafikon egy átlagos arány körüli görbét mutat, amely a kevésbé valószínű számok esetén elkeskenyedik.

## **Geometriai eloszlások**
A geometriai eloszlás modellezi **az első siker eléréséhez szükséges próbálkozások számát** ismételt független próbálkozások során.

$$P(X = k) = (1 - p)^{k - 1} p$$


> Példa: hány e-mailt kell elküldenie, mielőtt egy ügyfél vásárol. Segít megjósolni a siker esélyét az egyes próbák során.

> - Számolja a próbálkozásokat az első sikerig
> - Minden próba független, fix sikervalószínűséggel
> - Hasznos olyan kérdésekhez, mint például a „Hány e-mailt kell még várni az első vásárlásig?”
> - A grafikon csökkenő görbét mutat, kevesebb próbálkozás nagyobb valószínűséggel
