# Valószínűségszámítás

## Alapok
### Mekkora a valószínűsége?

A kedvező esetek és az összes eset hányadosa, valószínűségekkel súlyozva.
> Példa: mekkora eséllyel dobunk egy szabályos dobókockával legalább ötöst?
> - Számunkra **kedvező** esetek: 5 és 6
> - Az **összes** eset: 1, 2, 3, 4, 5, 6
> - Mindegyik eset **egyforma valószínűséggel** fordul elő
> - Tehát a megoldás:$$\frac{2}{6} = \frac{1}{3}$$


### Mekkora a várható értéke?

Az elemi eseményeket kell átlagolni azok valószínűségével súlyozva.

> Például, mennyi két dobókocka által dobott értékek szorzatának a várható értéke?

> - 6∙6 = **36 féle lehetőség** van, a **szorzatok összege** 1+2+3+4+5+6=**21** -> 21*21 = **441**
> - **A várható érték tehát** $$\frac{441}{36} = 12.2$$

###  **Valószínűség**

- **Definíció:** azt mondja meg, hogy egy esemény mekkora eséllyel következik be
- **Jelölése: `P`** és itt zárójelben adjuk meg azt, hogy mire vonatkozik
- **Diszkrét példa:** jelölje `A` azt az **eseményt,** hogy a dobás eredménye legalább 5. Ekkor $$P(A)=\frac{1}{3}$$
- **Folytonos példa:** 10 perc követési idejű villamosra történő várakozás esetén a 
várakozási idő legfeljebb 4 perc: $$P(x<4) = \frac{4}{10} = \frac{2}{5}$$
> Például annak a valószínűsége, hogy pont négyest dobunk: $$p(4)=\frac{1}{6}$$

###  **Valószínűségi tartomány**

Bármely esemény valószínűsége mindig 0 és 1 között van.
$$ 0 ≤ P(E) ≤ 1 $$

###  **Valószínűségi változó**
- **Definíció:** olyan jelenség, mely véletlentől függő értéket vesz fel
-  **Jelölése:** nagy betű az ábécé végéről, pl. X, Y
- **Diszkrét példa:** kockadobás
-  **Folytonos példa:** a várakozás kezdetétől a villamos érkezéséig eltelt idő


###  **Eseménytér**
- **Definíció:** a valószínűségi változó által felvehető lehetséges értékek a 
halmaza
- **Jelölése:** a görög ábécé végi nagybetű, `pl. Ω`. A lehetséges értékeket a valószínűségi változó betűjelének kisbetűs változatával jelöljük, pl. x, y.
- **Diszkrét példa:** kockadobás esetén a 1, 2, 3, 4, 5 és 6 értékeket jelentik
- **Folytonos példa:** a villamosra való várakozás esetén, pontos menetrendet és 10 perces követési időt feltételezve egy 0 és 10 közötti tetszőleges valós számot


### **Várható érték** - Expected value

Az egyes értékek átlaga vagy valószínűségükkel súlyozott összege
-  **Jelölése:** E(X), μ (mű) vagy ҧ 

### **Szórás** -  Standard deviation

 Egy valószínűségi változó értékeinek a várható értéktől való eltérésének a mértéke
- **Jelölése:** D(X), σ (szigma) vagy s (tapasztalati szórás)
- **Variancia - szórás négyzete $$Var(X) = σ^2$$**

## **Kísérleti valószínűség**

- **A valószínűség kiszámolása:** elméleti valószínűség
- **Kísérleti valószínűség:** sokszor végrehajtunk kísérleteket, és a pozitív események  számát elosztjuk az összes esemény számával
> Például egy dobókockával dobunk százszor, majd megszámoljuk a 4-es eredményeket; az eredmény és a dobások számának a hányadosa adja a kísérleti valószínűség eredményét

 ## Eloszlásfüggvény - F(x)
Megadja, hogy mekkora eséllyel lesz az eredmény kisebb a paraméterül átadott értéknél
> Példa: a villamosra várakozás esetén ha x 0 és 10 perc közötti, akkor F(x)=x/10. Tehát annak az  esélye, hogy legfeljebb 4 percet kell várni, 40%, tehát F(4)=0,4
 
 
 ## Sűrűségfüggvény - f(x)

Azt fejezi ki, hogy milyen értékek mentén mekkora a valószínűségi változó értéke. Folytonos esetben van jelentősége
> **Példa:** a villamosra várakozás esetén ha x 0 és 10 perc közötti, akkor f(x)=1/10, egyébként 0. 

>**Értelmezése:** folytonos esetben nem konkrét értékekben érdemes gondolkodnunk, hanem  intervallumokban, például 2 és 3 perc közötti a várakozási idő. Az, hogy az f(x) konstans, azaz nem  függ x-től, azt jelenti, hogy a valószínűség egyenletes, azaz ugyanakkora az esélye annak, hogy a  várakozási idő 2 és 3 perc közötti mint 7 és 8 perc közötti. A függvény úgy van megalkotva, hogy az  integrálja mínusz végtelentől plusz végtelenig 1 legyen

**Kapcsolat a kettő között**
- Az eloszlásfüggvény a sűrűségfüggvény határozott integrálja` -∞-től x-ig`
- A sűrűségfüggvény az eloszlásfüggvény deriváltja

## **Feltételes valószínűség**

 Egy esemény bekövetkezésének a valószínűsége, **feltéve, hogy egy másik esemény bekövetkezett**

 $$𝑃(𝐴|𝐵) = \frac{𝑃(𝐴∩𝐵)}{𝑃(𝐵)}$$

 - Mekkora eséllyel dobunk egy szabályos dobókockával páros számot, feltéve, hogy nem hatost dobtunk?
- **P(A)**: a páros dobás esélye, 1/2
- **P(B)**: a nem hatos dobás esélye: 5/6
- **P(A∩B)**: nem hatos páros dobás esélye *(2 lehetőség a 6-ból)*: 2/6 = 1/3
- Az eredeti feladat megoldása: $$\frac{\frac{1}{3}}{\frac{5}{6}} = \frac{2}{5} = 0.4 $$

> - A **nem 6-os lehetőségek:** 1, 2, 3, 4, 5; 
> - A **nem 6-os páros lehetőségek:** 2, 4, tehát 5 esetéből **2 a kedvező**, azaz 2/5  az eredmény

---
## A Bayes-formula
 $$𝑃(𝐴|𝐵) = \frac{𝑃(𝐵|𝐴) * 𝑃(𝐴)}{𝑃(𝐵)}$$

 $$𝑃(𝐵|𝐴) = \frac{𝑃(𝐴|𝐵) * 𝑃(𝐵)}{𝑃(𝐴)}$$


>Az előző példában: mekkora eséllyel dobtunk nem hatost, feltéve, hogy párost dobtunk?

 $$𝑃(𝐵|𝐴) = \frac{2/5 * 5/6}{1/2} = \frac{2}{3} $$

- **ami a következőképpen ellenőrizhető:** páros dobás esetén a lehetséges értékek: 2, 4, 6, azaz kétharmad eséllyel lesz nem hatos

- **Használata**: ha az egyik irányú feltételes valószínűség kiszámítása sokkal könnyebb mint a másik, de nekünk a másikra van szükség

## Több esemény együttes valószínűsége

> Mennyi két esemény bekövetkezésének az együttes valószínűsége?

Nem mindegy, hogy függetlenek vagy nem - **Szorzás törvénye** *(multiplication law)*

**Ha függetlenek:**  $$P(A, B) = P(A ∩ B) = P(A) * P(B)$$

-  **Példa**: mekkora eséllyel dobunk kétszer hatost egymás után egy dobókockával?
- A két esemény független, tehát össze kell szorozni az egyes valószínűségeket külön-külön
 $$P(A) * P(B) = 1/6 * 1/6 = 1/36$$

**Ha nem függetlenek:** $$P(A, B) = P(A ∩ B) = P(A) * P(B|A)$$

- **Példa:** egy pakli francia kártyából kihúzunk 2 lapot; mekkora eséllyel lesz mindkettő ász?
- A két esemény nem független, mert ha az első húzás eredménye ász, akkor a maradék 51 lapban 3 ász van, egyébként 4
 $$P(A) * P(B|A) = 4/52 * 3/51 = 1/221$$
