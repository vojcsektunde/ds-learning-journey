# Valószínűségszámítás

## Alapok
### Mekkora a valószínűsége?

A kedvező esetek és az összes eset hányadosa, valószínűségekkel súlyozva.
> Példa: mekkora eséllyel dobunk egy szabályos dobókockával legalább ötöst?
> - Számunkra **kedvező** esetek: 5 és 6
> - Az **összes** eset: 1, 2, 3, 4, 5, 6
> - Mindegyik eset **egyforma valószínűséggel** fordul elő
> - Tehát a megoldás:$$\frac{2}{6} = \frac{1}{3}$$


### Mekkora a várható értéke?

Az elemi eseményeket kell átlagolni azok valószínűségével súlyozva.

> Például, mennyi két dobókocka által dobott értékek szorzatának a várható értéke?

> - 6∙6 = **36 féle lehetőség** van, a **szorzatok összege** 1+2+3+4+5+6=**21** -> 21*21 = **441**
> - **A várható érték tehát** $$\frac{441}{36} = 12.2$$ 

###  **Valószínűség**

- **Definíció:** azt mondja meg, hogy egy esemény mekkora eséllyel következik be
- **Jelölése: `P`** és itt zárójelben adjuk meg azt, hogy mire vonatkozik
- **Diszkrét példa:** jelölje `A` azt az **eseményt,** hogy a dobás eredménye legalább 5. Ekkor $$P(A)=\frac{1}{3}$$
- **Folytonos példa:** 10 perc követési idejű villamosra történő várakozás esetén a 
várakozási idő legfeljebb 4 perc: $$P(x<4) = \frac{4}{10} = \frac{2}{5}$$
> Például annak a valószínűsége, hogy pont négyest dobunk: $$p(4)=\frac{1}{6}$$

###  **Valószínűségi tartomány**

Bármely esemény valószínűsége mindig 0 és 1 között van.
$$ 0 ≤ P(E) ≤ 1 $$

###  **Valószínűségi változó**
- **Definíció:** olyan jelenség, mely véletlentől függő értéket vesz fel
-  **Jelölése:** nagy betű az ábécé végéről, pl. X, Y
- **Diszkrét példa:** kockadobás
-  **Folytonos példa:** a várakozás kezdetétől a villamos érkezéséig eltelt idő


###  **Eseménytér**
- **Definíció:** a valószínűségi változó által felvehető lehetséges értékek a 
halmaza
- **Jelölése:** a görög ábécé végi nagybetű, `pl. Ω`. A lehetséges értékeket a valószínűségi változó betűjelének kisbetűs változatával jelöljük, pl. x, y.
- **Diszkrét példa:** kockadobás esetén a 1, 2, 3, 4, 5 és 6 értékeket jelentik
- **Folytonos példa:** a villamosra való várakozás esetén, pontos menetrendet és 10 perces követési időt feltételezve egy 0 és 10 közötti tetszőleges valós számot



## **Kísérleti valószínűség**

- **A valószínűség kiszámolása:** elméleti valószínűség
- **Kísérleti valószínűség:** sokszor végrehajtunk kísérleteket, és a pozitív események  számát elosztjuk az összes esemény számával
> Például egy dobókockával dobunk százszor, majd megszámoljuk a 4-es eredményeket; az eredmény és a dobások számának a hányadosa adja a kísérleti valószínűség eredményét
