## Vektorok

A vektorokat az adatok numerikus formában történő ábrázolására használják, hogy a gépi tanulási algoritmusok hatékonyan feldolgozhassák, elemezhessék és tanulhassanak belőlük. Segítenek a valós információkat strukturált formátumba konvertálni, amelyet a gépek megértenek.

A vektor egy **matematikai objektum, amelynek nagysága és iránya van.**

- Ez egy rendezett számlista, ahol minden szám egy jellemzőt jelöl
Adatpontok összetevőinek vagy elemeinek is nevezik
Széles körben használják osztályozásban, regresszióban, klaszterezésben és mélytanulásban
Számos gépi tanulási algoritmus alapját képezi

> Példa: Egy 2D vektor felírható így
> v = ( x1 , x2 )


```python
import numpy as np

vector = np.array([170, 65])

print("Vector:", vector)
```

## Vektorok a gépi tanulási modellekben
A vektorokat a gépi tanulási modellek különböző pontjain használják:

- **Bemenet** : Minden adatot, például képeket, szöveget vagy érzékelők által leolvasott adatokat numerikus formába kell alakítani. **Ezeket a számokat vektorokként rendezzük**, hogy a modellek hatékonyan feldolgozhassák őket.

- **Modell** : A gépi tanulási modellek, különösen a neurális hálózatok, vektor- és mátrixműveletekre támaszkodnak. A **lineáris algebrai műveletek**, mint például a szorzás, segítenek frissíteni a **modell paramétereit és javítani a pontosságot**.

- **Kimenetek**: A modell **kimenetei lehetnek számok, kategóriák vagy vektorok.** Az olyan feladatokban, mint az NLP vagy az ajánlások, a kimeneti vektorokat hasonlóságra, klaszterezésre vagy további előrejelzésekre használják.


## Vektortípusok

- Sor- és oszlopvektorok
    - A sorvektor egy egydimenziós tömb, amelyet sorformátumban ábrázolunk: (x1, x2, x3)
    - Az oszlopvektor egy egydimenziós tömb, amelyet oszlopjelöléssel ábrázolunk:
    $$\mathbf{x} = \begin{bmatrix} x_1 \\ x_2 \\ x_3 \\ \vdots \\ x_n \end{bmatrix}$$

- Nulla vektor
    - Egy vektor, amelynek minden eleme nulla. 
    - Hasznosak optimalizálási problémák megoldásánál, és a vektortérben az origót jelentik.

    $$v = (0, 0, 0)$$

- Egységvektor
    - Egy 1 nagyságrendű vektor. Gyakran használják irány jelölésére:
    $$\mathbf{u} = \frac{\mathbf{v}}{\Vert{}\mathbf{v}\Vert{}}$$
    ahol ∥v∥ a v vektor nagysága.

- Ritka és sűrű vektorok
    - A ritka vektorok elsősorban nullákból állnak, és szövegelemző, valamint ajánlórendszerekben használják őket.
    - A sűrű vektorok elsősorban nem nulla értékekből állnak, és képfeldolgozásban, valamint mélytanulásban használják őket.


## A vektorok fontossága a gépi tanulásban
### 1. Jellemzők ábrázolása
A vektorokat az adatpontok numerikus formában történő ábrázolására használják. Például a természetes nyelvi feldolgozásban (NLP) a szavakat szóvektorokká alakítják olyan technikákkal, mint a Word2Vec vagy a TF-IDF.

### 2. Távolság- és hasonlósági mértékek
Az adatpontok közötti hasonlóságot a gépi tanulásban jellemzően vektor távolságmértékekkel számítják ki, például:

- **Euklideszi távolság:** Két pont közötti egyenes vonalú távolság mennyiségi mértéke.
- **Koszinusz hasonlóság:** Két vektor közötti szög koszinuszát számszerűsíti a hasonlóság megállapítása érdekében.


### 3. Átalakítások és vetítések
A vektorok olyan matematikai transzformációkat tesznek lehetővé, mint az elforgatás, a skálázás és az eltolás. Ezeket olyan módszerekben alkalmazzák, mint a főkomponens-analízis (PCA), hogy az adathalmazokat alacsonyabb dimenziókba vetítsék ki.

## Vektorműveletek

Vektoros összeadás és kivonás

```python
import numpy as np

a = np.array([2, 3])
b = np.array([1, 4])

add = a + b

sub = a - b

print("Addition:", add)
print("Subtraction:", sub)
```

Skaláris szorzás

```python
import numpy as np

a = np.array([1, 2, 3])

scalar = 3

res = scalar * a

print("Scalar Multiplication:", res)
```

Pontszorzat

```python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

prod = np.dot(a, b)

print("Dot Product:", prod)
```

Keresztszorzat

```python
import numpy as np

c = np.array([1, 2, 3])
d = np.array([4, 5, 6])

prod = np.cross(c, d)

print("Cross Product:", prod)
```
